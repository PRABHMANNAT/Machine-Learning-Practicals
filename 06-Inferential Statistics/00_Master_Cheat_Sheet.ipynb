## 1. Test-selection map

| Outcome and design | Typical question | Test |
|---|---|---|
| quantitative, one sample, known $\sigma$ | mean versus $\mu_0$ | one-sample z |
| quantitative, one sample, unknown $\sigma$ | mean versus $\mu_0$ | one-sample t |
| quantitative, same units twice | mean change | paired t |
| quantitative, two independent groups | mean difference | Welch t |
| quantitative, 3+ independent groups | any mean differs | one-way ANOVA / Welch ANOVA |
| quantitative, same units in 3+ conditions | any condition mean differs | repeated-measures ANOVA / mixed model |
| quantitative, 2+ categorical factors | main effects/interactions | factorial ANOVA / linear model |
| one categorical variable | follows stated proportions | chi-square goodness of fit |
| two categorical variables | associated | chi-square independence |
| update probability after evidence | posterior probability | Bayes’ theorem |


## 2. Universal hypothesis-test workflow

1. Define population, parameter, and outcome.
2. Write $H_0$ and $H_a$.
3. Set $\alpha$ before seeing the answer.
4. Identify design and tail.
5. Check assumptions.
6. Calculate test statistic and p-value.
7. Reject or fail to reject $H_0$.
8. State the contextual conclusion.
9. Report estimate, CI, and effect size.

**Decision sentence:** Since $p$ [is/is not] below $\alpha$, we [reject/fail to reject] $H_0$. The data provide [sufficient/insufficient] evidence that …


## 3. Core formulas

| Topic | Formula |
|---|---|
| Mean z | $z=(\bar x-\mu_0)/(\sigma/\sqrt n)$ |
| One-sample t | $t=(\bar x-\mu_0)/(s/\sqrt n)$, df $=n-1$ |
| Mean CI, known $\sigma$ | $\bar x\pm z^*\sigma/\sqrt n$ |
| Mean CI, unknown $\sigma$ | $\bar x\pm t^*s/\sqrt n$ |
| Bayes | $P(A\mid B)=P(B\mid A)P(A)/P(B)$ |
| Chi-square | $\chi^2=\sum(O-E)^2/E$ |
| GOF df | $k-1$ minus parameters fitted |
| Independence df | $(r-1)(c-1)$ |
| ANOVA | $SS_T=SS_B+SS_W$ |
| ANOVA F | $F=MS_B/MS_W$ |
| Power | $1-\beta$ |


## 4. Error language

| Reality / decision | Result |
|---|---|
| $H_0$ true, reject | Type I error, probability $\alpha$ |
| $H_0$ false, fail to reject | Type II error, probability $\beta$ |
| $H_0$ false, reject | power / correct detection |
| $H_0$ true, fail to reject | correct non-rejection |

Never say:

- “p is the probability $H_0$ is true.”
- “The null is accepted.”
- “Significant means important.”
- “Non-significant means no effect.”
- “ANOVA proves every group differs.”
- “Association proves causation.”


## 5. Assumption snapshot

| Family | Main checks |
|---|---|
| z/t means | independence, credible sampling/design, normal/large-sample justification, outliers |
| paired t | normality/outliers of differences, correct pairing |
| classical ANOVA | independent observations, normal residuals, equal variances, influential cases |
| repeated measures | dependency model plus sphericity or correction |
| chi-square | independent counts, exclusive categories, adequate expected counts |
| Bayes | credible prior and likelihood; conditional structure is correctly modelled |

## 7. Reporting templates

### Mean test

> A [one/two]-sided [z/t] test found [evidence/no sufficient evidence] that [parameter statement], [statistic and df], p = [value], [effect estimate and CI].

### Chi-square

> The categorical distribution [did/did not] differ from expectation, $\chi^2(df,N=[N])=[value]$, p = [value], [effect size].

### ANOVA

> Mean [outcome] differed across [factor] levels, $F(df_B,df_W)=[value]$, p = [value], $\eta^2=[value]$. [Adjusted post-hoc result if applicable].

Use exact p-values when practical; report p < .001 rather than p = .000.


# Revision questions and answers

**Q1. Unknown sigma, one sample mean—what test?**

<details><summary>Answer</summary>

One-sample t-test.

</details>

---

**Q2. What is the p-value conditioned on?**

<details><summary>Answer</summary>

$H_0$ and all model assumptions being true.

</details>

---

**Q3. Two-sided alpha = .05 critical z?**

<details><summary>Answer</summary>

Approximately ±1.96.

</details>

---

**Q4. Type I error?**

<details><summary>Answer</summary>

Rejecting a true null hypothesis.

</details>

---

**Q5. Power?**

<details><summary>Answer</summary>

$1-\beta$.

</details>

---

**Q6. Expected GOF count?**

<details><summary>Answer</summary>

$E_i=np_i$.

</details>

---

**Q7. ANOVA alternative?**

<details><summary>Answer</summary>

At least one population mean differs.

</details>

---

**Q8. ANOVA identity?**

<details><summary>Answer</summary>

$SS_T=SS_B+SS_W$.

</details>

---

**Q9. Bayes in words?**

<details><summary>Answer</summary>

Posterior is proportional to likelihood times prior.

</details>

---

**Q10. What accompanies a p-value?**

<details><summary>Answer</summary>

Effect estimate, confidence interval, assumptions/design context, and usually an effect size.

</details>
