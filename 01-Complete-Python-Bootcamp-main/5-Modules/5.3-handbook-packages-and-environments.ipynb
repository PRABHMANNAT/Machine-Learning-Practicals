## Project 1 — Greetings module (completed safely in memory)

**Problem:** create a reusable greeting module and import it. **Approach:** write module source in a temporary project, load it with `importlib`, call it, and inspect `__name__`.

````python
import importlib.util, tempfile
from pathlib import Path
with tempfile.TemporaryDirectory() as folder:
    module_path=Path(folder)/'greetings.py'
    module_path.write_text('def greet(name):\n    return f"Hello, {name}!"\n',encoding='utf-8')
    spec=importlib.util.spec_from_file_location('greetings',module_path)
    greetings=importlib.util.module_from_spec(spec); spec.loader.exec_module(greetings)
    print(greetings.greet('Maya'))
    print(greetings.__name__)
````

## Built-in vs standard vs third-party vs custom

| Kind | Example | Installation |
|---|---|---|
| Built-in | `len` | Always available |
| Standard library | `math`, `pathlib` | Ships with Python |
| Third-party | NumPy, Requests | Install with pip/manager |
| Custom | `package.prabhmaths` | Your project code |

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Create venv | `python -m venv .venv` |
| Install | `python -m pip install name` |
| Snapshot | `python -m pip freeze > requirements.txt` |
| Install snapshot | `python -m pip install -r requirements.txt` |
| Entry guard | `if __name__ == "__main__": main()` |

## Common mistakes and interview tips

- Do not commit virtual environments or secrets.
- Do not use `pip` from an unknown interpreter.
- Avoid import-time application startup.
- A requirements snapshot is not a substitute for explaining direct dependencies.

**Revision habit:** explain what each line does, predict the result, run it, and test one edge case.
