# Understanding `__init__.py` — The Package Welcome Desk 🛎️

## Goal

`__init__.py` is code that runs when its package is first imported. It can document the package, expose a friendly public API, set metadata, or initialize lightweight package state.

## 1. Existing package structure

```text
5-Modules/
├── package/
│   ├── __init__.py
│   ├── prabhmaths.py
│   └── subpackages/
│       ├── __init__.py
│       └── mult.py
└── test.py
```

The current `package/__init__.py` contains a package docstring, so functions are imported from their exact modules. A future re-export could intentionally enable shorter calls such as `package.addition(...)`.

````python
import package
from package.prabhmaths import addition, substraction
from package.subpackages.mult import multiply
print(package.__file__)
print(addition(2,3))
print(substraction(4,3))
print(multiply(4,5))
````

## 2. What happens during import

1. Python finds the package.
2. It creates a module object and places it in `sys.modules`.
3. It executes `package/__init__.py` once.
4. Names assigned/imported there become `package.name`.

Keep initialization fast and unsurprising: avoid network calls, large computations, or destructive side effects.

````python
import sys
first_id=id(package)
import package as package_again
print(first_id == id(package_again))
print('package' in sys.modules)
````

## 3. Friendly public API and `__all__`

An `__init__.py` can contain:

```python
from .prabhmaths import addition, substraction
from .subpackages.mult import multiply
__all__ = ["addition", "substraction", "multiply"]
__version__ = "1.0.0"
```

`__all__` mainly controls wildcard imports and communicates intended public names; explicit imports can still reach other attributes.

## 4. Relative and absolute imports inside packages

`from .prabhmaths import addition` is relative to the current package. `from package.prabhmaths import addition` is absolute. Relative imports are concise internally; absolute imports reveal the full package path. Never run a relative-import module as a loose file—use `python -m package.module`.

## 5. Is `__init__.py` always required?

Modern Python supports namespace packages without it, but regular teaching/application packages commonly include it for clarity, initialization, metadata, and stable APIs. It is different from a class method named `__init__`: the file initializes a package; the method initializes an object instance.

## Key concepts summary

`__init__.py` is the package welcome desk: it runs once on import and can expose a clean API.

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Relative re-export | `from .module import name` |
| Public names | `__all__ = ["name"]` |
| Metadata | `__version__ = "1.0.0"` |
| Package run | `python -m package.module` |

## Common mistakes and interview tips

- Do not confuse the file with a class `__init__` method.
- Avoid heavy side effects during import.
- Re-export deliberately; do not expose every internal helper.
- Circular imports often arise from too much package-level importing.

**Revision habit:** explain what each line does, predict the result, run it, and test one edge case.
