# Control Flow 2 — Loops: Teach Python to Repeat 🔁

## Goal

A loop repeats instructions without copying the same code many times. Python has two main loop statements:

- `for`: visit each item in an iterable;
- `while`: repeat while a condition remains true.

You will also learn ranges, strings and containers, dictionaries, `enumerate`, `zip`, `reversed`, `break`, `continue`, `pass`, loop `else`, nested loops, comprehensions, sentinel loops, and common traps.

## 1. Loop vocabulary

- **Iterable:** something you can visit item by item, such as a string, list, dictionary, or `range`.
- **Iteration:** one trip around the loop.
- **Loop variable:** the temporary name for the current item.
- **Body:** the indented instructions repeated each time.

## 2. `range()` makes a counting sequence

`range(start, stop, step)` stops **before** `stop`.

- `range(5)` means `0, 1, 2, 3, 4`.
- `range(1, 6)` means `1, 2, 3, 4, 5`.
- A negative step counts backward.

`range` is lazy: it remembers how to generate numbers instead of storing a huge list.

```python
print(range(5))
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(1, 10, 3)))
print(list(range(10, 1, -1)))
print(list(range(10, 1, -2)))
```

Why is `1` missing from `range(10, 1, -1)`? Because the stop value is never included. To include `1`, stop at `0`: `range(10, 0, -1)`.

## 3. `for` loops visit each item

The original counting examples are preserved here.

```python
for i in range(5):
    print(i)

print("Count 1 to 5:")
for i in range(1, 6):
    print(i)

print("Jump by 3:")
for i in range(1, 10, 3):
    print(i)
```

## 4. Loop over a string or list

Use a descriptive loop name. Avoid naming a variable `str`, because `str` is Python's text-conversion tool.

```python
student_name = "Prabhmannat Singh"
for character in student_name:
    print(character)

fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(f"I like {fruit}.")
```

## 5. Loop through dictionaries

Looping over a dictionary alone gives keys. Use `.values()` for values and `.items()` for key-value pairs.

```python
student = {"name": "Maya", "score": 9, "city": "Pune"}

for key in student:
    print("key:", key)

for key, value in student.items():
    print(key, "->", value)
```

## 6. `enumerate()` gives position and item

This is clearer than manually updating an index. Choose `start=1` for human-friendly numbering.

```python
colors = ["red", "green", "blue"]
for position, color in enumerate(colors, start=1):
    print(position, color)
```

## 7. `zip()` walks through iterables together

`zip` stops when the shortest iterable ends. Use it when matching related positions.

```python
names = ["Maya", "Leo", "Zara"]
scores = [9, 7, 10]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

## 8. `reversed()` and sorted iteration

`reversed()` visits an ordered sequence backward. `sorted()` creates a sorted list for iteration without changing the original.

```python
numbers = [3, 1, 4, 2]
print("Backward:")
for number in reversed(numbers):
    print(number, end=" ")

print("\nSorted:")
for number in sorted(numbers):
    print(number, end=" ")
print()
```

## 9. `while` repeats while a question is true

A `while` loop needs:

1. a starting value;
2. a condition;
3. an update that eventually makes the condition false.

Without the update, the loop may continue forever.

```python
count = 0

while count < 5:
    print(count)
    count = count + 1
```

### Sentinel loop

A sentinel is a special stop value. Interactive input is commented out; the sample list simulates a person entering words and then `quit`.

```python
simulated_inputs = iter(["hello", "python", "quit"])

while True:
    # word = input("Type a word, or quit: ")
    word = next(simulated_inputs)
    if word == "quit":
        break
    print("You typed:", word)
```

## 10. `break`, `continue`, and `pass`

- `break` exits the nearest loop immediately.
- `continue` skips the rest of this iteration and starts the next one.
- `pass` does nothing; it is a placeholder for an empty block.

```python
print("break example:")
for i in range(10):
    if i == 5:
        break
    print(i)

print("continue example (odd numbers):")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)

print("pass example (all values still print):")
for i in range(5):
    if i == 3:
        pass
    print(i)
```

`pass` does **not** skip `3`; it literally performs no action. If you want to skip printing `3`, use `continue`.

## 11. Loop `else`: “we did not break”

The `else` block runs when a loop finishes normally. It does not run after `break`. This is useful for searching.

```python
items = [4, 7, 9, 12]
target = 9

for item in items:
    if item == target:
        print("Found it!")
        break
else:
    print("Not found")
```

## 12. Nested loops: a loop inside a loop

For each outer iteration, the entire inner loop runs. Three outer trips × two inner trips = six printed pairs.

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

### Build a tiny grid

```python
for row in range(3):
    for column in range(4):
        print("⭐", end=" ")
    print()
```

## 13. Accumulator pattern: build a result little by little

Do not call a variable `sum`, because that hides Python's built-in `sum()` function. Use `total`.

```python
# Original while-loop sum example
n = 10
total = 0
count = 1

while count <= n:
    total += count
    count += 1

print("Sum of first 10 natural numbers:", total)

# Original for-loop idea
total = 0
for number in range(11):
    total += number
print(total)

# Python's built-in shortcut
print(sum(range(1, 11)))
```

## 14. Prime numbers and the important `for`/`else`

If a divisor is found, `break` skips the loop's `else`. If no divisor is found, the `else` prints the prime.

```python
primes = []
for number in range(2, 101):
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            break
    else:
        primes.append(number)

print(primes)
```

## 15. Comprehensions: compact loops that build containers

Use comprehensions when the idea stays short and readable. A normal loop is better for several steps or side effects.

```python
squares = [number ** 2 for number in range(1, 6)]
even_squares = [number ** 2 for number in range(1, 11) if number % 2 == 0]
square_map = {number: number ** 2 for number in range(1, 6)}
unique_lengths = {len(word) for word in ["cat", "lion", "tiger"]}

print(squares)
print(even_squares)
print(square_map)
print(unique_lengths)
```

## 16. Common loop mistakes

- Off-by-one errors because `range` excludes `stop`.
- An infinite `while` loop because its condition never changes.
- Changing a list while iterating over it.
- Hiding built-ins with names such as `str`, `sum`, or `list`.
- Using `range(len(items))` when direct iteration or `enumerate()` is clearer.
- Expecting `break` to exit every nested loop—it exits only the nearest one.
- Forgetting that sets do not promise a meaningful display order.

Iterate over a copy when removing items.

```python
numbers = [1, 2, 3, 4, 5, 6]
for number in numbers.copy():
    if number % 2 == 0:
        numbers.remove(number)
print(numbers)
```

## Checks — predict, run, change

1. Count from `5` down to `1`.
2. Print only words longer than three letters.
3. Use `while` to double a number until it reaches at least `100`.
4. Find the first number divisible by both `7` and `5`.

```python
for number in range(5, 0, -1):
    print(number)

for word in ["cat", "lion", "elephant"]:
    if len(word) > 3:
        print(word)

value = 3
while value < 100:
    value *= 2
print(value)

for number in range(1, 101):
    if number % 7 == 0 and number % 5 == 0:
        print("First shared multiple:", number)
        break
```

## Easy revision cheat sheet

| Need | Best tool | Pattern |
|---|---|---|
| Visit every item | `for` | `for item in items:` |
| Repeat while true | `while` | `while condition:` |
| Count | `range` | `range(start, stop, step)` |
| Get number + item | `enumerate` | `for i, item in enumerate(items):` |
| Pair sequences | `zip` | `for a, b in zip(xs, ys):` |
| Exit now | `break` | Leaves nearest loop |
| Skip this trip | `continue` | Starts next iteration |
| Empty placeholder | `pass` | Does nothing |
| Detect no `break` | loop `else` | Search/prime patterns |
| Build a simple list | comprehension | `[f(x) for x in xs if test(x)]` |

**Five-second memory line:** `for` visits items; `while` watches a condition; `range` excludes its stop; always make a `while` loop able to finish.
