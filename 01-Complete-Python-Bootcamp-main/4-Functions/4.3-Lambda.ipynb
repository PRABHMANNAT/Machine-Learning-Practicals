# Lambda Functions — Tiny One-Line Recipes ⚡

## Goal

`lambda` creates a small anonymous function with any number of parameters but exactly **one expression**. Use it when a short function is needed briefly—especially as `key=` or as an argument to `map`/`filter`.

## 1. Syntax and named-function comparison

`lambda arguments: expression`

The expression is returned automatically. There is no `return` keyword and no statement block.

```python
def add_named(a, b):
    return a + b

add_lambda = lambda a, b: a + b
print(add_named(2, 3))
print(type(add_lambda).__name__)
print(add_lambda(5, 6))
```

## 2. Original parity and three-number examples

The named version should explicitly return `False` for odd numbers; the Boolean expression already produces the correct answer.

```python
def even(number):
    return number % 2 == 0

even_lambda = lambda number: number % 2 == 0
add_three = lambda x, y, z: x + y + z
print(even(24), even_lambda(12))
print(add_three(12, 13, 14))
```

## 3. Conditional expression inside lambda

A lambda may contain an expression such as `yes if condition else no`, but not statement-style `if`, loops, `try`, or assignments.

```python
label = lambda number: "even" if number % 2 == 0 else "odd"
print(label(7), label(8))
```

## 4. Lambda with `map`

`map` feeds every item to the lambda. Convert the lazy result to a list when you need all values now.

```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(map(lambda x: x ** 2, numbers))
print(result)
```

## 5. Lambda with `filter`

The lambda is a yes/no gate. Only values producing truthy results survive.

```python
numbers = list(range(1, 11))
print(list(filter(lambda number: number % 2 == 0, numbers)))
```

## 6. Best use: sorting with `key=`

The lambda says which feature to compare. Here, sort by descending score and ascending name.

```python
students = [("Zara", 90), ("Maya", 95), ("Leo", 90)]
ranked = sorted(students, key=lambda row: (-row[1], row[0]))
print(ranked)
```

## 7. Closures and captured values

A lambda remembers names from its enclosing scope. This can create a tiny configurable function.

```python
def make_multiplier(factor):
    return lambda number: number * factor

triple = make_multiplier(3)
print(triple(7))
```

## 8. Late-binding loop trap

Closures look up captured names when called. Lambdas made in a loop may all see the final loop value. Capture the current value with a default argument.

```python
wrong = [lambda: i for i in range(3)]
fixed = [lambda i=i: i for i in range(3)]
print([function() for function in wrong])
print([function() for function in fixed])
```

## 9. When not to use lambda

Prefer `def` when logic needs a name, docstring, annotations, several steps, error handling, or reuse. Avoid assigning complicated lambdas to names; a named `def` is clearer.

## Revision cheat sheet

| Need | Pattern |
|---|---|
| One expression | `lambda x: expression` |
| Several inputs | `lambda x, y: x + y` |
| Tiny choice | `lambda x: a if test else b` |
| Sort feature | `sorted(data, key=lambda row: row[1])` |
| Capture loop value | `lambda x=x: x` |

**Memory line:** lambda is a short disposable expression-function; clarity decides whether it belongs.
