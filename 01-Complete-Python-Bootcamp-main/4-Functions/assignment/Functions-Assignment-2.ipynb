### Q1. Create a reusable bill calculator

**Concept tested:** parameters, defaults, return values

**Question:** Write `calculate_bill(prices, tax_rate=0, discount=0)` and return the final amount.

**Hint:** Compute subtotal, subtract percentage discount, then add tax to the discounted amount.

**Step-by-step logic:** Compute subtotal, subtract percentage discount, then add tax to the discounted amount. Then return structured data where possible so the result is easy to test and reuse.

**What was learned:** Parameters make one calculation work for many bills.

```python
def calculate_bill(prices, tax_rate=0, discount=0):
    subtotal = sum(prices)
    after_discount = subtotal * (1 - discount / 100)
    return after_discount * (1 + tax_rate / 100)

print(f"{calculate_bill([100, 50], tax_rate=10, discount=20):.2f}")
```

### Q15. Build a retry decorator

**Concept tested:** decorators, nested functions, `*args`, `**kwargs`

**Question:** Create `retry(times)` that retries a function after exceptions and returns the successful result.

**Hint:** A decorator factory remembers the limit; wrapper forwards all arguments and stores the last error.

**Step-by-step logic:** A decorator factory remembers the limit; wrapper forwards all arguments and stores the last error. Then return structured data where possible so the result is easy to test and reuse.

**Expected output:** The executed output is saved directly beneath the solution cell.

**What was learned:** Decorators wrap behavior while preserving a reusable call interface.

```python
def retry(times):
    def decorate(function):
        def wrapper(*args, **kwargs):
            last_error = None
            for _ in range(times):
                try:
                    return function(*args, **kwargs)
                except ValueError as error:
                    last_error = error
            raise last_error
        return wrapper
    return decorate

attempts = {"count": 0}
@retry(3)
def unstable():
    attempts["count"] += 1
    if attempts["count"] < 3: raise ValueError("try again")
    return "success"
print(unstable(), attempts["count"])
```

### Q16. Write a generic memoization decorator

**Concept tested:** closures, dictionaries, argument unpacking

**Question:** Cache function results by positional and sorted keyword arguments.

**Hint:** Build a hashable key, check the closure's cache, and forward arguments only on a miss.

**Step-by-step logic:** Build a hashable key, check the closure's cache, and forward arguments only on a miss. Then return structured data where possible so the result is easy to test and reuse.

**Expected output:** The executed output is saved directly beneath the solution cell.

**What was learned:** Tuples turn compatible arguments into compound cache keys.

```python
def memoize(function):
    cache = {}
    def wrapper(*args, **kwargs):
        key = (args, tuple(sorted(kwargs.items())))
        if key not in cache:
            cache[key] = function(*args, **kwargs)
        return cache[key]
    return wrapper

@memoize
def power(base, exponent=2): return base ** exponent
print(power(3, exponent=4), power(3, exponent=4))
```

### Q26. Implement a configurable group-by

### Q28. Create a transaction processor with rollback

**Concept tested:** nested functions, nonlocal state, practical design

**Question:** Apply operations to a balance; if any would make it negative, return the original balance and failure.

**Hint:** Keep current balance in enclosing scope, update with `nonlocal`, and stop on invalid state.

**Step-by-step logic:** Keep current balance in enclosing scope, update with `nonlocal`, and stop on invalid state. Then return structured data where possible so the result is easy to test and reuse.

**Expected output:** The executed output is saved directly beneath the solution cell.

**What was learned:** Closures can manage local state, while explicit return data reports success.

```python
def process_transactions(opening, transactions):
    balance = opening
    def apply(amount):
        nonlocal balance
        if balance + amount < 0: return False
        balance += amount
        return True
    for amount in transactions:
        if not apply(amount): return opening, False
    return balance, True

print(process_transactions(100, [-20, 50, -200]))
```

### Q29. Create a tiny expression evaluator

**Concept tested:** function dispatch, recursion, validation

**Question:** Evaluate nested tuples such as `('add', 2, ('mul', 3, 4))` safely without `eval`.

**Hint:** Map allowed operator names to functions; recursively evaluate tuple operands.

**Step-by-step logic:** Map allowed operator names to functions; recursively evaluate tuple operands. Then return structured data where possible so the result is easy to test and reuse.

**Expected output:** The executed output is saved directly beneath the solution cell.

**What was learned:** A whitelist plus recursive functions is safer than executing arbitrary text.

```python
from operator import add, sub, mul, truediv
OPS = {"add": add, "sub": sub, "mul": mul, "div": truediv}

def evaluate(expression):
    if isinstance(expression, (int, float)): return expression
    operation, left, right = expression
    if operation not in OPS: raise ValueError("unsupported operation")
    return OPS[operation](evaluate(left), evaluate(right))

print(evaluate(("add", 2, ("mul", 3, 4))))
```

### Q30. Build a complete analytics report function

**Concept tested:** composition, `*args`, `**kwargs`, map/filter, closures

**Question:** Clean readings, keep values in a configurable range, transform them, and return summary statistics.

**Hint:** Create a range predicate closure, map conversion, filter validity, map transformation, then summarize.

**Step-by-step logic:** Create a range predicate closure, map conversion, filter validity, map transformation, then summarize. Then return structured data where possible so the result is easy to test and reuse.

**Expected output:** The executed output is saved directly beneath the solution cell.

**What was learned:** Later problems combine flexible arguments, closures, lazy tools, defaults, and safe aggregation.

```python
def between(low, high):
    return lambda value: low <= value <= high

def analytics(*readings, transform=lambda x: x, **limits):
    low, high = limits.get("low", float("-inf")), limits.get("high", float("inf"))
    converted = map(float, readings)
    valid = filter(between(low, high), converted)
    values = list(map(transform, valid))
    return {"count": len(values), "values": values,
            "average": sum(values)/len(values) if values else None}

print(analytics("2", 5, 12, 8, low=1, high=10, transform=lambda x: x*x))
```

## Final revision cheat sheet

| Family | Core idea |
|---|---|
| Inputs | positional, keyword, defaults, `*args`, `**kwargs` |
| Outputs | return reusable values; use structured results |
| Scope | local → enclosing → global → built-in |
| Higher-order | accept, return, store, and compose functions |
| Lambda | one short expression, often a local callback |
| `map` | lazy transformation |
| `filter` | lazy selection using a predicate |
| Closure | returned function remembers enclosing values |
| Decorator | wraps a function with reusable behavior |
| Recursion | base case plus smaller self-call |

**Final habit:** define the function contract, validate inputs, keep one clear job, return data, and test normal, boundary, empty, and invalid cases.
