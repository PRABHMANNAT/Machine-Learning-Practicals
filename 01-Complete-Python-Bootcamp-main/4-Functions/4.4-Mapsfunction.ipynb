# `map()` — Give Every Item the Same Makeover 🪄

## Goal

`map(function, iterable, ...)` transforms items lazily. It returns a `map` iterator, not a list. Imagine a makeover machine: each incoming item receives the same recipe.

## 1. Named function with `map`

Step 1: define one-item transformation. Step 2: give the function—not `square()`—to `map`. Step 3: consume the iterator.

```python
def square(x):
    return x * x

print(square(10))
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
mapped = map(square, numbers)
print(type(mapped).__name__)
print(list(mapped))
```

## 2. Lambda with `map`

Use a lambda when the transformation is truly tiny and local.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
print(list(map(lambda x: x * x, numbers)))
```

## 3. Multiple iterables

`map` passes one value from each iterable to the function and stops when the **shortest** iterable ends.

```python
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]
print(list(map(lambda x, y: x + y, numbers1, numbers2)))
print(list(map(lambda x, y: x + y, [1, 2, 3], [10])))
```

## 4. Reuse built-in functions and methods

Functions such as `int`, `str.upper`, `abs`, and `len` can be passed directly.

```python
str_numbers = ["1", "2", "3", "4", "5"]
print(list(map(int, str_numbers)))

words = ["apple", "banana", "cherry"]
print(list(map(str.upper, words)))
print(list(map(len, words)))
```

## 5. Transform records

The original example extracts each person's name. A comprehension often reads more naturally for simple extraction.

```python
def get_name(person):
    return person["name"]

people = [{"name": "Prabh", "age": 32}, {"name": "Jack", "age": 33}]
print(list(map(get_name, people)))
print([person["name"] for person in people])
```

## 6. Laziness and one-time consumption

No transformation happens until the iterator is consumed. After consumption, the same iterator is exhausted.

```python
def announce_square(number):
    print("processing", number)
    return number ** 2

lazy = map(announce_square, [1, 2, 3])
print("created")
print(list(lazy))
print("second use:", list(lazy))
```

## 7. Error handling during conversion

`map(int, values)` stops at the first invalid string. If failures should be handled per item, use a named safe-conversion function.

```python
def safe_int(text):
    try:
        return int(text)
    except (TypeError, ValueError):
        return None

print(list(map(safe_int, ["10", "oops", None, "30"])))
```

## 8. `itertools.starmap`

When each iterable item already contains the arguments for a multi-parameter function, `starmap` unpacks each item.

```python
from itertools import starmap
from operator import mul

pairs = [(2, 3), (4, 5), (6, 7)]
print(list(starmap(mul, pairs)))
```

## 9. `map` versus comprehension

| Prefer | When |
|---|---|
| `map(existing_function, data)` | A named/built-in transformation already exists |
| `[expression for x in data]` | The expression is clearer inline |
| Generator expression | You want lazy values with readable expression syntax |

## Revision cheat sheet

`map(f, xs)` lazily produces `f(x)` for each `x`. It can accept multiple iterables, stops at the shortest, and is consumed once. Pass the function object (`square`), not the result of calling it (`square()`).
