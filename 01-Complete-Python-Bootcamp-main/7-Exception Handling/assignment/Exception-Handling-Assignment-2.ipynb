### Q15. Aggregate failures from reusable rules.

**Concept tested:** validation framework

**Question:** Aggregate failures from reusable rules.

**Hint:** Each rule raises a custom error.

**Solution approach:** Collect messages and raise once.

**Step-by-step explanation:** Collect messages and raise once. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Later problems combine functions and deliberate error aggregation.

```python
class ValidationError(Exception): pass
def positive(n):
 if n<=0: raise ValidationError('not positive')
def even(n):
 if n%2: raise ValidationError('not even')
def validate(v,rules):
 errors=[]
 for rule in rules:
  try: rule(v)
  except ValidationError as e: errors.append(str(e))
 if errors: raise ValidationError('; '.join(errors))
try: validate(-3,[positive,even])
except ValidationError as e: print(e)
```
