# Exception Handbook — Five Completed Exercises 🚑

Expected problems deserve useful recovery. The original handbook examples and exercises are completed below without interactive blocking.

## Core pattern

```python
def safe_divide(a,b):
    try: return a/b
    except ZeroDivisionError: return None
print('Division by zero is not allowed.' if safe_divide(10,0) is None else safe_divide(10,0))
```

## 1. Keep asking until valid digits (simulated inputs)

```python
def first_valid_integer(inputs):
    for text in inputs:
        try: return int(text)
        except ValueError: print('Please enter whole digits.')
    raise ValueError('no valid input supplied')
print(first_valid_integer(['hello','3.5','7']))
```

## 2. Display infinite for division by zero

```python
def divide_or_infinite(a,b):
    try: return a/b
    except ZeroDivisionError: return float('inf')
print(divide_or_infinite(10,0))
```

## 3. Open several files without stopping

```python
from pathlib import Path
def read_available(paths):
    results={}
    for path in map(Path,paths):
        try: results[str(path)]=path.read_text(encoding='utf-8')
        except FileNotFoundError as error: results[str(path)]=f'MISSING: {error.filename}'
    return results
for path,result in read_available(['example1.txt','missing-a.txt','missing-b.txt']).items(): print(path,'->',result[:50])
```

## 4. Validate age 0–130

```python
def set_age(age):
    if not isinstance(age,int) or isinstance(age,bool): raise TypeError('age must be an integer')
    if not 0<=age<=130: raise ValueError('age must be between 0 and 130')
    return age
try: set_age(150)
except (TypeError,ValueError) as error: print('Invalid input:',error)
```

## 5. Why broad catching is harmful

`except Exception` can hide typos and broken assumptions. Catch specific expected errors near the risky operation. Use a broad handler only at a deliberate application boundary where you log/re-raise or produce a final safe response.

## Key concepts summary

The exercises demonstrate bounded retry, specific recovery, per-file isolation, validation, and narrow exception handling.

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Specific catch | `except ValueError:` |
| Capture message | `except OSError as error:` |
| Always cleanup | context manager / `finally` |
| Contract error | `raise ValueError(...)` |

## Common mistakes and interview tips

- Do not convert every failure to `None`.
- Do not hide bugs with broad catches.
- Validate types and ranges separately.
- Process independent items separately when one failure should not stop the batch.

**Revision habit:** explain what each line does, predict the result, run it, and test one edge case.
