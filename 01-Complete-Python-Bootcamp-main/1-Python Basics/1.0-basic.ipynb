## Goal

Imagine Python is a friendly robot. It is very helpful, but it follows instructions **exactly**. In this lesson you will learn how to:

- run code in a notebook;
- print messages and simple answers;
- understand syntax, meaning, comments, and indentation;
- tell expressions from statements;
- read common errors without being scared;
- build and explain a tiny program.

> **Learning recipe:** First **predict** what a cell will do, then **run** it, then **change** one small thing and run it again.

## Setup — How a notebook works

A notebook is like an exercise book made of boxes called **cells**.

- A **Markdown cell** contains notes like these.
- A **code cell** contains Python instructions.
- Click a code cell and press **Shift + Enter** to run it.
- The number beside a code cell shows when it ran.
- If you change an earlier value, run the later cells again so they use the new value.

Python normally reads a code cell from **top to bottom**, like reading a shopping list.

## 1. `print()` makes Python speak

`print()` displays something. Text must live inside quotes. The quotes are like a fence telling Python, “This is text.”

```python
print("Hello, learner!")
print("Python is my friendly robot.")
print(2 + 3)
```

`print("2 + 3")` prints the characters `2 + 3`, but `print(2 + 3)` asks Python to do the addition. Try both.

```python
print("2 + 3")
print(2 + 3)
print("Two apples plus three apples is", 2 + 3)
```

## 2. Syntax and semantics

- **Syntax** is the grammar or shape of the instruction: quotes, parentheses, colons, and spelling.
- **Semantics** is its meaning: what the valid instruction actually does.

`print("Hi")` has valid syntax and means “show the word Hi.” A missing closing quote breaks the grammar.

## 3. Python is case-sensitive

Python treats uppercase and lowercase letters as different labels. `name`, `Name`, and `NAME` can hold three different things.

```python
name = "Maya"
Name = "Leo"
NAME = "Zara"
print(name)
print(Name)
print(NAME)
```

## 4. Comments are notes for humans

Anything after `#` on a line is a **comment**. Python ignores it. Good comments explain **why** something is happening, not merely repeat the code.

```python
# We use 5 because each bag contains five marbles.
bags = 3
marbles = bags * 5  # A short comment can also go after code.
print("Total marbles:", marbles)
```

## 5. Indentation makes instruction boxes

Spaces at the start of a line are called **indentation**. After a line ending in `:`, Python expects an indented block. Think of the indented lines as toys placed inside one toy box.

Python convention uses **four spaces** per level. Do not mix tabs and spaces.

```python
age = 8  # Change this to 3 and run again.

if age >= 5:
    print("You can start school")
    print("This line is inside the if block")

print("This line is outside, so it always runs")
```

## 6. Values, expressions, and statements

- A **value** is a piece of data: `5`, `"cat"`, or `True`.
- An **expression** produces a value: `2 + 3` produces `5`.
- A **statement** tells Python to do something: `x = 5`, `print(x)`, and `if ...` are statements.

A small program usually combines several statements, and those statements contain expressions.

```python
x = 5             # assignment statement
y = 10            # assignment statement
z = x + y         # x + y is an expression
print("z =", z)   # function-call statement
```

## 7. A tiny program: input → process → output

Many programs follow the same three-part story:

1. **Input:** collect or store information.
2. **Process:** calculate or decide something.
3. **Output:** show the result.

```python
# Input
apples = 4
oranges = 3

# Process
total_fruit = apples + oranges

# Output
print("Fruit in the basket:", total_fruit)
```

## 8. Escape characters and multi-line text

Special combinations beginning with `\` help format text:

- `\n` starts a new line;
- `\t` inserts a tab;
- `\"` places a double quote inside double-quoted text.

Triple quotes can hold text across several lines.

```python
print("Line one\nLine two")
print("Name:\tMaya")
print("She said, \"Hello!\"")

little_poem = """Roses are red,
Python is neat,
I give it instructions,
It makes them complete!"""
print(little_poem)
```

## 9. Errors are clues, not monsters 👾

An error message tells you where Python became confused. Read the **last line first**.

| Error | Child-friendly meaning | Common cause |
|---|---|---|
| `SyntaxError` | “I cannot read this sentence.” | Missing quote, bracket, or colon |
| `IndentationError` | “This line is in the wrong box.” | Missing or uneven spaces |
| `NameError` | “I do not know that label.” | Misspelled or not-yet-created name |
| `TypeError` | “These kinds of things do not work together.” | Adding text and a number directly |
| `ZeroDivisionError` | “You cannot split into zero groups.” | Dividing by `0` |

The broken examples below are comments so the whole notebook still runs. Remove one leading `#` at a time, observe the error, then put it back.

```python
# print("missing quote)
# print(unknown_name)
# print("Age: " + 8)
# print(10 / 0)

print("Safe fixed versions:")
print("Age:", 8)
print(10 / 2)
```

## 10. Helpful built-ins

Python arrives with useful tools called **built-in functions**:

- `print()` shows a value;
- `type()` tells its kind;
- `len()` counts items or characters;
- `help()` explains a tool.

```python
word = "python"
print(type(word))
print(len(word))
print(word.upper())

# Try help(print) or help(str.upper) when you want Python's built-in documentation.
```
