# Python Basics 1 — Variables: Labels on Boxes 📦

## Goal

A variable is a **name that points to a value**. Imagine putting `25` in a box and sticking the label `age` on it. You will learn assignment, naming rules, reassignment, multiple assignment, swapping, conversion, input, constants, `None`, and common mistakes.

## 1. Create and use variables

`=` is the **assignment operator**. Read `age = 25` as “put the value `25` under the name `age`.” It does **not** ask whether two things are equal; `==` does that.

```python
age = 25
height_m = 1.65
name = "Maya"
is_student = True

print(f"{name} is {age} years old")
print("Height:", height_m, "metres")
print("Student?", is_student)
```

## 2. Good names tell a story

Names may contain letters, numbers, and `_`, but cannot start with a number. They are case-sensitive and cannot be Python keywords.

| Prefer ✅ | Avoid ❌ | Why |
|---|---|---|
| `items_in_cart` | `x` | Meaning is clear |
| `student_name` | `first-name` | `-` means subtraction |
| `total_price` | `2price` | Names cannot start with a digit |
| `class_name` | `class` | `class` is a keyword |

Python style normally uses `snake_case`. Use nouns for data and later use verbs for functions.

```python
import keyword

print("Is 'if' a keyword?", keyword.iskeyword("if"))
print("Is 'student_name' a keyword?", keyword.iskeyword("student_name"))
```

## 3. Reassignment: replace what the label points to

The right side is calculated **first**, then the name on the left receives the result.

```python
score = 10
print("Start:", score)

score = score + 5
print("After earning 5:", score)

score += 5   # Short form of score = score + 5
score *= 2   # Short form of score = score * 2
print("After shortcuts:", score)
```

## 4. Python is dynamically typed

The **value** has a type. A name can later point to a different kind of value. This is flexible, but changing meaning unexpectedly can confuse readers.

```python
thing = 10
print(thing, type(thing).__name__)

thing = "ten"
print(thing, type(thing).__name__)
```

## 5. Assign several variables and swap them

Python can unpack several values at once. The number of names must match the number of values.

```python
width, height = 4, 3
print("Before swap:", width, height)

width, height = height, width
print("After swap:", width, height)

red = green = blue = 0
print(red, green, blue)
```

## 6. Converting values

Conversion is like moving the same idea into a different costume:

- `int("25")` → `25`
- `float("3.5")` → `3.5`
- `str(25)` → `"25"`
- `bool(1)` → `True`

Bad numeric text such as `int("hello")` causes a `ValueError`.

```python
age_text = "25"
age = int(age_text)
price = float("19.99")

print(f"Age next year: {age + 1}")
print(f"Price: {price}")
print("A number turned into text:", str(42))
```

Even if a person types `12`, `input()` gives Python the string `"12"`. Convert it before doing arithmetic. The interactive version is commented out so “Run All” never pauses.

## 8. F-strings make sentences readable

Put `f` before the opening quote and place expressions inside `{}`. Formatting rules can control decimal places.

```python
product = "notebook"
price = 49.5
quantity = 3
total = price * quantity

print(f"{quantity} {product}s cost ₹{total:.2f}")
```

## 9. Constants are a promise made by programmers

Python does not lock ordinary variables. By convention, an uppercase name means “please do not change this value.”

```python
PI = 3.14159
MAX_PLAYERS = 4
radius = 2
area = PI * radius ** 2
print(f"Circle area: {area:.2f}")
```

## 10. `None` means “no value yet”

`None` is not zero and not empty text. It is a special placeholder meaning “nothing has been chosen or returned yet.” Use `is None` to check it.

```python
winner = None
print("Winner stored yet?", winner is not None)

winner = "Maya"
print("Winner stored yet?", winner is not None)
```

## 11. Names can point to the same mutable object

This is a peek ahead. Two labels can be stuck on the **same list box**, so changing the list through one label is visible through the other. Use `.copy()` when you need a separate list.

```python
first_basket = ["apple"]
same_basket = first_basket
separate_basket = first_basket.copy()

same_basket.append("banana")
print("First:", first_basket)
print("Same object:", same_basket)
print("Separate copy:", separate_basket)
```

## Checks — Original practice, kept and explained

1. Store a product, price, and quantity, then calculate the total.
2. Convert `"42"` to an integer and add `8`.
3. Convert Celsius with `F = C × 9/5 + 32`.

```python
product = "soap"
price = 20
quantity = 300
total = price * quantity
print(f"{product} costs {price} each and quantity is {quantity}. Total cost: {total}")

number_text = "42"
number = int(number_text)
new_number = number + 8
print(f"New number is {new_number} and type is {type(new_number).__name__}")

celsius = 48
fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

## Easy revision cheat sheet

| Need | Pattern | Example |
|---|---|---|
| Store a value | `name = value` | `age = 12` |
| Compare equality | `==` | `age == 12` |
| Clear name | `snake_case` | `total_price` |
| Update | `+=`, `-=`, `*=`, `/=` | `score += 1` |
| Multiple assignment | `a, b = 1, 2` | Two labels at once |
| Swap | `a, b = b, a` | No temporary name needed |
| Convert | `int()`, `float()`, `str()` | `int("12")` |
| Insert in text | `f"...{value}..."` | `f"Age: {age}"` |
| No value yet | `None` | `answer = None` |
| Intended constant | uppercase name | `PI = 3.14` |

**Common traps:** names cannot start with digits; `input()` returns text; `=` stores while `==` compares; avoid changing a variable's meaning halfway through a program.
