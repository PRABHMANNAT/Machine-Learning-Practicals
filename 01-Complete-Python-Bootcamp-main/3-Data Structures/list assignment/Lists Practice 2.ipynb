### Q20. Merge two already sorted lists

**Concept tested:** two-pointer algorithm

**Solution logic:** Compare current items, take the smaller, then append all leftovers.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Two pointers merge sorted inputs in linear time.

```python
left, right = [1, 4, 7], [2, 3, 8]
merged, i, j = [], 0, 0
while i < len(left) and j < len(right):
    if left[i] <= right[j]:
        merged.append(left[i]); i += 1
    else:
        merged.append(right[j]); j += 1
merged.extend(left[i:]); merged.extend(right[j:])
print(merged)
```

### Q25. Sort student records by two rules

**Solution logic:** Use negative score for descending scores and name for ascending ties.

### Q26. Check balanced parentheses with a stack

**Concept tested:** stack pattern

**Solution logic:** Push opening brackets; each closing bracket must match the newest opening bracket.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** A list's `append/pop` pair implements last-in-first-out behavior.

```python
text = "([{}])"
pairs = {")":"(", "]":"[", "}":"{"}
stack = []
balanced = True
for char in text:
    if char in "([{":
        stack.append(char)
    elif not stack or stack.pop() != pairs[char]:
        balanced = False
        break
balanced = balanced and not stack
print(balanced)
```

### Q27. Process people in queue order

**Concept tested:** deque as a queue

**Solution logic:** Append arrivals on the right and remove the oldest from the left.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** `deque.popleft()` avoids the shifting cost of list `pop(0)`.

```python
from collections import deque
queue = deque(["Maya", "Leo"])
queue.append("Zara")
served = []
while queue:
    served.append(queue.popleft())
print(served)
```

### Q29. Find the maximum-sum contiguous sublist

**Concept tested:** Kadane's algorithm

**Solution logic:** Track the best sum ending here and the best sum seen anywhere.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Dynamic programming can reduce a nested-search problem to one pass.

```python
numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
best_here = best_total = numbers[0]
for number in numbers[1:]:
    best_here = max(number, best_here + number)
    best_total = max(best_total, best_here)
print(best_total)
```

### Q30. Find all pairs that reach a target

**Concept tested:** set-assisted list algorithm

**Solution logic:** Remember seen numbers; the needed partner is `target - number`; normalize pairs to avoid duplicates.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** A helper set makes complement lookup fast while the result remains ordered for display.

```python
numbers = [2, 7, 11, 15, 3, 6, 4, 5]
target, seen, pairs = 9, set(), set()
for number in numbers:
    needed = target - number
    if needed in seen:
        pairs.add(tuple(sorted((number, needed))))
    seen.add(number)
print(sorted(pairs))
```
