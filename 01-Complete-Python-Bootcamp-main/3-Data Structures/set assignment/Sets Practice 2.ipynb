### Q19. Combine role permissions

**Concept tested:** union across sets

**Solution logic:** Start empty and union every role's permission set.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Sets naturally model permissions without duplicates.

```python
roles = {"viewer":{"read"}, "editor":{"read","write"}}
assigned = ["viewer", "editor"]
permissions = set()
for role in assigned:
    permissions |= roles[role]
print(sorted(permissions))
```

### Q20. Compare two documents with Jaccard similarity

**Concept tested:** intersection/union ratio

**Solution logic:** Divide shared unique words by all unique words.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Jaccard similarity compares overlap while ignoring duplicates.

```python
a = set("python data science".split())
b = set("python machine learning".split())
score = len(a & b) / len(a | b)
print(f"{score:.2f}")
```

### Q25. Group words by unique-letter signature

**Concept tested:** frozenset grouping

**Solution logic:** Use each word's frozenset of letters as the dictionary key.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** A frozenset is useful when order and repetition do not matter.

```python
words = ["eat", "tea", "tan", "ant"]
groups = {}
for word in words:
    groups.setdefault(frozenset(word), []).append(word)
print(sorted(sorted(group) for group in groups.values()))
```

### Q26. Generate a power set

**Concept tested:** set combinations

**Solution logic:** Start with the empty subset; for each value, add it to every existing subset.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Frozensets let a set contain other set-like subsets.

```python
values = {1, 2, 3}
power = {frozenset()}
for value in values:
    power |= {subset | {value} for subset in power}
print(sorted((sorted(s) for s in power), key=lambda x: (len(x), x)))
```

### Q27. Choose stores with a greedy set cover

**Concept tested:** set-cover heuristic

**Solution logic:** Repeatedly select the store covering the most still-needed items.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Greedy set cover is practical, though not always globally optimal.

```python
needed = {"milk", "bread", "eggs", "fruit"}
stores = {"A":{"milk","bread"}, "B":{"eggs","fruit"}, "C":{"bread","eggs"}}
chosen = []
while needed:
    best = max(stores, key=lambda name: len(stores[name] & needed))
    chosen.append(best)
    needed -= stores[best]
print(chosen)
```

### Q28. Validate a Sudoku row

**Solution logic:** Valid digits must equal the complete set 1 through 9.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Set equality can validate exact membership requirements.

```python
row = [5, 3, 4, 6, 7, 8, 9, 1, 2]
valid = len(row) == 9 and set(row) == set(range(1, 10))
print(valid)
```

### Q29. Find elements appearing in an odd number of sets

**Concept tested:** repeated symmetric difference

**Solution logic:** Fold `^` across all sets so even appearances cancel.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Symmetric difference behaves like parity across memberships.

```python
groups = [{1,2,3}, {2,3,4}, {3,4,5}]
odd = set()
for group in groups:
    odd ^= group
print(sorted(odd))
```

### Q30. Build unique unordered pairs

**Concept tested:** frozenset and combinations

**Solution logic:** Create all two-item combinations and freeze each pair so order is irrelevant.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Frozenset represents an unordered, unique compound value.

```python
from itertools import combinations
values = [1, 2, 2, 3]
pairs = {frozenset(pair) for pair in combinations(values, 2) if pair[0] != pair[1]}
print(sorted(sorted(pair) for pair in pairs))
```
