# Python Data Structures 2 — Tuples: A Sealed Lunchbox 🍱

## Goal

A tuple is an **ordered collection that cannot be structurally changed after creation**. It is like a sealed lunchbox: you can look at each compartment, but you cannot add, remove, or replace compartments. Tuples are excellent for fixed records, safe returns, coordinates, and dictionary keys.

## 1. Create tuples—and remember the comma

Parentheses improve readability, but the **comma** creates a tuple. `(5)` is just integer `5`; `(5,)` is a one-item tuple.

```python
empty_tuple = ()
also_empty = tuple()
numbers = tuple([1, 2, 3, 4, 5, 6])
mixed_tuple = (1, "Hello World", 3.14, True)
single_wrong = (5)
single_right = (5,)

print(empty_tuple, type(empty_tuple).__name__)
print(numbers)
print(mixed_tuple)
print(type(single_wrong).__name__, type(single_right).__name__)
print(list((1, 2, 3, 4, 5, 6)))
```

## 2. Index, slice, and iterate

Tuple seats are numbered exactly like list seats. Slicing returns a new tuple and `stop` is excluded.

```python
print(numbers[2])
print(numbers[-1])
print(numbers[0:4])
print(numbers[::-1])

for number in numbers:
    print(number, end=" ")
print()
```

## 3. Tuple operations

`+` concatenates tuples, `*` repeats them, `in` tests membership, and built-ins such as `len`, `min`, `max`, and `sum` work when appropriate.

```python
concatenated = numbers + mixed_tuple
print(concatenated)
print(mixed_tuple * 3)
print(numbers * 3)
print(3 in numbers, len(numbers), sum(numbers))
```

## 4. Immutability

You may change a list seat, but tuple item assignment raises `TypeError`. The broken line is caught so the notebook stays runnable.

```python
editable = [1, 2, 3, 4, 5]
editable[1] = "Prabh"
print(editable)

try:
    numbers[1] = "Krish"
except TypeError as error:
    print("TypeError:", error)
print(numbers)
```

### Important edge case: immutable container, mutable child

A tuple cannot point to a different child, but a mutable child such as a list can change itself. “Tuple is immutable” means its references are fixed—not that everything nested inside is frozen.

```python
record = ("Maya", [90, 95])
record[1].append(100)
print(record)
```

## 5. Only two tuple methods

Because tuples cannot grow or shrink, they need only `count(value)` and `index(value)`. Missing `index` raises `ValueError`.

```python
values = (1, 2, 1, 3, 1)
print(values.count(1))
print(values.index(3))
```

## 6. Packing and unpacking

Packing bundles values. Unpacking gives each position a name. The counts must match unless one name uses `*` to collect leftovers into a list.

```python
packed_tuple = 1, "Hello", 3.14
print(packed_tuple)
a, b, c = packed_tuple
print(a, b, c)

numbers = (1, 2, 3, 4, 5, 6)
first, *middle, last = numbers
print(first, middle, last)
```

### Swapping and ignoring values

Python uses packing/unpacking behind the scenes. `_` conventionally means “I intentionally do not need this value.”

```python
left, right = "L", "R"
left, right = right, left
print(left, right)

name, _, score = ("Maya", 12, 95)
print(name, score)
```

## 7. Nested tuples and mixed nesting

Use one index per level. The original nested-list/tuple examples are preserved here.

```python
nested_list = [[1, 2, 3, 4], [6, 7, 8, 9], [1, "Hello", 3.14, "c"]]
print(nested_list[0][0:3])

mixed_nested = [[1, 2, 3, 4], [6, 7, 8, 9], (1, "Hello", 3.14, "c")]
print(mixed_nested[2][0:3])

nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))
print(nested_tuple[0])
print(nested_tuple[1][2])
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```

## 8. Functions often return tuples

Returning several values actually returns one tuple. The caller can unpack it.

```python
def minimum_and_maximum(values):
    return min(values), max(values)

result = minimum_and_maximum([8, 3, 11, 4])
print(result, type(result).__name__)
smallest, largest = result
print(smallest, largest)
```

## 9. Tuple comparison

Tuples compare lexicographically: compare first items; if tied, compare second items, and so on. This is useful for multi-rule sorting.

```python
print((1, 5) < (2, 0))
print((1, 5) < (1, 8))

students = [(90, "Zara"), (95, "Maya"), (90, "Leo")]
print(sorted(students, key=lambda item: (-item[0], item[1])))
```

## 10. Hashability and dictionary keys

A tuple can be a set item or dictionary key **only if every nested item is hashable**. A tuple containing a list is not hashable.

```python
board = {(0, 0): "start", (2, 3): "treasure"}
print(board[(2, 3)])

try:
    hash((1, [2, 3]))
except TypeError as error:
    print("TypeError:", error)
```

## 11. Named tuples: tuple safety with readable labels

`namedtuple` keeps tuple behavior but lets us say `.x` and `.y` instead of remembering positions.

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
point = Point(3, 4)
print(point)
print(point.x, point[0])
print(point._replace(y=10))
```

## 12. Generator expression versus tuple comprehension

Python has no special tuple-comprehension syntax. `(x*x for x in data)` creates a generator. Wrap it with `tuple(...)` to collect a tuple.

```python
generator = (number ** 2 for number in range(5))
print(type(generator).__name__)
squares = tuple(number ** 2 for number in range(5))
print(squares)
```

## 13. Practical cases

- Coordinates: `(latitude, longitude)`
- RGB color: `(red, green, blue)`
- Database row or function result
- Compound dictionary key
- Fixed configuration that should not be edited accidentally

```python
locations = [("Home", 28.61, 77.21), ("Park", 28.59, 77.23)]
for place, latitude, longitude in locations:
    print(f"{place}: {latitude}, {longitude}")

rgb_red = (255, 0, 0)
print("Red channel:", rgb_red[0])
```

## 14. Common mistakes

- Forgetting the comma in a one-item tuple.
- Trying `append`, `remove`, or item assignment—tuples do not have these.
- Unpacking into the wrong number of names.
- Believing nested mutable objects become frozen.
- Using a tuple containing a list as a set item or dictionary key.
- Expecting `(x for x in values)` to create a tuple; it creates a generator.

## Easy revision cheat sheet

| Need | Pattern |
|---|---|
| Create | `(1, 2)` or `tuple(iterable)` |
| One item | `(value,)` |
| Read/slice | `items[i]`, `items[a:b:c]` |
| Join/repeat | `a + b`, `a * 3` |
| Count/find | `count(x)`, `index(x)` |
| Unpack | `a, b = pair` |
| Gather middle | `first, *middle, last = values` |
| Fixed compound key | `{(row, col): value}` |
| Readable fixed record | `collections.namedtuple` |

**Memory line:** tuples are ordered and immutable, allow duplicates, and use commas. A one-item tuple needs a trailing comma.
