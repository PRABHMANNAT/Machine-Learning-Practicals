## Challenging questions (21–30)

### Q21. Deep-merge nested settings

**Concept tested:** recursive dictionaries

**Solution logic:** When both colliding values are dictionaries, merge recursively; otherwise use the override.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** A shallow merge would replace the entire nested `db` mapping.

```python
def deep_merge(left, right):
    result = left.copy()
    for key, value in right.items():
        if key in result and isinstance(result[key], dict) and isinstance(value, dict):
            result[key] = deep_merge(result[key], value)
        else:
            result[key] = value
    return result

base = {"db": {"host": "local", "port": 5432}, "debug": False}
override = {"db": {"port": 6000}}
print(deep_merge(base, override))
```

### Q22. Invert a dictionary with duplicate values

**Concept tested:** grouped inversion

**Solution logic:** Append every original key under its value using `defaultdict(list)`.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** When values repeat, inverted values must hold multiple keys.

```python
from collections import defaultdict
data = {"Maya": "red", "Leo": "blue", "Zara": "red"}
inverse = defaultdict(list)
for key, value in data.items():
    inverse[value].append(key)
print(dict(inverse))
```

### Q23. Flatten a nested dictionary

**Concept tested:** recursion and compound path keys

**Solution logic:** Recurse into dictionaries while joining path parts with dots.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Recursive traversal can turn hierarchy into path-value pairs.

```python
def flatten(data, prefix=""):
    result = {}
    for key, value in data.items():
        path = f"{prefix}.{key}" if prefix else key
        if isinstance(value, dict):
            result.update(flatten(value, path))
        else:
            result[path] = value
    return result

print(flatten({"user": {"name": "Maya", "address": {"city": "Pune"}}}))
```

### Q24. Rebuild a nested dictionary from dotted paths

**Concept tested:** setdefault and traversal

**Solution logic:** Walk all path pieces except the last, creating dictionaries as needed.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** A cursor variable can progressively descend into nested mappings.

```python
flat = {"user.name": "Maya", "user.address.city": "Pune"}
nested = {}
for path, value in flat.items():
    cursor = nested
    parts = path.split(".")
    for part in parts[:-1]:
        cursor = cursor.setdefault(part, {})
    cursor[parts[-1]] = value
print(nested)
```

### Q25. Memoize Fibonacci values

**Concept tested:** dictionary cache

**Solution logic:** Return cached answers; otherwise compute, store, and return them.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Memoization trades memory for speed by avoiding repeated work.

```python
cache = {0: 0, 1: 1}

def fibonacci(n):
    if n not in cache:
        cache[n] = fibonacci(n - 1) + fibonacci(n - 2)
    return cache[n]

print(fibonacci(10))
print(cache)
```

### Q26. Group anagrams

**Concept tested:** normalized tuple keys

**Solution logic:** Sort each word's letters into a hashable tuple key and append the word.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Canonical hashable signatures group equivalent records.

```python
from collections import defaultdict
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = defaultdict(list)
for word in words:
    groups[tuple(sorted(word))].append(word)
print(sorted((sorted(group) for group in groups.values()), key=lambda group: (len(group), group)))
```

### Q29. Apply inventory transactions

**Concept tested:** nested lookup and validation

**Solution logic:** Check item existence and sufficient stock before subtracting each sale.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Mappings make keyed state updates and validation straightforward.

```python
inventory = {"pen": 10, "book": 4}
sales = {"pen": 3, "book": 5}
for item, quantity in sales.items():
    if quantity <= inventory.get(item, 0):
        inventory[item] -= quantity
    else:
        print(f"Cannot sell {quantity} {item}(s)")
print(inventory)
```

### Q30. Build a student report

**Concept tested:** nested aggregation

**Solution logic:** Compute each student's average, then choose the maximum report entry by average.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Nested dictionaries can store derived summaries for later decisions.

```python
students = {"Maya": [90, 95, 100], "Leo": [80, 85, 90], "Zara": [88, 92, 96]}
report = {
    name: {"average": sum(scores) / len(scores), "passed": min(scores) >= 40}
    for name, scores in students.items()
}
top = max(report, key=lambda name: report[name]["average"])
print(report)
print("Top:", top)
```
