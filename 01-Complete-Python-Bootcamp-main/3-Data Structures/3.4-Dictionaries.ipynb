# Python Data Structures 4 — Dictionaries: Labeled Drawers 🗄️

## Goal

A dictionary maps a unique **key** to a **value**. Imagine labeled drawers: instead of remembering seat number `2`, ask for drawer `"grade"`. Dictionaries are mutable and preserve insertion order in modern Python.

## 1. Create dictionaries

Keys must be unique and hashable; values may be anything. Repeating a key in a literal silently keeps the last value.

```python
empty_dict = {}
also_empty = dict()
student = {"name": "Prabh", "age": 32, "grade": 24}
duplicate_key = {"name": "Prabh", "age": 32, "name": 24}
from_pairs = dict([("a", 1), ("b", 2)])
with_keywords = dict(name="Maya", age=12)

print(type(empty_dict).__name__, also_empty)
print(student)
print(duplicate_key)
print(from_pairs, with_keywords)
```

## 2. Read values: brackets versus `get`

`mapping[key]` raises `KeyError` if missing. `get(key)` returns `None`, or your supplied default. Use brackets when absence means a bug; use `get` when absence is normal.

```python
student = {"name": "Prabh", "age": 32, "grade": "A"}
print(student["grade"])
print(student["age"])
print(student.get("grade"))
print(student.get("last_name"))
print(student.get("last_name", "Not Available"))

try:
    print(student["last_name"])
except KeyError as error:
    print("KeyError:", error)
```

## 3. Add, update, and merge

Assigning a new key adds a drawer; assigning an existing key replaces its content. `update` adds/replaces many pairs. `|` creates a merged dictionary in Python 3.9+; later values win on collisions.

```python
student["age"] = 20
student["address"] = "India"
student.update({"grade": "A+", "active": True})
print(student)

defaults = {"theme": "light", "language": "en"}
choices = {"theme": "dark"}
print(defaults | choices)
```

## 4. Remove items

- `del d[key]` removes a known key and raises if absent.
- `pop(key, default)` removes and returns a value.
- `popitem()` removes and returns the most recently inserted pair.
- `clear()` empties the dictionary.

```python
working = student.copy()
del working["grade"]
address = working.pop("address")
missing = working.pop("phone", "Not found")
last_pair = working.popitem()
print(address, missing, last_pair)
print(working)
working.clear()
print(working)
```

## 5. Views: `keys`, `values`, and `items`

These methods return dynamic **view objects**, not frozen lists. If the dictionary changes, the view reflects it. Convert to `list` when a snapshot or indexing is needed.

```python
student = {"name": "Prabh", "age": 20, "address": "India"}
keys_view = student.keys()
print(keys_view)
print(student.values())
print(student.items())
student["grade"] = "A"
print("Dynamic view:", keys_view)
print("Snapshot:", list(keys_view))
```

## 6. Iterate over dictionaries

Plain iteration gives keys. Use `.values()` for values and `.items()` to unpack key-value pairs.

```python
for key in student:
    print(key)
for value in student.values():
    print(value)
for key, value in student.items():
    print(f"{key}: {value}")
```

## 7. `setdefault`: get or create once

`setdefault(key, default)` returns the existing value; if absent, it inserts and returns the default. It is useful for grouping, but `defaultdict` is often cleaner.

```python
groups = {}
for name, team in [("Maya", "red"), ("Leo", "blue"), ("Zara", "red")]:
    groups.setdefault(team, []).append(name)
print(groups)
```

## 8. Copying and aliasing

Assignment shares one dictionary. `.copy()` makes a shallow outer copy. Nested mutable values remain shared; use `deepcopy` when independent nested data is required.

```python
from copy import deepcopy

student = {"name": "Prabh", "scores": [80, 90]}
alias = student
shallow = student.copy()
deep = deepcopy(student)

student["name"] = "Prabh2"
student["scores"].append(100)
print("alias:", alias)
print("shallow:", shallow)
print("deep:", deep)
```

## 9. Nested dictionaries

Each student ID maps to another dictionary. Use one key per level and nested loops to visit everything.

```python
students = {
    "student1": {"name": "Krish", "age": 32},
    "student2": {"name": "Peter", "age": 35},
}
print(students)
print(students["student2"]["name"])
print(students["student2"]["age"])
print(students.items())

for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"  {key}: {value}")
```

## 10. Dictionary comprehensions

The shape is `{key_expression: value_expression for item in iterable if condition}`. Duplicate generated keys keep the last value.

```python
squares = {x: x ** 2 for x in range(5)}
even_squares = {x: x ** 2 for x in range(10) if x % 2 == 0}
swapped = {value: key for key, value in {"a": 1, "b": 2}.items()}
print(squares)
print(even_squares)
print(swapped)
```

## 11. Count frequencies

The original manual pattern teaches dictionary thinking. `get` makes it shorter, and `Counter` is the standard specialized tool.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print(frequency)

frequency_with_get = {}
for number in numbers:
    frequency_with_get[number] = frequency_with_get.get(number, 0) + 1
print(frequency_with_get)

from collections import Counter
print(Counter(numbers))
```

## 12. Merge dictionaries and collision rules

`**` unpacks dictionary pairs into a new literal. Think of emptying labeled lunchboxes into one larger box: a later duplicate label replaces the earlier value.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}
merged_unpacking = {**dict1, **dict2}
merged_operator = dict1 | dict2
print(merged_unpacking)
print(merged_operator)
```

## 13. Hashable keys

Strings, numbers, and hashable tuples may be keys. Lists and dictionaries may not, because changing them would make their storage location unreliable.

```python
grid = {(0, 0): "start", (2, 3): "treasure"}
print(grid[(2, 3)])

try:
    broken = {[1, 2]: "not allowed"}
except TypeError as error:
    print("TypeError:", error)
```

## 14. `defaultdict` for automatic defaults

It calls a factory when a key is first missing. Accessing a missing key **creates it**, which differs from ordinary `get`.

```python
from collections import defaultdict

words = ["apple", "ant", "banana", "boat"]
by_first_letter = defaultdict(list)
for word in words:
    by_first_letter[word[0]].append(word)
print(dict(by_first_letter))
```

## 15. Sorting dictionary information

Dictionaries preserve insertion order, but they are not automatically sorted. Sort their keys or item pairs to produce an ordered report.

```python
scores = {"Zara": 90, "Maya": 95, "Leo": 88}
print(dict(sorted(scores.items())))
ranking = sorted(scores.items(), key=lambda pair: pair[1], reverse=True)
print(ranking)
```

## 16. `fromkeys` mutable-default trap

`dict.fromkeys(keys, mutable_value)` gives every key the same object. Use a comprehension for separate lists.

```python
shared = dict.fromkeys(["red", "blue"], [])
shared["red"].append("Maya")
print("Shared lists:", shared)

separate = {team: [] for team in ["red", "blue"]}
separate["red"].append("Maya")
print("Separate lists:", separate)
```

## 17. Common mistakes and safe patterns

- Bracket access raises `KeyError`; use `get` for optional data.
- Duplicate literal keys silently overwrite earlier values.
- Do not add/remove keys while iterating over the same dictionary; iterate over `list(d)` or build a new dictionary.
- Assignment is aliasing, not copying.
- Shallow copies still share nested mutable values.
- `in dictionary` checks keys, not values.
- Never use a mutable default with `fromkeys` when each key needs its own container.

```python
inventory = {"apple": 3, "banana": 0, "mango": 5}
for item in list(inventory):
    if inventory[item] == 0:
        del inventory[item]
print(inventory)
print("apple" in inventory)
print(5 in inventory.values())
```

## Easy revision cheat sheet

| Need | Tool |
|---|---|
| Create | `{key: value}` or `dict(...)` |
| Required lookup | `d[key]` |
| Optional lookup | `d.get(key, default)` |
| Add/update one | `d[key] = value` |
| Add/update many | `d.update(other)` |
| Remove/return | `d.pop(key, default)` |
| Loop pairs | `for key, value in d.items():` |
| Create default once | `setdefault` / `defaultdict` |
| Count | `d.get(x, 0) + 1` / `Counter` |
| Merge new dict | `{**a, **b}` or `a | b` |
| Build compactly | `{k: v for ...}` |
| Independent nested copy | `copy.deepcopy` |

**Memory line:** dictionaries map unique hashable keys to values, preserve insertion order, and are mutable. Brackets demand a key; `get` handles optional keys.
