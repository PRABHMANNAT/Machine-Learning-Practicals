# Object-Oriented Programming (OOP) — Explained Like You Are 5

## Big idea

Imagine you want to make toy cars:

- A **class** is the drawing or blueprint of a car.
- An **object** is a real toy car created from that blueprint.
- An **attribute** describes the toy, such as its color, doors, or name.
- A **method** is something the toy can do, such as drive, bark, deposit money, or withdraw money.

> **Simple rule:** A class is the plan. An object is the real thing made from the plan.

## 1. Creating a class

The code below says: **Make a blueprint named `Car`.**

`pass` means: **The blueprint is empty for now, but that is okay.**

```python
class Car:
    pass
```

Now we use the `Car` blueprint to create two separate cars:

- `audi` is one object.
- `bmw` is another object.

`type(audi)` asks Python: **What kind of thing is Audi?**

```python
audi = Car()
bmw = Car()

print(type(audi))
```

## 2. Strange-looking object output

Printing an object may show something like `<__main__.Car object at 0x...>`.

Python is saying:

> This is a `Car` object, and I stored it somewhere in the computer's memory.

The strange number is its memory location. You do not need to memorize it.

```python
print(audi)
print(bmw)
```

## 3. Giving an object information

An **attribute** is information attached to an object.

`audi.windows = 4` is like attaching a label that says: **This Audi has 4 windows.**

```python
audi.windows = 4
print(audi.windows)
```

## 4. Why Tata gives an error

We give Tata a `doors` attribute, but we never give it a `windows` attribute. Asking for `tata.windows` therefore causes an `AttributeError`.

It is like asking a child, **What is inside the box?**, before putting anything in the box.

> The `try` and `except` below let us look at the error without stopping the whole notebook.

```python
tata = Car()
tata.doors = 4

try:
    print(tata.windows)
except AttributeError as error:
    print(f"AttributeError: {error}")
```

## 5. What does `dir(tata)` do?

`dir(tata)` asks Python: **Show me everything Tata knows or can do.**

Python supplies many special names beginning with `__`. The important name we added is `doors`. To keep the output simple, the next cell shows Tata's own attributes and checks whether `doors` appears in `dir(tata)`.

```python
print("Tata's own attributes:", tata.__dict__)
print("Is 'doors' in dir(tata)?", "doors" in dir(tata))
```

## 6. Creating dogs properly

`__init__` is an automatic setup method. It runs whenever a new dog is created.

Think of `self` as meaning **this particular dog**. Therefore, `self.name = name` means: **Give this dog the name we received.**

Buddy and Lucy use the same `Dog` blueprint, but each dog keeps its own name and age.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age


dog1 = Dog("Buddy", 3)
dog2 = Dog("Lucy", 4)

print(dog1.name, dog1.age)
print(dog2.name, dog2.age)
```

## 7. Giving dogs an action

A function inside a class is called a **method**. The `bark()` method lets every dog bark.

The same method produces different words because `self.name` refers to the dog that called it.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")


dog1 = Dog("Buddy", 3)
dog2 = Dog("Lucy", 4)

dog1.bark()
dog2.bark()
```

## 8. The bank account example

The `BankAccount` class keeps two pieces of information:

- `owner`: the person's name
- `balance`: the money in the account

It also provides three actions:

1. `deposit()` adds money.
2. `withdraw()` removes money when enough is available.
3. `get_balance()` returns the current balance.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")

    def get_balance(self):
        return self.balance
```

### Create an account

`BankAccount("Krish", 5000)` creates Krish's account with a starting balance of 5,000.

```python
account = BankAccount("Krish", 5000)
print("Starting balance:", account.get_balance())
```

### Deposit money

Depositing 100 changes the balance from 5,000 to 5,100.

```python
account.deposit(100)
```

### Withdraw money

Withdrawing 300 changes the balance from 5,100 to 4,800.

```python
account.withdraw(300)
```

### Trying to withdraw too much

If we request more money than the account contains, the method protects the account and prints `Insufficient funds!`.

```python
account.withdraw(10000)
```

### Check the balance

`get_balance()` uses `return` to give the current balance back to us.

```python
print("Current balance:", account.get_balance())
```

## Tiny cheat sheet

| Word | Child-friendly meaning |
|---|---|
| `class` | Blueprint for making things |

| attribute | Information about the thing |
| method | An action the thing can perform |
| `__init__` | Automatic setup when creating an object |
| `self` | This particular object |
| `pass` | Do nothing for now |
| `return` | Give an answer back |
