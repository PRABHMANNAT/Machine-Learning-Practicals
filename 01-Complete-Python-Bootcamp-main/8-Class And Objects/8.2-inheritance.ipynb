# Inheritance in Python — Explained Like You Are 5

## Goal

Inheritance lets one class receive useful information and actions from another class.

Imagine a parent knows how to ride a bicycle. The child can learn that skill from the parent and can also learn new skills. In Python, a **child class** can reuse the code of a **parent class**.

> **Big idea:** Write common code once in the parent, then let child classes reuse it.

## Important words

| Word | Easy meaning |
|---|---|
| Parent class / Base class | The class that shares its features |
| Child class / Derived class | The class that receives those features |
| Inherit | Receive attributes and methods from another class |
| Override | Replace a parent's method with a child's special version |
| `super()` | Ask the parent class to do its part |
| MRO | The order Python follows when searching parent classes |

## 1. The smallest inheritance example

`Child(Parent)` means: **Child inherits from Parent.**

The child below does not define `say_hello()`, but it can still use that method because the parent owns it.

````python
class Parent:
    def say_hello(self):
        print("Hello from the parent!")


class Child(Parent):
    pass


child = Child()
child.say_hello()
````

## 2. Single inheritance: Car and Tesla

**Single inheritance** means one child class has one parent class.

```text
Car
 └── Tesla
```

The `Car` parent stores the features shared by ordinary cars and special cars.

````python
class Car:
    def __init__(self, windows, doors, engine_type):
        self.windows = windows
        self.doors = doors
        self.engine_type = engine_type

    def drive(self):
        print(f"Driving a {self.engine_type} car")


car = Car(4, 5, "petrol")
car.drive()
````

### Using `super()`

Tesla is a special kind of Car. It needs the normal car information plus one extra value: `is_self_driving`.

`super().__init__(...)` says: **Parent Car, please set up the windows, doors, and engine.** Tesla then sets up its own special feature. This avoids repeating the parent's setup code.

````python
class Tesla(Car):
    def __init__(self, windows, doors, engine_type, is_self_driving):
        super().__init__(windows, doors, engine_type)
        self.is_self_driving = is_self_driving

    def show_self_driving(self):
        print(f"Tesla supports self-driving: {self.is_self_driving}")


tesla = Tesla(4, 5, "electric", True)
tesla.show_self_driving()
````

### What did Tesla inherit?

Tesla inherited the `windows`, `doors`, and `engine_type` attributes. It also inherited the `drive()` method. Tesla did not need to rewrite them.

````python
print("Windows:", tesla.windows)
print("Doors:", tesla.doors)
print("Engine:", tesla.engine_type)
tesla.drive()
````

## 3. Method overriding

Sometimes a child needs its own version of a parent's action. This is called **method overriding**.

The parent `Animal` has a general `speak()` method. The child classes replace it with sounds that fit them.

````python
class Animal:
    def speak(self):
        return "The animal makes a sound"


class Dog(Animal):
    def speak(self):
        return "The dog says woof"


class Cat(Animal):
    def speak(self):
        return "The cat says meow"


animal = Animal()
dog = Dog()
cat = Cat()

print(animal.speak())
print(dog.speak())
print(cat.speak())
````

### Extending instead of completely replacing

A child can use `super()` inside an overridden method. This keeps the parent's behavior and then adds something new.

````python
class Worker:
    def work(self):
        print("The worker starts working.")


class Developer(Worker):
    def work(self):
        super().work()
        print("The developer writes Python code.")


developer = Developer()
developer.work()
````

## 4. Multilevel inheritance

**Multilevel inheritance** is like a family chain. A child can receive features from its parent and grandparent.

```text
LivingThing
 └── Animal
      └── Puppy
```

The Puppy below can use methods from both classes above it.

````python
class LivingThing:
    def breathe(self):
        print("I can breathe.")


class Animal(LivingThing):
    def eat(self):
        print("I can eat.")


class Puppy(Animal):
    def play(self):
        print("I can play with a ball.")


puppy = Puppy()
puppy.breathe()  # From LivingThing
puppy.eat()      # From Animal
puppy.play()     # From Puppy
````

## 5. Hierarchical inheritance

**Hierarchical inheritance** means several children share one parent.

```text
Vehicle
 ├── Bike
 └── Bus
```

Both children can use the parent's `start()` method, while each child adds its own action.

````python
class Vehicle:
    def start(self):
        print("The vehicle starts.")


class Bike(Vehicle):
    def ring_bell(self):
        print("Ring ring!")


class Bus(Vehicle):
    def open_doors(self):
        print("The bus doors open.")


bike = Bike()
bus = Bus()

bike.start()
bike.ring_bell()
bus.start()
bus.open_doors()
````

## 6. Multiple inheritance

**Multiple inheritance** means one child has more than one parent.

```text
Animal ──┐
          ├── PetDog
Pet ─────┘
```

`NamedAnimal` gives the dog a name. `Pet` gives the dog an owner. `PetDog` combines both.

````python
class NamedAnimal:
    def __init__(self, name):
        self.name = name


class Pet:
    def __init__(self, owner):
        self.owner = owner


class PetDog(NamedAnimal, Pet):
    def __init__(self, name, owner):
        NamedAnimal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says woof"


buddy = PetDog("Buddy", "Krish")
print(buddy.speak())
print("Owner:", buddy.owner)
````

## 7. Method Resolution Order (MRO)

What if two parents have methods with the same name? Python needs a rule for choosing one.

**MRO** is the order Python follows when searching for a method. In `Duck(Flyer, Swimmer)`, Python checks `Duck`, then `Flyer`, then `Swimmer`, and finally `object`. Because `Flyer` appears first, its `move()` method wins.

> In beginner code, keep multiple inheritance simple. It can become confusing when parents use the same method names.

````python
class Flyer:
    def move(self):
        return "Flying through the air"


class Swimmer:
    def move(self):
        return "Swimming through the water"


class Duck(Flyer, Swimmer):
    pass


duck = Duck()
print(duck.move())
print("Search order:", [item.__name__ for item in Duck.mro()])
````

## 8. Checking inheritance relationships

Python provides two useful helpers:

- `isinstance(object, Class)` asks: **Is this object made from this class or one of its child classes?**
- `issubclass(Child, Parent)` asks: **Does this child class inherit from this parent class?**

````python
print("Is Tesla a child of Car?", issubclass(Tesla, Car))
print("Is tesla a Tesla object?", isinstance(tesla, Tesla))
print("Is tesla also a Car object?", isinstance(tesla, Car))
print("Is an ordinary car a Tesla?", isinstance(car, Tesla))
````

## 9. Real-world example: savings account

A savings account **is a** bank account, so inheritance makes sense. The parent handles deposits. The child adds interest.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. Balance: {self.balance}")


class SavingsAccount(BankAccount):
    def add_interest(self, rate):
        interest = self.balance * rate / 100
        self.balance += interest
        print(f"Added {interest:.2f} interest. Balance: {self.balance:.2f}")


savings = SavingsAccount("Krish", 1000)
savings.deposit(500)       # Inherited from BankAccount
savings.add_interest(10)   # Defined by SavingsAccount
````

## 10. Common mistake: forgetting the parent's setup

If a child writes its own `__init__()` but does not call `super().__init__()`, the parent's attributes are not created automatically.

The next example safely displays the missing-attribute error instead of stopping the notebook.

````python
class Person:
    def __init__(self, name):
        self.name = name


class StudentWithoutSuper(Person):
    def __init__(self, grade):
        self.grade = grade


student = StudentWithoutSuper(5)

try:
    print(student.name)
except AttributeError as error:
    print(f"AttributeError: {error}")
````

### Correct version

Call `super().__init__(name)` so the parent creates `name`, then let the child create `grade`.

````python
class Student(Person):
    def __init__(self, name, grade):
        super().__init__(name)
        self.grade = grade


student = Student("Aarav", 5)
print("Name:", student.name)
print("Grade:", student.grade)
````

## 11. When should you use inheritance?

Use inheritance when the child truly **is a** type of the parent:

- A Tesla **is a** Car. Good use of inheritance.
- A Dog **is an** Animal. Good use of inheritance.
- A SavingsAccount **is a** BankAccount. Good use of inheritance.

Do not use inheritance for a **has a** relationship:

- A Car **has an** Engine. An engine should normally be stored inside a car, not used as the car's parent.

Also avoid very deep inheritance chains. They make it difficult to discover where behavior comes from.

## 12. Mini practice example

Here, both `Teacher` and `SchoolStudent` inherit a name from `SchoolMember`, then add their own information and behavior. Try changing the names and subjects.

````python
class SchoolMember:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"Hello, I am {self.name}."


class Teacher(SchoolMember):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def teach(self):
        return f"I teach {self.subject}."


class SchoolStudent(SchoolMember):
    def __init__(self, name, grade):
        super().__init__(name)
        self.grade = grade

    def study(self):
        return f"I study in grade {self.grade}."


teacher = Teacher("Meera", "Python")
school_student = SchoolStudent("Riya", 6)

print(teacher.introduce(), teacher.teach())
print(school_student.introduce(), school_student.study())
````

## Final recap

| Type | Shape | Meaning |
|---|---|---|
| Single | `A → B` | One child inherits from one parent |
| Multilevel | `A → B → C` | A child inherits through a family chain |
| Hierarchical | `A → B` and `A → C` | Several children share one parent |
| Multiple | `A + B → C` | One child inherits from multiple parents |

### Remember these four ideas

1. Inheritance helps us **reuse code**.
2. A child can use the parent's attributes and methods.
3. A child can add new features or override old behavior.
4. `super()` lets the child ask the parent to do its part.

> **One-sentence summary:** Inheritance lets a new class reuse an older class's information and abilities while adding or changing features of its own.
