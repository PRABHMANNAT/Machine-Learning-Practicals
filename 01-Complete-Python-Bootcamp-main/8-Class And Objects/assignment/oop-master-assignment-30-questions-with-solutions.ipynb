## Question 3 — Rectangle Calculator

**Difficulty:** 1/3  
**Concept:** Instance methods

### Task

Create a `Rectangle` class that stores width and height. Add `area()` and `perimeter()` methods. Test a rectangle measuring 4 by 5.

> **Hint:** Area is `width * height`; perimeter is `2 * (width + height)`.

**Solution idea:** Different methods can use the same object attributes for different calculations.

```python
# Solution 3

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


rectangle = Rectangle(4, 5)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

## Question 4 — Safe Bank Account

**Difficulty:** 2/3  
**Concept:** State-changing methods and validation

### Task

Create a `BankAccount` with an owner and balance. Support positive deposits and withdrawals. Reject non-positive amounts and withdrawals larger than the balance.

> **Hint:** Validate the amount before changing `self.balance`.

**Solution idea:** The object protects its own balance by allowing changes only through checked methods.

```python
# Solution 4

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            return "Deposit must be positive."
        self.balance += amount
        return f"Deposited {amount}. Balance: {self.balance}"

    def withdraw(self, amount):
        if amount <= 0:
            return "Withdrawal must be positive."
        if amount > self.balance:
            return "Insufficient funds."
        self.balance -= amount
        return f"Withdrew {amount}. Balance: {self.balance}"


account = BankAccount("Krish", 1000)
print(account.deposit(500))
print(account.withdraw(300))
print(account.withdraw(5000))
```

## Question 6 — Build a Student from Text

**Difficulty:** 2/3  
**Concept:** Class method and static method

### Task

Create a `Result` class with `name` and `marks`. Add `from_string()` to create an object from text such as `Riya-88`. Add `is_passing()` that returns whether marks are at least 40.

> **Hint:** Use `@classmethod` for the alternative constructor and `@staticmethod` for logic that needs no object data.

**Solution idea:** The class method creates an object; the static method is a related utility function.

```python
# Solution 6

class Result:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @classmethod
    def from_string(cls, text):
        name, marks = text.split("-")
        return cls(name, int(marks))

    @staticmethod
    def is_passing(marks):
        return marks >= 40


result = Result.from_string("Riya-88")
print(result.name, result.marks)
print("Passing:", Result.is_passing(result.marks))
```

## Question 7 — Validated Product Price

**Difficulty:** 2/3  
**Concept:** Property and encapsulation

### Task

Create a `Product` whose `price` property rejects negative values. Demonstrate one valid update and safely demonstrate an invalid update.

> **Hint:** Store the real value in `_price` and validate inside `@price.setter`.

**Solution idea:** The property looks like a normal attribute but runs validation behind the scenes.

```python
# Solution 7

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative.")
        self._price = value


product = Product("Book", 500)
product.price = 450
print("Updated price:", product.price)

try:
    product.price = -10
except ValueError as error:
    print(f"ValueError: {error}")
```

## Question 17 — Polymorphic Payment System

**Difficulty:** 3/3  
**Concept:** Abstract class and polymorphism

### Task

Create an abstract `PaymentMethod` with `pay(amount)`. Implement Card and UPI children, then process both using one checkout function.

> **Hint:** Decorate `pay()` with `@abstractmethod`.

**Solution idea:** The abstract class enforces one interface while each child supplies different behavior.

```python
# Solution 17

from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid {amount} using a card"


class UPIPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid {amount} using UPI"


def checkout(method, amount):
    print(method.pay(amount))


checkout(CardPayment(), 500)
checkout(UPIPayment(), 500)
```

## Question 21 — Shape with Two Rules

**Difficulty:** 3/3  
**Concept:** Multiple abstract methods

### Task

Create an abstract `MeasuredShape` requiring `area()` and `perimeter()`. Implement both methods in `MeasuredCircle`.

> **Hint:** A concrete child must implement every required method.

**Solution idea:** Abstract classes can enforce several related rules at once.

```python
# Solution 21

from math import pi


class MeasuredShape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class MeasuredCircle(MeasuredShape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return pi * self.radius**2

    def perimeter(self):
        return 2 * pi * self.radius


measured_circle = MeasuredCircle(3)
print(f"Area: {measured_circle.area():.2f}")
print(f"Perimeter: {measured_circle.perimeter():.2f}")
```

## Question 27 — Add and Subtract Vectors

**Difficulty:** 2/3  
**Concept:** `__add__`, `__sub__`, and `__repr__`

### Task

Create a two-dimensional Vector supporting `+` and `-`. Return new vectors and display them clearly.

> **Hint:** Combine matching `x` and `y` coordinates.

**Solution idea:** Operators should return the useful result type without unexpectedly changing their inputs.

```python
# Solution 27

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x - other.x, self.y - other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"


vector_one = Vector(2, 3)
vector_two = Vector(4, 5)
print(vector_one + vector_two)
print(vector_one - vector_two)
```

## Question 28 — Add Money Safely

**Difficulty:** 3/3  
**Concept:** Type-safe operator overloading

### Task

Create a Money class supporting `+` only when both objects use the same currency. Safely demonstrate a valid addition and a currency mismatch.

> **Hint:** Raise `ValueError` when currencies differ; return `NotImplemented` for unrelated types.

**Solution idea:** Meaningful validation prevents an operator from silently producing nonsense.

```python
# Solution 28

class Money:
    def __init__(self, amount, currency):
        self.amount = amount
        self.currency = currency

    def __add__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        if self.currency != other.currency:
            raise ValueError("Currencies must match.")
        return Money(self.amount + other.amount, self.currency)

    def __repr__(self):
        return f"Money({self.amount}, {self.currency!r})"


wallet = Money(500, "INR") + Money(250, "INR")
print(wallet)

try:
    print(Money(500, "INR") + Money(10, "USD"))
except ValueError as error:
    print(f"ValueError: {error}")
```

## Question 29 — Sort Products by Price

**Difficulty:** 3/3  
**Concept:** `__lt__`, `__eq__`, and sorting

### Task

Create Products that compare by price. Sort a product list and test whether two differently named products with the same price compare equal.

> **Hint:** `sorted()` can use the class's `__lt__` method.

**Solution idea:** Comparison magic methods define what ordering and equality mean for the domain.

```python
# Solution 29

class ComparableProduct:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __lt__(self, other):
        if not isinstance(other, ComparableProduct):
            return NotImplemented
        return self.price < other.price

    def __eq__(self, other):
        if not isinstance(other, ComparableProduct):
            return NotImplemented
        return self.price == other.price

    def __repr__(self):
        return f"ComparableProduct({self.name!r}, {self.price})"


products = [
    ComparableProduct("Bag", 800),
    ComparableProduct("Pen", 20),
    ComparableProduct("Book", 250),
]

print("Sorted:", sorted(products))
print("Equal prices:", ComparableProduct("Pen A", 20) == ComparableProduct("Pen B", 20))
```

## Question 30 — Capstone: Library Borrowing Basket

**Difficulty:** 3/3  
**Concept:** Abstraction, inheritance, polymorphism, magic methods, and operator overloading

### Task

Create an abstract `LibraryItem` requiring `loan_period()`. Implement BookItem and DVDItem with different periods. Create a `BorrowingBasket` that supports adding baskets with `+`, `len()`, iteration, readable representation, and total loan days.

> **Hint:** Give every item the same method interface, then let the basket call it polymorphically.

**Solution idea:** This design combines a common abstract rule, specialized children, polymorphic calls, container behavior, and a meaningful `+` operator.

```python
# Solution 30

from abc import ABC, abstractmethod


class LibraryItem(ABC):
    def __init__(self, title):
        self.title = title

    @abstractmethod
    def loan_period(self):
        pass

    def loan_message(self):
        return f"{self.title}: {self.loan_period()} days"


class BookItem(LibraryItem):
    def loan_period(self):
        return 14


class DVDItem(LibraryItem):
    def loan_period(self):
        return 7


class BorrowingBasket:
    def __init__(self, items=None):
        self.items = list(items or [])

    def __add__(self, other):
        if not isinstance(other, BorrowingBasket):
            return NotImplemented
        return BorrowingBasket(self.items + other.items)

    def __len__(self):
        return len(self.items)

    def __iter__(self):
        return iter(self.items)

    def __repr__(self):
        titles = [item.title for item in self.items]
        return f"BorrowingBasket({titles!r})"

    def total_loan_days(self):
        return sum(item.loan_period() for item in self.items)


reading = BorrowingBasket([BookItem("Python Basics"), BookItem("Clean Code")])
watching = BorrowingBasket([DVDItem("Learning OOP")])
combined_basket = reading + watching

print(combined_basket)
print("Number of items:", len(combined_basket))

for item in combined_basket:
    print(item.loan_message())

print("Total loan days:", combined_basket.total_loan_days())
```
