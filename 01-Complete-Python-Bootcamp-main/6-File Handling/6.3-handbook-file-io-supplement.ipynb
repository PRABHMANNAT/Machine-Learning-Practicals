# File I/O Handbook — Seven Completed Projects 🗃️

Files make data survive. These projects complete every original handbook exercise using disposable files.

```python
from pathlib import Path
import tempfile
root_context=tempfile.TemporaryDirectory(); root=Path(root_context.name)
demo=root/'story.txt'; demo.write_text('Twinkle star\nI learn Python\nDonkey walks\n',encoding='utf-8')
print(demo.read_text(encoding='utf-8'))
```

## 1. Case-insensitive search

```python
text=demo.read_text(encoding='utf-8')
print('twinkle' in text.casefold())
```

## 2. Copy and compare

```python
copy=root/'copy.txt'; copy.write_text(text,encoding='utf-8')
print(demo.read_bytes()==copy.read_bytes())
```

## 3. Find line number containing Python

```python
matches=[n for n,line in enumerate(text.splitlines(),1) if 'python' in line.casefold()]
print(matches)
```

## 4. Generate multiplication tables 2–20

```python
tables=root/'tables'; tables.mkdir()
for n in range(2,21):
    lines=[f'{n} x {i} = {n*i}' for i in range(1,11)]
    (tables/f'table_{n}.txt').write_text('\n'.join(lines)+'\n',encoding='utf-8')
print(len(list(tables.glob('table_*.txt'))))
```

## 5. Censor several words

```python
def censor(text,words):
    import re
    pattern=re.compile('|'.join(map(re.escape,words)),re.IGNORECASE)
    return pattern.sub('#####',text)
print(censor(text,['donkey','python']))
```

## 6. Update a high score only when larger

```python
score_file=root/'high_score.txt'; score_file.write_text('80',encoding='utf-8')
def update_high_score(path,new_score):
    old=int(path.read_text(encoding='utf-8')) if path.exists() else 0
    if new_score>old: path.write_text(str(new_score),encoding='utf-8')
    return max(old,new_score)
print(update_high_score(score_file,95),score_file.read_text())
```

## 7. Rename after safety checks

```python
old=root/'old.txt'; old.write_text('safe',encoding='utf-8'); new=root/'new.txt'
if old.is_file() and not new.exists() and old.resolve().is_relative_to(root.resolve()):
    old.rename(new)
print(new.exists())
root_context.cleanup()
```

## Key concepts summary

Every project separates reading, transformation, validation, and writing; temporary paths protect real data.

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Small text | `Path.read_text/write_text` |
| Stream | `with path.open() as file:` |
| Compare exact | `read_bytes()` |
| Safe rename | validate then `rename()` |

## Common mistakes and interview tips

- Preserve originals for risky transformations.
- Normalize only when the problem calls for it.
- Report line numbers for bad records.
- Validate targets before rename/delete.

**Revision habit:** explain what each line does, predict the result, run it, and test one edge case.
