### Q1. Count lines, words, and characters in example.txt.

**Concept tested:** streaming

**Question:** Count lines, words, and characters in example.txt.

**Hint:** Loop once over the file.

**Solution approach:** Update three counters while streaming.

**Step-by-step explanation:** Update three counters while streaming. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Streaming bounds memory.

```python
from pathlib import Path
p=Path.cwd().parent/'example.txt'; lines=words=chars=0
with p.open(encoding='utf-8') as f:
 for line in f: lines+=1; words+=len(line.split()); chars+=len(line)
print(lines,words,chars)
```

### Q8. Copy and compare SHA-256.

**Concept tested:** copy verification

**Question:** Copy and compare SHA-256.

**Hint:** Use copy2 and file_digest.

**Solution approach:** Copy to temp and compare.

**Step-by-step explanation:** Copy to temp and compare. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Hashes compare content.

```python
import tempfile,shutil,hashlib
from pathlib import Path
s=Path.cwd().parent/'example.txt'
with tempfile.TemporaryDirectory() as d:
 t=Path(d)/'copy'; shutil.copy2(s,t)
 def h(p):
  with p.open('rb') as f: return hashlib.file_digest(f,'sha256').hexdigest()
 print(h(s)==h(t))
```

### Q9. Copy binary data in two-byte chunks.

**Concept tested:** chunks

**Question:** Copy binary data in two-byte chunks.

**Hint:** Loop until empty bytes.

**Solution approach:** Stream source to temp output.

**Step-by-step explanation:** Stream source to temp output. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Chunking controls memory.

```python
import tempfile
from pathlib import Path
s=Path.cwd().parent/'example.bin'
with tempfile.TemporaryDirectory() as d:
 t=Path(d)/'copy.bin'
 with s.open('rb') as a,t.open('wb') as b:
  while chunk:=a.read(2): b.write(chunk)
 print(s.read_bytes()==t.read_bytes())
```

### Q14. Group files with matching hashes.

**Concept tested:** duplicates

**Question:** Group files with matching hashes.

**Step-by-step explanation:** Create files and show duplicate groups. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Hashes identify equal content.

```python
import tempfile,hashlib
from pathlib import Path
with tempfile.TemporaryDirectory() as d:
 r=Path(d)
 for n,x in [('a','same'),('b','same'),('c','other')]: (r/n).write_text(x)
 g={}
 for p in r.iterdir(): g.setdefault(hashlib.sha256(p.read_bytes()).hexdigest(),[]).append(p.name)
 print([sorted(v) for v in g.values() if len(v)>1])
```
