# File Paths and Directories — Addresses for Files 🗺️

## Goal

A relative path starts from the current working directory; an absolute path begins from a filesystem root. Use `pathlib` for new code and understand `os.path` for existing code.

```python
import os
from pathlib import Path
print('cwd:',os.getcwd())
print('absolute:',Path('example.txt').resolve())
print('exists/file/dir:',Path('example.txt').exists(),Path('example.txt').is_file(),Path('.').is_dir())
print(os.path.join('folder','file.txt'))
```

## 1. Useful `Path` pieces

`/` joins paths portably. `.name`, `.stem`, `.suffix`, `.parent`, `.resolve`, `.exists`, `.is_file`, and `.is_dir` inspect them.

```python
path=Path('reports')/'sales.csv'
print(path,path.name,path.stem,path.suffix,path.parent)
print(Path('example1.txt').exists())
```

## 2. List and search

`iterdir()` lists one level; `glob` matches one level/pattern; `rglob` searches recursively. Sort when stable output matters.

```python
print([p.name for p in sorted(Path('.').iterdir())])
print([p.name for p in sorted(Path('.').glob('*.txt'))])
```

## 3. Create, rename, move, copy, delete—safely

The original `os.mkdir("package")` would fail when repeated. `mkdir(exist_ok=True)` is repeatable. All mutations below occur in a verified temporary directory.

```python
import tempfile,shutil
with tempfile.TemporaryDirectory() as folder:
    root=Path(folder).resolve()
    source=root/'folder'; source.mkdir(parents=True,exist_ok=True)
    file=source/'old.txt'; file.write_text('hello',encoding='utf-8')
    renamed=file.rename(source/'new.txt')
    copied=shutil.copy2(renamed,root/'copy.txt')
    moved=shutil.move(str(copied),str(root/'moved.txt'))
    Path(moved).unlink()
    shutil.rmtree(source)
    print('remaining:',list(root.iterdir()))
```

## 4. Path safety

Resolve targets before destructive work, confirm they are inside the intended root, reject root/home/workspace-wide targets, avoid untrusted `..`, and prefer recoverable moves/backups. `Path.resolve()` normalizes the path; `Path.is_relative_to(root)` checks containment in modern Python.

## Key concepts summary

Paths are addresses; `Path` joins and inspects them portably. Resolve and validate before destructive actions.

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Join | `Path(folder) / "file.txt"` |
| Absolute | `path.resolve()` |
| Create dirs | `mkdir(parents=True,exist_ok=True)` |
| List/search | `iterdir()`, `glob()`, `rglob()` |
| Rename/delete | `rename()`, `unlink()`, `rmdir()` |
| Copy/move trees | `shutil` |

## Common mistakes and interview tips

- Do not assume the current directory.
- Never concatenate separators manually.
- `mkdir` fails if present unless `exist_ok=True`.
- Verify resolved targets before deletion/move.
- Handle `FileNotFoundError`, `PermissionError`, and collisions.
