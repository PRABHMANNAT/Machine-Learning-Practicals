# 1. What Is Probability?

The transcript defines probability as determining the **likelihood of an event or experiment**.

A more precise beginner-friendly definition is:

> **Probability measures how likely an event is to occur.**

If all basic outcomes are equally likely:

\[
P(A)
=
\frac{\text{number of favorable outcomes}}
{\text{total number of possible outcomes}}
\]

### Example: fair coin

Possible outcomes:

\[
S=\{H,T\}
\]

Probability of heads:

\[
P(H)=\frac{1}{2}=0.5=50\%
\]

Probability of tails:

\[
P(T)=\frac{1}{2}=0.5=50\%
\]

## 1.1 Four Words You Must Know

### 1. Experiment
A process that produces an outcome.

Examples:
- toss a coin
- roll a die
- pick a card

### 2. Outcome
One possible result.

Example:
- coin → `Heads`
- die → `4`

### 3. Sample Space
The set of **all possible outcomes**.

Coin:

\[
S=\{H,T\}
\]

Die:

\[
S=\{1,2,3,4,5,6\}
\]

### 4. Event
A collection of one or more outcomes that we care about.

Example:

\[
A=\{\text{rolling an even number}\}
=\{2,4,6\}
\]

## 1.2 Dice Example

For a fair six-sided die:

\[
S=\{1,2,3,4,5,6\}
\]

Probability of rolling `1`:

\[
P(X=1)=\frac{1}{6}
\]

Probability of rolling `2`:

\[
P(X=2)=\frac{1}{6}
\]

Every face has equal probability because the die is assumed to be fair.

````python
from fractions import Fraction

p_one = Fraction(1, 6)
print("P(rolling 1) =", p_one)
print("Decimal =", float(p_one))
print("Percent =", float(p_one) * 100)
````

## 1.3 The Basic Formula Is Not Universal

The formula

\[
\frac{\text{favorable}}{\text{total}}
\]

works directly when outcomes are **equally likely**.

Example: fair die ✅

But imagine a biased coin:

```text
P(H) = 0.8
P(T) = 0.2
```

There are still two outcomes, but:

\[
P(H)\neq\frac12
\]

So in real statistics, probabilities may come from:

- symmetry/equally likely outcomes
- empirical data
- models
- known distributions

# 2. Probability Notation

Let:

- \(A\) = event A
- \(B\) = event B

Common notation:

\[
P(A)
\]

means:

> probability that event A happens.

### OR

\[
P(A\cup B)
\]

Read:

> probability of **A OR B**

### AND

\[
P(A\cap B)
\]

Read:

> probability of **A AND B**

### NOT A

\[
P(A^c)
\]

Read:

> probability that A does **not** happen.

## 2.1 OR vs AND — Critical Memory Trick

### OR → think **addition**

Usually start from:

\[
P(A\cup B)
\]

### AND → think **multiplication**

Usually start from:

\[
P(A\cap B)
\]

But there are conditions and overlap corrections, so never blindly add or multiply.

# 3. Complement Rule

The complement of event \(A\) means **A does not happen**.

Because either A happens or it does not:

\[
P(A)+P(A^c)=1
\]

Therefore:

\[
\boxed{P(A^c)=1-P(A)}
\]

### Example

If:

\[
P(\text{rain})=0.3
\]

then:

\[
P(\text{no rain})=1-0.3=0.7
\]

````python
p_rain = 0.30
p_no_rain = 1 - p_rain

print("P(rain)    =", p_rain)
print("P(no rain) =", p_no_rain)
````

# 4. Mutually Exclusive Events

The transcript defines two events as mutually exclusive if they **cannot occur at the same time**.

That means:

\[
A\cap B=\varnothing
\]

and therefore:

\[
P(A\cap B)=0
\]

### Coin example

For one coin toss:

- \(A\) = Heads
- \(B\) = Tails

You cannot get both heads and tails on the same single toss.

So these events are mutually exclusive.

## 4.1 Mutually Exclusive Addition Rule

For mutually exclusive events:

\[
\boxed{P(A\cup B)=P(A)+P(B)}
\]

because there is **no overlap** to subtract.

### Example: Heads OR Tails

\[
P(H\cup T)=P(H)+P(T)
\]

\[
=\frac12+\frac12
\]

\[
=1
\]

This makes sense because a coin toss must result in heads or tails under the simple model.

### Example: Roll 1 OR 5

On one fair die:

\[
P(1)=\frac16
\]

\[
P(5)=\frac16
\]

They cannot occur on the same roll.

Therefore:

\[
P(1\text{ or }5)
=
\frac16+\frac16
=
\frac26
=
\frac13
\]

````python
p_1_or_5 = Fraction(1, 6) + Fraction(1, 6)
print("P(1 or 5) =", p_1_or_5)
print("Decimal =", float(p_1_or_5))
````

# 5. Non-Mutually Exclusive Events

Two events are **non-mutually exclusive** when both can happen at the same time.

Example: draw one card.

Let:

- \(K\) = “card is a king”
- \(H\) = “card is a heart”

The **King of Hearts** belongs to both events.

So:

\[
K\cap H\neq\varnothing
\]

## 5.1 General Addition Rule

If events can overlap:

\[
\boxed{
P(A\cup B)
=
P(A)+P(B)-P(A\cap B)
}
\]

### Why subtract?

Because adding \(P(A)+P(B)\) counts the overlapping part twice.

So we subtract it once.

## 5.2 King OR Heart Example

A standard deck has 52 cards.

- 4 kings
- 13 hearts
- 1 King of Hearts

So:

\[
P(K)=\frac4{52}
\]

\[
P(H)=\frac{13}{52}
\]

\[
P(K\cap H)=\frac1{52}
\]

Therefore:

\[
P(K\cup H)
=
\frac4{52}
+
\frac{13}{52}
-
\frac1{52}
\]

\[
=
\frac{16}{52}
=
\frac4{13}
\]

````python
p_king_or_heart = Fraction(4, 52) + Fraction(13, 52) - Fraction(1, 52)

print("P(King or Heart) =", p_king_or_heart)
print("Decimal =", float(p_king_or_heart))
````

## 5.3 One Formula to Remember

The **general** addition rule is always:

\[
P(A\cup B)
=
P(A)+P(B)-P(A\cap B)
\]

If A and B are mutually exclusive:

\[
P(A\cap B)=0
\]

so the formula automatically becomes:

\[
P(A\cup B)=P(A)+P(B)
\]

This is better than memorizing two totally separate formulas.

# 6. Independent Events

The transcript describes independent events as events that **do not affect one another**.

More precisely:

Events \(A\) and \(B\) are independent when knowing that one happened does not change the probability of the other.

\[
P(B\mid A)=P(B)
\]

### Example: two coin tosses

First toss = Heads.

Does that change the probability of Tail on the second toss?

No.

\[
P(T\text{ on second toss})=\frac12
\]

So the tosses are independent.

## 6.1 Independent Multiplication Rule

For independent events:

\[
\boxed{
P(A\cap B)=P(A)P(B)
}
\]

### Example: Head then Tail

\[
P(H\text{ then }T)
=
\frac12\times\frac12
=
\frac14
\]

````python
p_head_then_tail = Fraction(1, 2) * Fraction(1, 2)

print("P(H then T) =", p_head_then_tail)
print("Percent =", float(p_head_then_tail) * 100)
````

### Example: Roll 1 then Roll 2

If two independent die rolls are made:

\[
P(1\text{ then }2)
=
\frac16\times\frac16
=
\frac1{36}
\]

# 7. Dependent Events

Two events are dependent when the first event changes the probability of the second.

Classic example:

> Draw a king, **do not replace the card**, then draw a queen.

Initially:

\[
52\text{ cards}
\]

After removing one king:

\[
51\text{ cards remain}
\]

So the probability for the second draw changes.

## 7.1 Conditional Probability

The notation:

\[
P(B\mid A)
\]

means:

> probability of B **given that A has already happened**.

Formula:

\[
\boxed{
P(B\mid A)=\frac{P(A\cap B)}{P(A)}
}
\]

as long as \(P(A)>0\).

## 7.2 General Multiplication Rule

For any two events:

\[
\boxed{
P(A\cap B)=P(A)P(B\mid A)
}
\]

This is the general multiplication rule.

If A and B are independent:

\[
P(B\mid A)=P(B)
\]

so it simplifies to:

\[
P(A\cap B)=P(A)P(B)
\]

## 7.3 King Then Queen Without Replacement

First draw:

\[
P(K)=\frac4{52}
\]

After one king is removed:

- cards remaining = 51
- queens remaining = 4

So:

\[
P(Q\mid K)=\frac4{51}
\]

Therefore:

\[
P(K\cap Q)
=
\frac4{52}
\times
\frac4{51}
\]

````python
p_king_then_queen = Fraction(4, 52) * Fraction(4, 51)

print("P(King then Queen without replacement) =", p_king_then_queen)
print("Decimal =", float(p_king_then_queen))
print("Percent =", float(p_king_then_queen) * 100)
````

## 7.4 With Replacement vs Without Replacement

### With replacement

You put the first card back.

Deck size returns to 52.

Events may become independent under the standard random-draw model.

Example:

\[
P(K\text{ then }Q)
=
\frac4{52}
\times
\frac4{52}
\]

### Without replacement

You do not put the card back.

Deck size changes to 51.

Events are dependent.

\[
P(K\text{ then }Q)
=
\frac4{52}
\times
\frac4{51}
\]

````python
with_replacement = Fraction(4, 52) * Fraction(4, 52)
without_replacement = Fraction(4, 52) * Fraction(4, 51)

print("With replacement   :", with_replacement, float(with_replacement))
print("Without replacement:", without_replacement, float(without_replacement))
````

# 8. Important Accuracy Note: Conditional Probability ≠ Bayes' Theorem

The transcript says the dependent-event formula is “also called Bayes theorem” and then introduces conditional probability.

The **idea is connected**, but these are not exactly the same formula.

### Conditional probability

\[
P(B\mid A)
=
\frac{P(A\cap B)}{P(A)}
\]

### Multiplication rule

\[
P(A\cap B)
=
P(A)P(B\mid A)
\]

### Bayes' theorem

\[
\boxed{
P(A\mid B)
=
\frac{P(B\mid A)P(A)}
{P(B)}
}
\]

Bayes' theorem is obtained by combining conditional-probability relationships.

So:

> **Conditional probability is the foundation. Bayes' theorem is a specific rule for reversing a condition.**

# 9. Bayes' Theorem — Advanced but Simple

Bayes helps answer:

> “I observed B. How should that update my belief about A?”

Formula:

\[
P(A\mid B)
=
\frac{P(B\mid A)P(A)}
{P(B)}
\]

Where:

- \(P(A)\) = prior probability
- \(P(B\mid A)\) = likelihood
- \(P(B)\) = evidence / total probability of B
- \(P(A\mid B)\) = posterior probability

## 9.1 Spam Filter Example

Suppose:

- 20% of messages are spam
- 70% of spam messages contain the word “WIN”
- 10% of non-spam messages contain “WIN”

Question:

> If a message contains “WIN”, what is the probability it is spam?

Let:

- \(S\) = spam
- \(W\) = contains “WIN”

Given:

\[
P(S)=0.20
\]

\[
P(W\mid S)=0.70
\]

\[
P(W\mid S^c)=0.10
\]

First calculate:

\[
P(W)
=
P(W\mid S)P(S)
+
P(W\mid S^c)P(S^c)
\]

````python
p_spam = 0.20
p_not_spam = 1 - p_spam

p_win_given_spam = 0.70
p_win_given_not_spam = 0.10

p_win = (
    p_win_given_spam * p_spam
    + p_win_given_not_spam * p_not_spam
)

p_spam_given_win = p_win_given_spam * p_spam / p_win

print("P(WIN) =", round(p_win, 4))
print("P(Spam | WIN) =", round(p_spam_given_win, 4))
````

So seeing “WIN” increases the spam probability from **20%** to about **63.6%** in this toy example.

That is Bayesian updating.

# 10. Law of Total Probability — Useful Extension

If events \(A_1,A_2,\dots,A_k\) divide all possibilities, then:

\[
P(B)
=
\sum_i P(B\mid A_i)P(A_i)
\]

For the spam example:

\[
P(W)
=
P(W\mid S)P(S)
+
P(W\mid S^c)P(S^c)
\]

This denominator often appears inside Bayes' theorem.

# 11. Mutually Exclusive vs Independent — Do NOT Confuse Them

This is a very common interview trap.

### Mutually exclusive

They cannot happen together.

\[
P(A\cap B)=0
\]

### Independent

One does not change the probability of the other.

\[
P(A\cap B)=P(A)P(B)
\]

### Important fact

If two non-zero-probability events are mutually exclusive, they **cannot** be independent.

Why?

If \(P(A)>0\) and \(P(B)>0\):

\[
P(A)P(B)>0
\]

but mutually exclusive requires:

\[
P(A\cap B)=0
\]

Those cannot both be true.

## 11.1 Quick Comparison Table

| Concept | Main idea | Key formula |
|---|---|---|
| Mutually exclusive | Cannot happen together | \(P(A\cap B)=0\) |
| Non-mutually exclusive | Can overlap | use general addition rule |
| Independent | One does not affect the other | \(P(A\cap B)=P(A)P(B)\) |
| Dependent | One changes probability of the other | \(P(A\cap B)=P(A)P(B\mid A)\) |
| Conditional probability | Probability after knowing information | \(P(B\mid A)\) |

# 12. Probability Sanity Rules

Every valid probability must satisfy:

\[
0\le P(A)\le1
\]

The entire sample space has probability:

\[
P(S)=1
\]

Impossible event:

\[
P(\varnothing)=0
\]

Complement:

\[
P(A^c)=1-P(A)
\]

If your calculation gives:

```text
1.4
-0.2
```

for a probability, something went wrong.

# 13. Simulation: Probability by Experiment

Probability can be estimated by repeating an experiment many times.

This is called **empirical probability**.

\[
\hat P(A)
=
\frac{\text{number of times A occurred}}
{\text{number of trials}}
\]

As the number of trials becomes large, the estimate often settles near the theoretical probability.

````python
import numpy as np

rng = np.random.default_rng(42)

for n in [10, 100, 1_000, 100_000]:
    tosses = rng.integers(0, 2, size=n)  # 1=head, 0=tail
    estimated = tosses.mean()
    print(f"{n:>7,} tosses -> estimated P(Head) = {estimated:.4f}")
````

This illustrates the intuition behind the **Law of Large Numbers**:

> With many independent repetitions, observed proportions tend to get closer to the true probability.

## 13.1 Simulate “1 OR 5” on a Die

````python
n = 200_000
rolls = rng.integers(1, 7, size=n)

event = (rolls == 1) | (rolls == 5)
estimated_probability = event.mean()

print("Estimated P(1 or 5):", round(estimated_probability, 4))
print("Theoretical P(1 or 5):", float(Fraction(1, 3)))
````

## 13.2 Simulate Head Then Tail

````python
n = 200_000

first = rng.integers(0, 2, size=n)   # 1 H, 0 T
second = rng.integers(0, 2, size=n)

head_then_tail = (first == 1) & (second == 0)

print("Estimated:", round(head_then_tail.mean(), 4))
print("Theoretical:", 0.25)
````

## 13.3 Simulate King Then Queen Without Replacement

Instead of trusting only the formula, we can simulate actual deck draws.

````python
import random

# rank labels only; suits are not needed for this experiment
deck = (
    ["K"] * 4
    + ["Q"] * 4
    + ["Other"] * 44
)

trials = 100_000
success = 0

random.seed(42)

for _ in range(trials):
    draw = random.sample(deck, 2)  # without replacement
    if draw[0] == "K" and draw[1] == "Q":
        success += 1

estimated = success / trials
theoretical = float(Fraction(4, 52) * Fraction(4, 51))

print("Estimated :", round(estimated, 5))
print("Theoretical:", round(theoretical, 5))
````

# 14. Probability Tree Thinking

For multi-step problems, draw a tree.

Example: two coin tosses.

```text
Start
├── H (1/2)
│   ├── H (1/2) → HH = 1/4
│   └── T (1/2) → HT = 1/4
└── T (1/2)
    ├── H (1/2) → TH = 1/4
    └── T (1/2) → TT = 1/4
```

### Rule

- **Multiply along a path**
- **Add separate successful paths**

Example:

Probability of exactly one head:

Successful paths:

```text
HT
TH
```

\[
P(\text{exactly one H})
=
\frac14+\frac14
=
\frac12
\]

````python
outcomes = {
    "HH": 1/4,
    "HT": 1/4,
    "TH": 1/4,
    "TT": 1/4,
}

p_exactly_one_head = outcomes["HT"] + outcomes["TH"]
print("P(exactly one head) =", p_exactly_one_head)
````

# 15. More Than Two Independent Events

If \(A,B,C\) are independent:

\[
P(A\cap B\cap C)
=
P(A)P(B)P(C)
\]

Example: three heads in three fair coin tosses:

\[
P(HHH)
=
\left(\frac12\right)^3
=
\frac18
\]

````python
p_three_heads = Fraction(1, 2) ** 3
print("P(HHH) =", p_three_heads)
````

# 16. “At Least One” Problems

Often the easiest method is the complement.

Example:

> What is the probability of getting **at least one head** in 3 fair tosses?

Instead of counting HHH, HHT, HTH, etc., calculate:

\[
P(\text{at least one H})
=
1-P(\text{no heads})
\]

“No heads” means all tails:

\[
P(TTT)=\left(\frac12\right)^3=\frac18
\]

Therefore:

\[
P(\text{at least one H})
=
1-\frac18
=
\frac78
\]

````python
p_at_least_one_head = 1 - Fraction(1, 2) ** 3
print("P(at least one head in 3 tosses) =", p_at_least_one_head)
````

# 17. Probability in Machine Learning

The transcript specifically connects probability to **classification algorithms** and thresholds.

A classifier may output:

```text
P(spam) = 0.91
P(not spam) = 0.09
```

Then a decision threshold might be:

```text
if P(spam) >= 0.50:
    predict spam
else:
    predict not spam
```

### Important idea

The model's probability and the final class label are not the same thing.

Probability:

```text
0.73
```

Decision:

```text
Class 1
```

Thresholding converts a probability into a class decision.

````python
probabilities = np.array([0.20, 0.49, 0.51, 0.73, 0.95])
threshold = 0.50

predicted_class = (probabilities >= threshold).astype(int)

for p, label in zip(probabilities, predicted_class):
    print(f"P(class=1)={p:.2f} -> predicted class {label}")
````

## 17.1 Why Change the Threshold?

Threshold `0.5` is common, but not mandatory.

Examples:

- spam detection
- fraud detection
- defect detection
- churn prediction

A business may choose `0.7`, `0.3`, or another threshold depending on the cost of false positives and false negatives.

This is a practical reason probability matters in ML.

# 18. Connection to Naive Bayes

The transcript mentions Naive Bayes.

Naive Bayes uses Bayes' theorem:

\[
P(C\mid X)
\propto
P(X\mid C)P(C)
\]

where:

- \(C\) = class
- \(X\) = observed features

The “naive” assumption is that features are conditionally independent given the class.

You do **not** need to master the full classifier yet.

For now remember:

> **Conditional probability + Bayes' theorem are mathematical foundations behind Naive Bayes.**

# 19. Common Probability Mistakes

## Mistake 1 — Adding when the question says AND

Wrong:

\[
P(H\text{ AND }T)=P(H)+P(T)
\]

For independent sequential tosses, use multiplication.

---

## Mistake 2 — Multiplying when the question says OR

For “A OR B”, start with the addition rule.

---

Wrong for overlapping events:

\[
P(A\cup B)=P(A)+P(B)
\]

Correct:

\[
P(A\cup B)=P(A)+P(B)-P(A\cap B)
\]

---

## Mistake 4 — Assuming independence

Without replacement, card draws are usually dependent.

---

## Mistake 5 — Confusing mutually exclusive with independent

These are different concepts.

---

## Mistake 6 — Forgetting the denominator changes

After drawing a card without replacement:

```text
52 → 51 → 50 → ...
```

---

## Mistake 7 — Treating conditional probability as Bayes' theorem

They are related, but not identical formulas.

---

## Mistake 8 — Probability above 1

Always sanity-check:

\[
0\le P\le1
\]

# 20. Final Practical Mini Project — E-commerce Conversion Probability

Imagine an online store.

We observe 20,000 sessions.

For every session we track:

- whether the visitor came from an ad
- whether the visitor used mobile
- whether they purchased

We will use probability to answer:

1. What is \(P(\text{purchase})\)?
2. What is \(P(\text{ad OR mobile})\)?
3. What is \(P(\text{ad AND purchase})\)?
4. What is \(P(\text{purchase}\mid \text{ad})\)?
5. Are `ad` and `purchase` approximately independent?
6. How can this help business decisions?

````python
import pandas as pd

rng = np.random.default_rng(123)
n = 20_000

ad = rng.random(n) < 0.35
mobile = rng.random(n) < 0.65

# Purchase probability depends on traffic source and device
purchase_probability = (
    0.04
    + 0.05 * ad
    + 0.02 * (~mobile)
)

purchase = rng.random(n) < purchase_probability

df = pd.DataFrame({
    "ad": ad,
    "mobile": mobile,
    "purchase": purchase
})

df.head()
````

## 20.1 Probability of Purchase

````python
p_purchase = df["purchase"].mean()
print("P(purchase) =", round(p_purchase, 4))
````

## 20.2 Addition Rule: Ad OR Mobile

\[
P(A\cup M)
=
P(A)+P(M)-P(A\cap M)
\]

````python
p_ad = df["ad"].mean()
p_mobile = df["mobile"].mean()
p_ad_and_mobile = (df["ad"] & df["mobile"]).mean()

p_ad_or_mobile_formula = p_ad + p_mobile - p_ad_and_mobile
p_ad_or_mobile_direct = (df["ad"] | df["mobile"]).mean()

print("Using formula:", round(p_ad_or_mobile_formula, 4))
print("Direct count :", round(p_ad_or_mobile_direct, 4))
````

## 20.3 AND Event: Ad AND Purchase

# 22. Formula Cheat Sheet

## Basic equally likely outcomes

\[
P(A)
=
\frac{\text{favorable outcomes}}
{\text{total outcomes}}
\]

## Complement

\[
P(A^c)=1-P(A)
\]

## General addition rule

\[
P(A\cup B)
=
P(A)+P(B)-P(A\cap B)
\]

## Mutually exclusive special case

\[
P(A\cup B)
=
P(A)+P(B)
\]

because:

\[
P(A\cap B)=0
\]

## Conditional probability

\[
P(B\mid A)
=
\frac{P(A\cap B)}{P(A)}
\]

## General multiplication rule

\[
P(A\cap B)
=
P(A)P(B\mid A)
\]

## Independent multiplication rule

\[
P(A\cap B)
=
P(A)P(B)
\]

## Bayes' theorem

\[
P(A\mid B)
=
\frac{
P(B\mid A)P(A)
}{
P(B)
}
\]

# 24. Interview Questions

### Q1. What is probability?
A measure of the likelihood that an event occurs.

### Q2. What is a sample space?
The set of all possible outcomes of an experiment.

### Q3. What are mutually exclusive events?
Events that cannot occur simultaneously.

### Q4. Addition rule for mutually exclusive events?
\[
P(A\cup B)=P(A)+P(B)
\]

### Q5. General addition rule?
\[
P(A\cup B)=P(A)+P(B)-P(A\cap B)
\]

### Q6. Independent vs dependent events?
Independent events do not alter one another's probability; dependent events do.

### Q7. Multiplication rule for independent events?
\[
P(A\cap B)=P(A)P(B)
\]

### Q8. General multiplication rule?
\[
P(A\cap B)=P(A)P(B\mid A)
\]

### Q9. What does \(P(B\mid A)\) mean?
Probability of B given that A has occurred.

### Q10. Mutually exclusive vs independent?
Mutually exclusive means cannot occur together. Independent means occurrence of one does not change the probability of the other.

### Q11. Is conditional probability the same as Bayes' theorem?
No. Bayes' theorem uses conditional probabilities to reverse conditioning.

### Q12. Why does without-replacement sampling create dependence?
Because the population/deck composition changes after each draw.

### Q13. Where is probability used in ML?
Classification probabilities, probabilistic models, Naive Bayes, uncertainty estimates, threshold-based decisions, and more.

# 27. Final Mental Model

Think of probability problems like this:

```text
1. What is the experiment?
2. What is the sample space?
3. What event am I interested in?
4. Is the question saying OR or AND?
5. If OR:
      Can both happen together?
6. If AND:
      Does the first event change the second?
7. Is there a "given that..." condition?
8. Do I need complement?
9. Sanity check: answer must be between 0 and 1.
```

## One sentence to remember

> **OR usually leads you toward addition, AND usually leads you toward multiplication, and “given” leads you toward conditional probability.**
