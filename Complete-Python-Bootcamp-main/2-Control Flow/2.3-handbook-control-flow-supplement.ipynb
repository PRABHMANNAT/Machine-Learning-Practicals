# Handbook Supplement: Decisions and Loops

A program becomes useful when it can choose and repeat. This note collects the handbook's conditional and loop patterns with safer examples.

## Conditions and an elif ladder

`if` runs a block when its condition is true. `elif` means "otherwise, try this condition". The ladder stops at the first true branch; `else` runs only when all tests fail.

```python
marks = 72
if marks >= 90:
    grade = "A+"
elif marks >= 60:
    grade = "B"
else:
    grade = "Needs practice"
```

Use comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) and combine conditions with `and`, `or`, and `not`.

````python
marks = 72
if marks >= 90:
    grade = "A+"
elif marks >= 60:
    grade = "B"
elif marks >= 40:
    grade = "C"
else:
    grade = "F"
print(grade)
````

## While, for, range, and loop control

Use `while` when you repeat until a condition changes. Update the value that controls the loop. Use `for` when you want one turn for every item in an iterable. `range(start, stop, step)` excludes `stop`.

- `break` exits the loop immediately.
- `continue` skips the rest of this turn.
- `pass` is a deliberate no-op placeholder.
- A loop's `else` runs when the loop finishes normally, not when `break` stops it.

````python
count = 1
while count <= 3:
    print("while:", count)
    count += 1

for number in range(1, 6, 2):
    print("for:", number)
else:
    print("for loop finished normally")
````

## Patterns worth practicing

```python
# Multiplication table
number = 7
for multiplier in range(1, 11):
    print(f"{number} x {multiplier} = {number * multiplier}")

# Search with for-else
for item in [2, 4, 6]:
    if item == 5:
        print("found")
        break
else:
    print("not found")
```
