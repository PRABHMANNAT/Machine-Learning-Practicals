# Control Flow 1 — Conditional Statements: Python Makes Choices 🚦

## Goal

A condition is a yes/no question. Python uses the answer to choose which path to follow. You will learn `if`, `elif`, `else`, nesting, truthiness, compound conditions, membership, conditional expressions, `match`, validation patterns, and common mistakes.

## 1. How a decision works

```text
ask a True/False question
        ├── True  → run the indented block
        └── False → skip it or use another branch
```

The colon opens a block; four spaces show what belongs inside.

## 2. `if`: run something only when the question is true

The original voting example is preserved below. Change `age` and predict whether the message appears.

````python
age = 20

if age >= 18:
    print("You are allowed to vote in the elections")
````

The expression `age >= 18` itself is a Boolean value:

````python
print(age >= 18)
````

## 3. `if`/`else`: exactly one of two paths

`else` means “if the earlier condition was false, do this instead.” It never has its own condition.

````python
age = 16

if age >= 18:
    print("You are eligible for voting")
else:
    print("You are a minor")
````

## 4. `elif`: choose the first matching path

Python checks from top to bottom and stops at the **first true branch**. Order conditions from the most specific/strict to the more general when ranges overlap.

````python
age = 17

if age < 13:
    print("You are a child")
elif age < 18:
    print("You are a teenager")
else:
    print("You are an adult")
````

### Boundary values matter

Test the edges: just below, exactly at, and just above a boundary. Here, ages `12`, `13`, `17`, and `18` reveal every handoff.

````python
for test_age in [12, 13, 17, 18]:
    if test_age < 13:
        group = "child"
    elif test_age < 18:
        group = "teenager"
    else:
        group = "adult"
    print(test_age, "->", group)
````

## 5. Combine conditions with `and`, `or`, and `not`

Use parentheses when they make the question easier to read. Chained comparisons are perfect for ranges.

````python
age = 20
has_id = True
is_banned = False

if age >= 18 and has_id and not is_banned:
    print("Entry allowed")
else:
    print("Entry denied")

temperature = 22
if 18 <= temperature <= 25:
    print("Comfortable temperature")
````

## 6. Truthiness: conditions do not need `== True`

Empty strings/containers, zero, `None`, and `False` are falsy. Most other values are truthy.

````python
username = ""

if username:
    print(f"Welcome, {username}")
else:
    print("Please enter a username")
````

## 7. Membership makes choices clearer

Instead of many `or` comparisons, ask whether a value is `in` a collection.

````python
day = "Saturday"

if day in {"Saturday", "Sunday"}:
    print("It is the weekend")
else:
    print("It is a weekday")
````

## 8. Nested conditionals: a question inside a question

Nesting is useful when the second question only makes sense after the first. Avoid very deep nests—combined conditions or early exits in functions are often clearer.

````python
# Original positive/even/odd example, made reproducible
number = 7  # Interactive alternative: int(input("Enter the number: "))

if number > 0:
    print("The number is positive")
    if number % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
elif number == 0:
    print("The number is zero")
else:
    print("The number is negative")
````

## 9. Leap-year example

A leap year is divisible by 400, **or** divisible by 4 but not by 100. The original nested idea is preserved, followed by a flatter equivalent.

````python
year = 2024

# Original nested style
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            nested_result = "leap year"
        else:
            nested_result = "not a leap year"
    else:
        nested_result = "leap year"
else:
    nested_result = "not a leap year"

print(year, "is", nested_result)

# Flatter equivalent
is_leap = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
print("Compact check:", is_leap)
````

## 10. Conditional expression: a tiny one-line choice

Use `value_if_true if condition else value_if_false` only for a **small, readable** choice. It produces a value.

````python
number = 27
kind = "even" if number % 2 == 0 else "odd"
print(kind)
````

## 11. `match`/`case`: match one value against patterns

Available in Python 3.10+. It is helpful when comparing one subject with several exact shapes. `_` is the catch-all case.

````python
command = "start"

match command:
    case "start":
        print("Starting the game")
    case "pause":
        print("Pausing the game")
    case "quit" | "exit":
        print("Leaving the game")
    case _:
        print("Unknown command")
````

## 12. Calculator — original scenario, safer structure

The user-input lines are shown as comments; fixed sample values make the notebook reproducible. We check for division by zero before dividing.

````python
# num1 = float(input("Enter first number: "))
# num2 = float(input("Enter second number: "))
# operation = input("Enter operation (+, -, *, /): ")
num1, num2, operation = 12.0, 4.0, "/"

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/" and num2 != 0:
    result = num1 / num2
elif operation == "/":
    result = "Cannot divide by zero"
else:
    result = "Unknown operation"

print("Result:", result)
````

## 13. Ticket price — original scenario

Notice that student status matters only in selected age groups, so a small nested condition is reasonable.

````python
age = 16
is_student = "yes"

if age < 5:
    price = 0
elif age <= 12:
    price = 10
elif age <= 17:
    price = 12 if is_student == "yes" else 15
elif age <= 64:
    price = 20 if is_student == "yes" else 25
else:
    price = 10

print("Ticket price:", "Free" if price == 0 else f"${price}")
````

## 14. Employee bonus — original scenario

First choose a percentage from rating and service, then calculate money. `:.2f` displays exactly two decimal places.

````python
salary = 50_000
years_of_service = 8
performance_rating = 4.7

if performance_rating >= 4.5:
    if years_of_service > 10:
        bonus_percentage = 20
    elif years_of_service > 5:
        bonus_percentage = 15
    else:
        bonus_percentage = 10
elif performance_rating >= 3.5:
    bonus_percentage = 10 if years_of_service > 5 else 5
else:
    bonus_percentage = 0

bonus_amount = salary * bonus_percentage / 100
print(f"Bonus Percentage: {bonus_percentage}%")
print("Bonus Amount: ${:.2f}".format(bonus_amount))
````

## 15. Login check — original scenario

Real systems never store plain-text passwords like this; they use secure password hashing. This toy example is only for practicing decisions.

````python
stored_username = "admin"
stored_password = "password123"
username = "admin"
password = "password123"

if username == stored_username:
    if password == stored_password:
        print("Login successful!")
    else:
        print("Incorrect password.")
else:
    print("Username not found.")
````

## 16. Common mistakes and debugging checklist

- Missing `:` after `if`, `elif`, or `else`.
- Wrong indentation.
- Using `=` instead of `==` in a condition.
- Writing independent `if` statements when only one branch should run.
- Putting a broad condition before a narrower one.
- Forgetting boundary values (`<` versus `<=`).
- Comparing input text to a number without conversion.
- Writing `if condition == True:` when `if condition:` is clearer.

Trace a decision by printing the values and individual Boolean parts.

````python
age = 17
has_permission = True
print("age >= 18:", age >= 18)
print("has_permission:", has_permission)
print("combined:", age >= 18 or has_permission)
````

## Checks — test boundaries, not only happy paths

Try values on both sides of every boundary. For the leap-year rule, useful cases are `1900`, `2000`, `2023`, and `2024`.

````python
for test_year in [1900, 2000, 2023, 2024]:
    is_leap = test_year % 400 == 0 or (
        test_year % 4 == 0 and test_year % 100 != 0
    )
    print(test_year, "->", "leap" if is_leap else "not leap")
````

## Easy revision cheat sheet

| Need | Pattern |
|---|---|
| One optional path | `if condition:` |
| Two paths | `if ... else ...` |
| Many exclusive paths | `if ... elif ... else ...` |
| Question inside a valid path | nested `if` |
| Combine questions | `and`, `or`, `not` |
| Range | `low <= value <= high` |
| Container check | `item in collection` |
| Tiny value choice | `yes_value if condition else no_value` |
| Exact pattern choices | `match subject: case ...` |

**Golden rules:** conditions become truthy/falsy; Python chooses the first matching `if`/`elif` branch; indentation owns the block; test boundary values.
