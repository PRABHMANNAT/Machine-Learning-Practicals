# Data Analysis Assignment Matplotlib and Seaborn

This notebook contains 30 practice questions with runnable solutions: 15 for Matplotlib and 15 for Seaborn. Run the setup cell first, then try each task before running its solution. All examples use a reproducible synthetic employee-performance dataset, so no download is required.

**Learning goals:** choose appropriate charts, customize readable figures, compare groups, and explore relationships in data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(2026)
n = 180
departments = rng.choice(['Sales', 'Support', 'Engineering', 'HR'], n, p=[0.30, 0.27, 0.28, 0.15])
regions = rng.choice(['North', 'South', 'East'], n)
experience = rng.integers(0, 16, n)
training_hours = rng.integers(4, 41, n)
department_bonus = pd.Series(departments).map({'Sales': 7, 'Support': 2, 'Engineering': 9, 'HR': 4}).to_numpy()
score = np.clip(52 + 1.7 * experience + 0.45 * training_hours + department_bonus + rng.normal(0, 9, n), 35, 100)
salary = 28000 + 2300 * experience + 210 * training_hours + department_bonus * 450 + rng.normal(0, 5000, n)
revenue = np.clip((score - 45) * rng.uniform(600, 1300, n), 2000, None)
employees = pd.DataFrame({'department': departments, 'region': regions, 'experience': experience,
                          'training_hours': training_hours, 'score': score,
                          'salary': salary, 'revenue': revenue})
months = pd.date_range('2026-01-01', periods=12, freq='MS')
monthly_sales = pd.DataFrame({'month': months,
                              'sales': 42000 + np.arange(12) * 1800 + rng.normal(0, 3500, 12),
                              'target': 44000 + np.arange(12) * 1700})
sns.set_theme(style='whitegrid', context='notebook')
employees.head()
```

# Part A Matplotlib 15 Questions and Solutions

## 1. Question: Line chart

```python
# Plot monthly sales with markers, labels, title, and grid.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_sales['month'], monthly_sales['sales'], color='steelblue', marker='o', linewidth=2)
ax.set(title='Monthly Sales', xlabel='Month', ylabel='Sales amount')
ax.grid(alpha=0.3)
fig.autofmt_xdate()
plt.show()
```

## 2. Question: Compare two lines

```python
# Plot actual and target sales using different styles and a legend.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_sales['month'], monthly_sales['sales'], marker='o', label='Actual sales')
ax.plot(monthly_sales['month'], monthly_sales['target'], linestyle='--', color='darkorange', label='Target')
ax.set(title='Actual Sales Versus Target', xlabel='Month', ylabel='Sales amount')
ax.legend()
fig.autofmt_xdate()
plt.show()
```

## 3. Question: Bar chart with value labels

```python
# Show the mean score for each department and write the value over every bar.
mean_scores = employees.groupby('department')['score'].mean().sort_values()
fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(mean_scores.index, mean_scores.values, color='mediumpurple')
ax.bar_label(bars, fmt='%.1f', padding=3)
ax.set(title='Average Performance Score by Department', ylabel='Average score', ylim=(0, 100))
plt.show()
```

## 4. Question: Sorted horizontal bar chart

```python
# Plot total regional revenue from lowest to highest.
region_revenue = employees.groupby('region')['revenue'].sum().sort_values()
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(region_revenue.index, region_revenue.values, color='teal')
ax.set(title='Total Revenue by Region', xlabel='Revenue')
ax.ticklabel_format(style='plain', axis='x')
plt.show()
```

## 5. Question: Encoded scatter plot

```python
# Map training hours to marker size and salary to marker colour.
fig, ax = plt.subplots(figsize=(8, 5))
points = ax.scatter(employees['experience'], employees['score'],
                    s=employees['training_hours'] * 5, c=employees['salary'],
                    cmap='viridis', alpha=0.7)
fig.colorbar(points, ax=ax, label='Salary')
ax.set(title='Experience and Score', xlabel='Years of experience', ylabel='Performance score')
plt.show()
```

## 6. Question: Histogram and mean line

```python
# Use 12 bins and mark the mean score with a dashed vertical line.
mean_score = employees['score'].mean()
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(employees['score'], bins=12, color='cornflowerblue', edgecolor='white')
ax.axvline(mean_score, color='crimson', linestyle='--', label=f'Mean = {mean_score:.1f}')
ax.set(title='Distribution of Performance Scores', xlabel='Score', ylabel='Employee count')
ax.legend()
plt.show()
```

## 7. Question: Two-panel figure

```python
# Place a training-hours histogram beside a salary boxplot.
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(employees['training_hours'], bins=10, color='skyblue', edgecolor='white')
axes[0].set(title='Training Hours', xlabel='Hours', ylabel='Count')
axes[1].boxplot(employees['salary'], patch_artist=True, boxprops={'facecolor': 'lightcoral'})
axes[1].set(title='Salary Distribution', ylabel='Salary', xticks=[])
fig.tight_layout()
plt.show()
```

## 8. Question: Pie chart

```python
# Show department proportions with percentages.
counts = employees['department'].value_counts()
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90,
       wedgeprops={'edgecolor': 'white'})
ax.set_title('Employee Distribution by Department')
plt.show()
```

## 9. Question: Stacked bar chart

```python
# Show department headcount split by region.
department_region = pd.crosstab(employees['department'], employees['region'])
fig, ax = plt.subplots(figsize=(8, 5))
bottom = np.zeros(len(department_region))
for region in department_region.columns:
    ax.bar(department_region.index, department_region[region], bottom=bottom, label=region)
    bottom += department_region[region].to_numpy()
ax.set(title='Department Composition by Region', ylabel='Employee count')
ax.legend(title='Region')
plt.show()
```

## 10. Question: Error bars

```python
# Plot departmental mean scores with one standard-deviation error bars.
summary = employees.groupby('department')['score'].agg(['mean', 'std'])
fig, ax = plt.subplots(figsize=(7, 4))
ax.errorbar(summary.index, summary['mean'], yerr=summary['std'],
            fmt='o', capsize=6, color='navy')
ax.set(title='Mean Score With Standard Deviation', xlabel='Department', ylabel='Score')
plt.show()
```

## 11. Question: Highlight below-target area

```python
# Shade only the months where actual sales are below target.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_sales['month'], monthly_sales['sales'], label='Actual')
ax.plot(monthly_sales['month'], monthly_sales['target'], label='Target')
ax.fill_between(monthly_sales['month'], monthly_sales['sales'], monthly_sales['target'],
                where=monthly_sales['sales'] < monthly_sales['target'],
                color='crimson', alpha=0.2, label='Below target')
ax.set(title='Sales Gap to Target', ylabel='Sales amount')
ax.legend()
fig.autofmt_xdate()
plt.show()
```

## 12. Question: Department boxplots

```python
# Compare salary distributions and show the mean for each group.
names = sorted(employees['department'].unique())
groups = [employees.loc[employees['department'].eq(name), 'salary'] for name in names]
fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(groups, tick_labels=names, showmeans=True)
ax.set(title='Salary Distribution by Department', xlabel='Department', ylabel='Salary')
ax.ticklabel_format(style='plain', axis='y')
plt.show()
```

## 13. Question: Annotate the maximum

```python
# Label the highest monthly-sales point with an arrow.
peak = monthly_sales.loc[monthly_sales['sales'].idxmax()]
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_sales['month'], monthly_sales['sales'], marker='o')
ax.annotate(f'Peak: {peak.sales:,.0f}', xy=(peak.month, peak.sales),
            xytext=(0, 22), textcoords='offset points', ha='center',
            arrowprops={'arrowstyle': '->'})
ax.set(title='Monthly Sales Peak', ylabel='Sales amount')
fig.autofmt_xdate()
plt.show()
```

## 14. Question: Readable date labels

```python
# Plot monthly targets and rotate the x-axis labels by 45 degrees.
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(monthly_sales['month'], monthly_sales['target'], marker='s', color='darkorange')
ax.set(title='Monthly Sales Target', xlabel='Month', ylabel='Target amount')
plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
fig.tight_layout()
plt.show()
```

## 15. Question: Matplotlib correlation heatmap

```python
# Plot numeric correlations with imshow, a colour bar, and cell labels.
corr = employees[['experience', 'training_hours', 'score', 'salary', 'revenue']].corr()
fig, ax = plt.subplots(figsize=(7, 6))
image = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr.columns)), corr.columns, rotation=45, ha='right')
ax.set_yticks(range(len(corr.index)), corr.index)
for row in range(len(corr)):
    for col in range(len(corr)):
        ax.text(col, row, f'{corr.iloc[row, col]:.2f}', ha='center', va='center')
fig.colorbar(image, ax=ax, label='Correlation')
ax.set_title('Numeric Feature Correlation')
fig.tight_layout()
plt.show()
```

# Part B Seaborn 15 Questions and Solutions

## 1. Question: Histogram with KDE

```python
# Show score distribution with a histogram and density curve.
plt.figure(figsize=(8, 4))
sns.histplot(data=employees, x='score', bins=15, kde=True, color='royalblue')
plt.title('Score Distribution With Density Curve')
plt.show()
```

## 2. Question: Ordered count plot

```python
# Count employees in each department from most to least common.
order = employees['department'].value_counts().index
plt.figure(figsize=(7, 4))
sns.countplot(data=employees, x='department', order=order, palette='Blues_d')
plt.title('Employees per Department')
plt.show()
```

## 3. Question: Mean bar plot

```python
# Plot mean score by department with no confidence-interval bars.
plt.figure(figsize=(7, 4))
sns.barplot(data=employees, x='department', y='score', estimator=np.mean, ci=None, palette='Set2')
plt.title('Mean Score by Department')
plt.ylim(0, 100)
plt.show()
```

## 4. Question: Grouped boxplot

```python
# Compare scores by department and region.
plt.figure(figsize=(9, 5))
sns.boxplot(data=employees, x='department', y='score', hue='region', palette='Set3')
plt.title('Scores by Department and Region')
plt.show()
```

## 5. Question: Violin plot

```python
# Compare training-hour distribution shapes across departments.
plt.figure(figsize=(8, 5))
sns.violinplot(data=employees, x='department', y='training_hours', inner='quartile', palette='Pastel1')
plt.title('Training Hours by Department')
plt.show()
```

## 6. Question: Jittered strip plot

```python
# Show individual scores in each region.
plt.figure(figsize=(7, 4))
sns.stripplot(data=employees, x='region', y='score', jitter=0.25, alpha=0.6, palette='Dark2')
plt.title('Individual Scores by Region')
plt.show()
```

## 7. Question: Multi-encoding scatter plot

```python
# Plot experience against score, coloured by department and styled by region.
plt.figure(figsize=(8, 5))
sns.scatterplot(data=employees, x='experience', y='score', hue='department', style='region', s=70)
plt.title('Experience and Score by Group')
plt.show()
```

## 8. Question: Regression plot

```python
# Fit and display a linear relationship between training hours and score.
plt.figure(figsize=(8, 5))
sns.regplot(data=employees, x='training_hours', y='score',
            scatter_kws={'alpha': 0.45}, line_kws={'color': 'crimson'})
plt.title('Training Hours and Performance Score')
plt.show()
```

## 9. Question: Line plot with group comparison

```python
# Compare mean score by experience level for each department.
plt.figure(figsize=(9, 5))
sns.lineplot(data=employees, x='experience', y='score', hue='department', estimator=np.mean, ci=None)
plt.title('Mean Score by Experience and Department')
plt.show()
```

## 10. Question: Point plot

```python
# Compare department means across regions with a point plot.
plt.figure(figsize=(9, 5))
sns.pointplot(data=employees, x='department', y='score', hue='region', ci=None, dodge=True)
plt.title('Mean Score by Department and Region')
plt.show()
```

## 11. Question: Correlation heatmap

```python
# Plot numeric correlations with a Seaborn heatmap and annotations.
corr = employees[['experience', 'training_hours', 'score', 'salary', 'revenue']].corr()
plt.figure(figsize=(7, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True)
plt.title('Numeric Feature Correlation')
plt.show()
```

## 12. Question: Pair plot

```python
# Explore relationships among four numeric variables, coloured by department.
sns.pairplot(employees, vars=['experience', 'training_hours', 'score', 'salary'],
             hue='department', corner=True, plot_kws={'alpha': 0.55, 's': 28})
plt.show()
```

## 13. Question: Joint plot

```python
# Visualize the score-experience relationship with marginal distributions.
joint = sns.jointplot(data=employees, x='experience', y='score', kind='reg',
                      scatter_kws={'alpha': 0.45}, line_kws={'color': 'crimson'})
joint.fig.suptitle('Experience and Score With Marginal Distributions', y=1.02)
plt.show()
```

## 14. Question: Faceted categorical plot

```python
# Create one department score bar chart per region.
grid = sns.catplot(data=employees, x='department', y='score', col='region',
                   kind='bar', estimator=np.mean, ci=None, palette='muted',
                   height=4, aspect=0.9)
grid.set_axis_labels('Department', 'Mean score')
grid.fig.suptitle('Mean Department Score by Region', y=1.05)
plt.show()
```

## 15. Question: Faceted regression plot

```python
# Fit score-experience regression lines separately for each region.
grid = sns.lmplot(data=employees, x='experience', y='score', col='region',
                  hue='department', height=4, aspect=0.9,
                  scatter_kws={'alpha': 0.35, 's': 28})
grid.fig.suptitle('Score and Experience by Region', y=1.04)
plt.show()
```
