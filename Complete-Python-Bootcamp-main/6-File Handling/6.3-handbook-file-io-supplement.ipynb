# Handbook Supplement: File I/O

Variables disappear when a program ends. Files let data survive. Text files store characters; binary files store raw bytes such as images. Always use a path you understand and avoid overwriting important data accidentally.

## Open, read, write, and append

`open(path, mode, encoding=...)` returns a file object. Common modes:

- `r` read (default)
- `w` write, replacing existing content
- `a` append at the end
- `r+` read and write
- `rb`/`wb` binary read/write

The `with` statement closes the file automatically, even when an error happens. Prefer it over manually calling `.close()`.

````python
from pathlib import Path

path = Path("handbook_demo.txt")
path.write_text("first line\nsecond line\n", encoding="utf-8")

with path.open(encoding="utf-8") as file:
    text = file.read()
print(text)

# Read one line at a time:
with path.open(encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        print(line_number, line.strip())
````

## Paths and safe text processing

`pathlib.Path` is clearer and more portable than manually joining slash strings. Use `read_text()` and `write_text()` for small text files; stream line by line for large files.

A replacement workflow should read, transform, and write deliberately:

```python
text = path.read_text(encoding="utf-8")
updated = text.replace("Donkey", "#####")
path.write_text(updated, encoding="utf-8")
```

For important files, write to a new file first or keep a backup. Never use `w` casually: it erases old content.
