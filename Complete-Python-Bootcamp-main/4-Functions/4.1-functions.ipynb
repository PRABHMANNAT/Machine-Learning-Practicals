# Python Functions — Reusable Recipe Cards 🧑‍🍳

## Goal

A function is a named recipe. We give it ingredients (**arguments**), it performs steps, and it may hand back a result with `return`. Functions prevent repeated code and split big problems into small jobs.

> Predict → run → change → explain. All outputs are saved beneath their code cells.

## 1. Anatomy of a function

```python
def function_name(parameters):
    """Docstring: what the function does."""
    # indented function body
    return result
```

`def` creates the recipe; calling `function_name(...)` runs it. Defining alone does not run the body.

````python
def greet_learner():
    """Print one friendly greeting."""
    print("Hello, learner!")

print("The recipe now exists.")
greet_learner()
````

## 2. Why functions?

The original even/odd example first repeats logic directly, then places it in a reusable function. `% 2` gives remainder zero for an even number.

````python
number = 24
if number % 2 == 0:
    print("the number is even")
else:
    print("the number is odd")

def even_or_odd(number):
    """Return a label describing an integer's parity."""
    return "the number is even" if number % 2 == 0 else "the number is odd"

print(even_or_odd(24))
print(even_or_odd(17))
````

## 3. Parameters versus arguments

- **Parameter:** placeholder name in the recipe: `a`, `b`.
- **Argument:** real value supplied during a call: `2`, `4`.
- **Signature:** the function name and parameter rules.

Arguments are evaluated before the function runs.

````python
def add(a, b):
    return a + b

result = add(2, 4)
print(result)
````

## 4. `return` versus `print`

`print` shows something to a human and returns `None`. `return` sends a value back to the caller, so it can be stored or reused. A function stops immediately when it reaches `return`; a function with no explicit return gives `None`.

````python
def show_total(a, b):
    print(a + b)

def calculate_total(a, b):
    return a + b

shown = show_total(2, 3)
calculated = calculate_total(2, 3)
print("show_total returned:", shown)
print("Reusable result:", calculated * 10)
````

### Return multiple values

Comma-separated returned values are packed into one tuple. The caller may keep the tuple or unpack it.

````python
def multiply_details(a, b):
    return a * b, a

result = multiply_details(2, 3)
product, first_input = result
print(result)
print(product, first_input)
````

## 5. Default parameters

A default is a backup ingredient. Required parameters must come before defaulted ones. Defaults are used only when the caller omits that argument.

````python
def greet(name="Guest"):
    return f"Hello {name}, welcome to paradise"

print(greet("Prabh"))
print(greet())
````

### Mutable-default trap

Defaults are created once when `def` runs. A default list would be shared across calls. Use `None`, then create a fresh list inside.

````python
def add_task(task, tasks=None):
    if tasks is None:
        tasks = []
    tasks.append(task)
    return tasks

print(add_task("read"))
print(add_task("write"))
````

## 6. Positional and keyword arguments

Positional arguments match by order. Keyword arguments match by name and make calls self-explanatory. Positional arguments must appear before keyword arguments in a call.

````python
def describe_pet(name, animal="dog"):
    return f"{name} is a {animal}."

print(describe_pet("Buddy"))
print(describe_pet(animal="cat", name="Misty"))
````

### Positional-only and keyword-only parameters

`/` means names before it must be positional. `*` means names after it must be keywords. This advanced syntax can make an API safer and clearer.

````python
def format_price(amount, /, *, currency="₹", decimals=2):
    return f"{currency}{amount:.{decimals}f}"

print(format_price(49.5))
print(format_price(49.5, currency="$", decimals=1))
````

## 7. Variable-length arguments: `*args`

`*args` gathers extra positional arguments into a tuple. The name could differ, but `args` is the clear convention.

````python
def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(1, 2, 3, 4, 5, 6, 7, 8, "Prabh")

def total(*numbers):
    return sum(numbers)
print(total(2, 4, 6))
````

## 8. Variable-length keyword arguments: `**kwargs`

`**kwargs` gathers extra named arguments into a dictionary.

````python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Prabh", age="32", country="India")
````

### Combine normal parameters, `*args`, and `**kwargs`

A common order is: required positional → `*args` → keyword-only/defaults → `**kwargs`.

````python
def show_everything(title, *args, uppercase=False, **kwargs):
    heading = title.upper() if uppercase else title
    print(heading)
    for value in args:
        print(f"Positional argument: {value}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_everything("details", 1, 2, 3, "Krish", uppercase=True,
                name="Krish", age="32", country="India")
````

## 9. Unpacking arguments when calling

`*sequence` spreads positional values; `**mapping` spreads keyword pairs. The receiving signature must accept them.

````python
def rectangle_area(width, height):
    return width * height

size = (4, 5)
labels = {"width": 6, "height": 3}
print(rectangle_area(*size))
print(rectangle_area(**labels))
````

## 10. Scope and LEGB

Python searches names in this order: **Local → Enclosing → Global → Built-in**. A local assignment does not change a global name. Prefer returning values instead of changing global state.

````python
message = "global"

def outer():
    message = "enclosing"
    def inner():
        message = "local"
        return message
    return inner(), message

print(outer())
print(message)
````

### `global` and `nonlocal`

`global` rebinds a module-level name; `nonlocal` rebinds the nearest enclosing-function name. Use both sparingly because hidden state makes testing harder.

````python
def make_counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment

counter = make_counter()
print(counter(), counter(), counter())
````

## 11. Functions are first-class objects

A function can be stored in a variable, placed in a collection, passed to another function, or returned. A function that accepts/returns functions is called **higher-order**.

````python
def double(number): return number * 2
def apply_twice(function, value): return function(function(value))

operation = double
print(operation(5))
print(apply_twice(double, 3))
````

## 12. Documentation and type hints

A docstring explains the contract. Type hints help humans and tools but Python does not enforce them automatically.

````python
def divide(dividend: float, divisor: float) -> float:
    """Return dividend/divisor; divisor must not be zero."""
    if divisor == 0:
        raise ValueError("divisor must not be zero")
    return dividend / divisor

print(divide(10, 4))
print(divide.__annotations__)
print(divide.__doc__)
````

## 13. Best practices and common mistakes

- Give the function one clear job and a verb-based name.
- Return data; let the caller decide how to display it.
- Validate important assumptions and raise meaningful errors.
- Avoid mutable defaults and unnecessary globals.
- Do not shadow built-ins with names like `list`, `sum`, or `filter`.
- Keep calls readable; use keyword arguments for mysterious booleans/numbers.
- A docstring describes purpose, parameters, return, and important errors.

## Revision cheat sheet

| Need | Pattern |
|---|---|
| Define/call | `def f(x): ...` / `f(value)` |
| Send result back | `return value` |
| Backup value | `def f(x=default)` |
| Extra positional | `*args` → tuple |
| Extra named | `**kwargs` → dict |
| Spread values | `f(*sequence, **mapping)` |
| Keyword-only | place after `*` |
| Fixed positional | place before `/` |
| Enclosing state | `nonlocal name` |
| Documentation | first string inside function |

**Memory line:** parameters are recipe placeholders, arguments are real ingredients, and `return` hands the cooked result back.
