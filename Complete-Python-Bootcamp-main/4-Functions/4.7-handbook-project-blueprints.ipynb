## Project 2: Perfect Guess

Goal: generate a secret number and guide the player until the guess is correct.

```python
import random
secret = random.randint(1, 100)
```

The loop compares the guess with the secret, prints `Higher` or `Lower`, increments an attempt counter, and stops on equality. Handle non-numeric input with `try`/`except` and show the final attempt count.

````python
import random

def compare_guess(secret, guess):
    if guess < secret:
        return "Higher"
    if guess > secret:
        return "Lower"
    return "Correct"

secret = random.randint(1, 100)
print(compare_guess(secret, secret))  # a test that is always correct
````

## Mega project: Jarvis voice assistant

The handbook describes a voice-activated assistant with this pipeline:

1. Initialize and greet.
2. Listen for a wake word.
3. Convert speech to text.
4. Route the command: open a website, play music, fetch news, or ask an AI model.
5. Convert the response to speech.
6. Repeat until the user exits.

Possible libraries include `speech_recognition`, `pyttsx3`, `gTTS`, `pygame`, `webbrowser`, `requests`, and an AI SDK. Build a text-only command router first. Add voice one layer at a time. Store API keys in environment variables, never in the notebook.

## Mega project: Auto-reply chatbot

The second mega project automates a chat workflow: capture chat text, identify the latest sender, generate a response, paste it, and repeat.

Possible components include `pyautogui` for UI actions, `pyperclip` for clipboard data, `time` for delays, and an AI SDK for generation.

Important engineering rules:

- Get explicit permission before automating an account or sending messages.
- Add a dry-run mode that prints the proposed reply without sending.
- Rate-limit the loop and provide an emergency stop.
- Never expose private chat history or API keys.
- Treat generated text as untrusted and review it before automatic sending.

## Project checklist

1. Write the inputs, outputs, and failure cases.
2. Turn each responsibility into a function.
3. Test with fake data before using real services.
4. Add logging and exception handling.
5. Keep configuration and secrets outside source code.
6. Document how to run and safely stop the program.
