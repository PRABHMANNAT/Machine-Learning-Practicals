## 1. Define, call, return, and defaults

A parameter is a placeholder inside `def`; an argument is the real value passed during a call. `return` sends data back. Reaching the end without `return` gives `None`.

```python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

def greet(name="friend"):
    return f"Good day, {name}!"

print(greet())
print(celsius_to_fahrenheit(20))
```

## 2. Recursion

A recursive function calls itself. It needs a stopping **base case** and a recursive step that moves closer to it. A loop is often simpler and avoids Python's recursion-depth limit.

```python
def factorial(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("factorial needs a non-negative integer")
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(6))
```

## 3. Lambda, map, filter, reduce, format, and join

| Tool | Child-friendly job |
|---|---|
| `lambda` | Tiny one-expression recipe |
| `map` | Transform every item |
| `filter` | Keep items passing a test |
| `reduce` | Combine items into one result |
| `join` | Glue strings with a separator |
| f-string/`format` | Insert values into text |

Prefer `sum`, `min`, `max`, or a loop over `reduce` when they express the job directly.

```python
from functools import reduce

numbers = [1, 2, 3, 4]
squares = list(map(lambda n: n * n, numbers))
evens = list(filter(lambda n: n % 2 == 0, numbers))
total = reduce(lambda left, right: left + right, numbers)
print(squares)
print(evens)
print(total)
print(", ".join(map(str, squares)))
print("{} is learning {}".format("Maya", "Python"))
```

# Project 1 — Snake, Water, Gun

## Problem

Two players choose one of `snake`, `water`, or `gun`:

- snake drinks water → snake wins;
- water damages gun → water wins;
- gun defeats snake → gun wins;
- equal choices → tie.

## Approach

1. Normalize and validate both choices.
2. Store winning pairs in a set.
3. Return a result instead of printing inside game logic.
4. Let a separate function manage several rounds and scoring.

```python
VALID_CHOICES = {"snake", "water", "gun"}
WINNING_PAIRS = {("snake", "water"), ("water", "gun"), ("gun", "snake")}

def decide_round(player, computer):
    player, computer = player.lower(), computer.lower()
    if player not in VALID_CHOICES or computer not in VALID_CHOICES:
        raise ValueError("choice must be snake, water, or gun")
    if player == computer:
        return "tie"
    return "player" if (player, computer) in WINNING_PAIRS else "computer"

for player, computer in [("snake", "water"), ("water", "snake"), ("gun", "gun")]:
    print(player, computer, "->", decide_round(player, computer))
```

### Complete multi-round game

For testing, computer choices are supplied. In an interactive app, replace them with `random.choice(sorted(VALID_CHOICES))` and collect user input in a loop.

```python
def play_game(player_choices, computer_choices):
    score = {"player": 0, "computer": 0, "tie": 0}
    rounds = []
    for player, computer in zip(player_choices, computer_choices):
        winner = decide_round(player, computer)
        score[winner] += 1
        rounds.append((player, computer, winner))
    return rounds, score

rounds, score = play_game(
    ["snake", "water", "gun", "snake"],
    ["water", "snake", "gun", "gun"],
)
for round_number, result in enumerate(rounds, start=1):
    print(f"Round {round_number}: {result}")
print("Final score:", score)
```

# Project 2 — Perfect Guess

## Problem

Guess a secret integer. After each wrong guess, say `Higher` or `Lower`, count attempts, and stop on a match.

## Approach

Keep comparison logic pure in `guess_hint`. A game function processes guesses and returns history. This separation makes testing easy and avoids interactive blocking.

```python
def guess_hint(secret, guess):
    if guess < secret:
        return "Higher"
    if guess > secret:
        return "Lower"
    return "Correct"

def play_perfect_guess(secret, guesses):
    history = []
    for attempt, guess in enumerate(guesses, start=1):
        hint = guess_hint(secret, guess)
        history.append((attempt, guess, hint))
        if hint == "Correct":
            break
    return history

secret = 42  # Interactive version: random.randint(1, 100)
history = play_perfect_guess(secret, [50, 25, 40, 42])
for attempt, guess, hint in history:
    print(f"Attempt {attempt}: {guess} -> {hint}")
print("Solved in", len(history), "attempts")
```

# Mini Project 3 — Greatest of Three

**Problem:** return the largest of three values. **Logic:** use comparisons or the built-in `max`; here we show the explicit decisions for learning.

```python
def greatest_of_three(a, b, c):
    greatest = a
    if b > greatest:
        greatest = b
    if c > greatest:
        greatest = c
    return greatest

print(greatest_of_three(12, 7, 19))
```

# Mini Project 4 — Inches to Centimetres

**Problem:** convert a measurement. **Formula:** centimetres = inches × 2.54. Validate that the input is numeric and non-negative.

```python
def inches_to_centimetres(inches):
    if not isinstance(inches, (int, float)) or isinstance(inches, bool):
        raise TypeError("inches must be a number")
    if inches < 0:
        raise ValueError("inches must not be negative")
    return inches * 2.54

print(f"10 inches = {inches_to_centimetres(10):.2f} cm")
```

# Mini Project 5 — Multiplication Table

**Problem:** build—not only print—a reusable table. Returning a list lets the caller display, save, or test it.

```python
def multiplication_table(number, start=1, stop=10):
    if start > stop:
        return []
    return [f"{number} × {multiplier} = {number * multiplier}"
            for multiplier in range(start, stop + 1)]

print("\n".join(multiplication_table(5, stop=5)))
```

# Mini Project 6 — Recursive Sum from 1 to n

**Base case:** sum through `0` is `0`. **Recursive step:** `n + sum_to(n−1)`. Validate before recursion.

```python
def recursive_sum(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a non-negative integer")
    if n == 0:
        return 0
    return n + recursive_sum(n - 1)

print(recursive_sum(10))
```

# Mini Project 7 — Remove a Word and Strip Whitespace

**Problem:** remove exact unwanted words and clean surrounding spaces. Build a new list so the input is not mutated. `casefold()` makes matching case-insensitive.

```python
def remove_word_and_strip(words, unwanted):
    target = unwanted.casefold()
    return [word.strip() for word in words
            if word.strip().casefold() != target]

raw_words = [" apple ", "BANANA", " banana ", " mango "]
print(remove_word_and_strip(raw_words, "banana"))
print("Original unchanged:", raw_words)
```
