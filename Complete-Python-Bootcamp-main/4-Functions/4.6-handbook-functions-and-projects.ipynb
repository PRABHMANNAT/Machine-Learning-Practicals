# Handbook Supplement: Functions, Recursion, and Projects

Functions give a name to a reusable job. They reduce repetition and make a large program easier to understand.

## Define, call, return, and default arguments

A parameter is the name inside a function definition; an argument is the value passed during a call. `return` sends a value back. If a function reaches the end without `return`, it returns `None`.

```python
def greet(name="friend"):
    return f"Good day, {name}!"

message = greet("Maya")
```

Keep functions focused: one clear job, useful names, and inputs/outputs that are easy to test.

````python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

def greet(name="friend"):
    return f"Good day, {name}!"

print(greet())
print(celsius_to_fahrenheit(20))
````

## Recursion

A recursive function calls itself. It must have a **base case** that stops the calls and a step that moves toward it.

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)
```

Recursion can express a problem naturally, but a loop is often simpler and avoids hitting Python's recursion limit. Never recurse without a trustworthy stopping condition.

````python
def factorial(n):
    if n < 0:
        raise ValueError("factorial needs a non-negative number")
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(6))
````

## Lambda, map, filter, reduce, format, and join

- `lambda` makes a small one-expression function.
- `map` transforms every item.
- `filter` keeps items whose test is true.
- `reduce` combines items into one result; a normal loop or `sum()` is often clearer.
- `join` combines strings; `format` and f-strings insert values.

```python
from functools import reduce

numbers = [1, 2, 3, 4]
squares = list(map(lambda n: n * n, numbers))
evens = list(filter(lambda n: n % 2 == 0, numbers))
total = reduce(lambda left, right: left + right, numbers)
print(", ".join(map(str, squares)))
```

````python
from functools import reduce
numbers = [1, 2, 3, 4]
print(list(map(lambda n: n * n, numbers)))
print(list(filter(lambda n: n % 2 == 0, numbers)))
print(reduce(lambda left, right: left + right, numbers))
print("{} is learning {}".format("Maya", "Python"))
````
