##### Lambda Functions in Python
Lambda functions are small anonymous functions defined using the **lambda** keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.



````python
#Syntax
lambda arguments: expression
````

````python
def addition(a,b):
    return a+b
````

````python
addition(2,3)
````

````python
addition=lambda a,b:a+b
print(type(addition))
print(addition(5,6))
````

````python
def even(num):
    if num%2==0:
        return True
    
even(24)
````

````python
even = lambda num:num%2==0
print(even(24))
````

````python
even1=lambda num:num%2==0
even1(12)
````

````python
def addition(x,y,z):
    return x+y+z

addition(12,13,14)
````

````python
addition1=lambda x,y,z:x+y+z
addition1(12,13,14)
````

````python
## map()- applies a function to all items in a list
numbers=[1,2,3,4,5,6]
def square(number):
    return number**2

square(2)
````

````python
square=lambda number:number**2
square(2)
````

# Lambda with Map - Explain Like I'm 5

```python
list(map(lambda x:x**2,numbers))
```

---

## What Does This Do?

This takes a **list of numbers** and **squares each one** (multiplies each number by itself).

Example:
- Input: `[1, 2, 3, 4, 5, 6]`
- Output: `[1, 4, 9, 16, 25, 36]`

---

## Breaking It Down into 3 Parts:

### Part 1: Lambda (The Mini Function)
```python
lambda x:x**2
```

**`lambda`** = A quick, tiny function with no name

Think: "I'm creating a **secret recipe** that says: take a number and square it"

- `x` = the input (a number)
- `x**2` = the instruction (square it)

**Examples:**
- `lambda x: x**2` with input `3` → gives `9`
- `lambda x: x**2` with input `5` → gives `25`

---

### Part 2: Map (Apply to Everything)
```python
map(lambda x:x**2, numbers)
```

**`map()`** = "Apply this recipe to **every item** in the list"

Think: "I have a **magic machine** that takes my recipe and applies it to every number in my list"

So if `numbers = [1, 2, 3, 4, 5, 6]`:
- Apply to 1 → 1² = 1
- Apply to 2 → 2² = 4
- Apply to 3 → 3² = 9
- Apply to 4 → 4² = 16
- Apply to 5 → 5² = 25
- Apply to 6 → 6² = 36

---

### Part 3: List (Convert to List)
```python
list(map(...))
```

**`list()`** = Convert the result to a list

Think: "Put all the answers into a **box** (list)"

---

## Full Example:

```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(map(lambda x:x**2, numbers))
print(result)
```

**Output:**
```
[1, 4, 9, 16, 25, 36]
```

---

## Simple Analogy:

Imagine you have **6 chocolates** and you want to **double each one**:

1. **Lambda** = "My rule: Double the chocolate"
2. **Map** = "Apply this rule to ALL 6 chocolates"
3. **List** = "Put all the doubled chocolates in a box"

```
Input:  🍫 🍫 🍫 🍫 🍫 🍫
         1  2  3  4  5  6

Apply rule (×2):
        🍫🍫 🍫🍫 🍫🍫 🍫🍫 🍫🍫 🍫🍫
         2   4   6   8   10  12

Output: [2, 4, 6, 8, 10, 12]
```

## Comparison: Without Lambda

**Normal way (longer):**
```python
numbers = [1, 2, 3, 4, 5, 6]
result = []
for num in numbers:
    result.append(num**2)
print(result)  # [1, 4, 9, 16, 25, 36]
```

**Lambda way (shorter):**
```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(map(lambda x:x**2, numbers))
print(result)  # [1, 4, 9, 16, 25, 36]
```

**Both do the same thing!** Lambda is just a quicker way. 🚀
