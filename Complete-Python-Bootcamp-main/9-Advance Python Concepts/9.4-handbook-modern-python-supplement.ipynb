## Type hints

Type hints document expected types for humans and tools. Python does not enforce them at runtime by itself.

```python
def greeting(name: str) -> str:
    return f"Hello, {name}"

age: int = 25
```

Modern Python can also annotate collections: `list[int]`, `tuple[str, int]`, and `dict[str, int]`. Use `str | int` for a value that may be either type.

````python
def greeting(name: str) -> str:
    return f"Hello, {name}"

scores: dict[str, int] = {"Maya": 95}
print(greeting("Maya"))
print(scores)
````

## Walrus operator and match

The walrus operator `:=` assigns inside an expression. Use it sparingly when it avoids repeating an expensive calculation.

```python
if (length := len("Python")) > 3:
    print(length)
```

`match` compares a value with patterns and is useful when many exact cases exist. `_` is the default case.

```python
def status_message(status):
    match status:
        case 200: return "OK"
        case 404: return "Not Found"
        case _: return "Unknown"
```

````python
if (length := len("Python")) > 3:
    print("length:", length)

def status_message(status):
    match status:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case _:
            return "Unknown"

print(status_message(404))
````

## Comprehensions, enumerate, dictionary merge, and context managers

```python
values = [1, 7, 12, 11, 22]
larger = [value for value in values if value > 8]
for index, value in enumerate(values, start=1):
    print(index, value)

merged = {"a": 1} | {"b": 2}
```

A context manager such as `with open(...)` handles setup and cleanup. You can use multiple context-managed resources in one `with` statement.

Keep comprehensions short. If a comprehension needs several conditions or nested loops, write a normal loop for readability.

````python
values = [1, 7, 12, 11, 22]
print([value for value in values if value > 8])
for index, value in enumerate(values, start=1):
    print(index, value)
print({"a": 1} | {"b": 2})
````

## Global state and the main guard

Avoid `global` when possible: pass values into functions and return new values. If you must change a module-level name, `global` makes that intention explicit.

Use this guard so demo or command-line code runs only when the file is executed directly:

```python
def main():
    print("Running directly")

if __name__ == "__main__":
    main()
```
