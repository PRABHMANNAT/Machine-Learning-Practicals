## Goal

The standard library ships with Python, so no separate `pip install` is needed. We will use deterministic examples and temporary directories so existing course files stay safe.

## 1. Numbers: `array`, `math`, and reproducible `random`

`array.array` stores same-type primitive values. `math` provides numeric functions/constants. `random` is pseudo-random—not for passwords; use `secrets` for security.

```python
import array, math, random, secrets
values = array.array('i', [1, 2, 3, 4])
print(values.tolist())
print(math.sqrt(16), math.pi, math.isclose(0.1+0.2, 0.3))
rng = random.Random(42)
print(rng.randint(1, 10), rng.choice(['apple','banana','cherry']))
print('Secure token example length:', len(secrets.token_hex(4)))
```

## 2. `os`, `sys`, and `pathlib`

`os` talks to the operating system, `sys` to the Python runtime, and `Path` represents filesystem paths cleanly.

```python
import os, sys
from pathlib import Path
print('cwd:', Path.cwd())
print('Python:', sys.version_info[:3])
print('Path separator:', os.sep)
print('Existing test.py?', Path('test.py').is_file())
```

## 3. Safe file/directory work with `tempfile` and `shutil`

The original lesson created `test_dir` and copied `source.txt`. This version performs the same operations in a disposable directory and verifies the copy.

```python
import tempfile, shutil
from pathlib import Path
with tempfile.TemporaryDirectory() as temp_name:
    temp = Path(temp_name)
    test_dir = temp/'test_dir'; test_dir.mkdir()
    copied = test_dir/'destination.txt'
    shutil.copyfile('source.txt', copied)
    print(test_dir.is_dir())
    print(copied.read_text(encoding='utf-8'))
```

## 4. `json` and `csv`

JSON represents objects, arrays, strings, numbers, booleans, and null. CSV represents rows/columns. Serialization turns Python data into storable/transmittable text.

```python
import json, csv
data={'name':'Krish','age':25}
json_text=json.dumps(data, indent=2)
print(json_text)
print(json.loads(json_text))
with open('example.csv', encoding='utf-8', newline='') as file:
    print(list(csv.DictReader(file)))
```

## 5. Dates and time

Use timezone-aware datetimes for real systems. `timedelta` represents a duration. `time.perf_counter()` measures elapsed performance more reliably than wall-clock time.

```python
from datetime import datetime, timedelta, timezone
import time
now=datetime.now(timezone.utc)
print(now.tzinfo)
print((now-timedelta(days=1)).date())
start=time.perf_counter(); total=sum(range(10000)); elapsed=time.perf_counter()-start
print(total, elapsed >= 0)
```

## 6. `collections` and `itertools`

Specialized containers and lazy iterator recipes simplify common patterns.

```python
from collections import Counter, defaultdict, deque
from itertools import chain, combinations, islice
print(Counter('banana'))
groups=defaultdict(list)
for word in ['ant','apple','boat']: groups[word[0]].append(word)
print(dict(groups))
queue=deque([1,2]); queue.append(3); print(queue.popleft())
print(list(chain([1,2],[3,4])))
print(list(combinations('ABC',2)))
print(list(islice(range(100),5)))
```

## 7. Other everyday tools

`re` matches text patterns; `statistics` summarizes numbers; `functools` builds function helpers; `logging` records events; `argparse` defines command-line interfaces.

```python
import re, statistics
print(re.findall(r'\d+', 'There are 123 apples and 456 oranges'))
print(statistics.mean([2,4,6]), statistics.median([1,9,3]))
```

## Common mistakes and interview tips

- Do not use `random` for secrets.
- Avoid naive datetimes across time zones.
- Open CSV with `newline=""` and explicit encoding.
- Prefer `pathlib` over hand-built slash strings.
