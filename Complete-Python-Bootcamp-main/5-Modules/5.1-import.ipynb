# Modules and Imports — Borrowing Organized Toolboxes 🧰

## Goal

A **module** is usually one `.py` file. A **package** is a folder of related modules. Imports let one program reuse names from another. This notebook uses the existing `package/`, `test.py`, text files, and CSV file.

## 1. Import styles

| Style | Meaning |
|---|---|
| `import math` | Bring in the toolbox; use `math.sqrt` |
| `from math import sqrt` | Bring in one named tool |
| `import numpy as np` | Give a long module a conventional nickname |
| `from module import *` | Import many unknown names—avoid it |

Explicit module names show readers where a tool came from and prevent name collisions.

````python
import math
print(math.sqrt(16))

from math import sqrt, pi
print(sqrt(16), sqrt(25), pi)

import statistics as stats
print(stats.mean([2, 4, 6]))
````

## 2. Import a custom module and package

The existing `package/prabhmaths.py` defines `addition` and `substraction`. `package/subpackages/mult.py` defines `multiply`. The package's `__init__.py` re-exports names so callers may use a shorter path.

````python
from package.prabhmaths import addition, substraction
from package.subpackages.mult import multiply
import package

print(addition(2, 3))
print(substraction(4, 3))
print(multiply(4, 5))
print('Package file:',package.__file__)
````

## 3. Import happens once per process

Python executes a module's top-level code the first time it is imported, then caches the module in `sys.modules`. `importlib.reload` deliberately executes it again—useful during exploration, uncommon in applications.

````python
import sys
import package.prabhmaths as custom_math
print('package.prabhmaths' in sys.modules)
print(custom_math.__file__)
````

## 4. `__name__` and the main guard

Inside the file run directly, `__name__` is `"__main__"`. Inside an imported module, it is the import name. Put demos/CLI startup behind the guard so importing does not unexpectedly run the program.

```python
def main():
    print("Run the application")

if __name__ == "__main__":
    main()
```

````python
import runpy
try:
    namespace = runpy.run_path('test.py', run_name='imported_test')
    print('Loaded names include:', sorted(name for name in namespace if not name.startswith('__')))
except ModuleNotFoundError as error:
    print('Existing test.py import needs repair:', error)
print('Notebook __name__:', __name__)
````

## 5. Module search path

Python looks in already-loaded modules, built-ins, and directories in `sys.path`. The first matching name wins. `importlib.util.find_spec` explains where a module would load from. Avoid manually changing `sys.path` in normal packages; install the project instead.

````python
import importlib.util, sys
print('First search location:', sys.path[0])
print('math spec:', importlib.util.find_spec('math'))
print('package spec:', importlib.util.find_spec('package'))
````

## 6. Absolute versus relative imports

- Absolute: `from package.prabhmaths import addition`—clear from project root.
- Relative (inside a package): `from .prabhmaths import addition` or `from ..module import name`.

Relative imports need package context and often fail when an inner file is run directly. Prefer `python -m package.module` from the project root.

## 7. Built-in, standard-library, third-party, custom

`len` is built in; `math` ships with Python; NumPy is third-party; `package.prabhmaths` is custom. Third-party imports should be declared as project dependencies.

````python
from pathlib import Path
import csv

print(Path('source.txt').read_text(encoding='utf-8'))
with Path('example.csv').open(encoding='utf-8', newline='') as file:
    print(list(csv.DictReader(file)))
````

## 8. Common import errors

- `ModuleNotFoundError`: module is unavailable or search path/project directory is wrong.
- `ImportError`: module loaded, but requested name is missing or a circular import occurred.
- Shadowing: a local `math.py` hides standard `math`.
- Circular imports: A imports B while B imports A before initialization finishes.
- Typo/case mismatch: file systems may treat case differently.

Debug with `module.__file__`, `sys.path`, and `find_spec`; never “fix” blindly with random path changes.

## Key concepts summary

Modules separate reusable code; packages organize modules; imports bind names and may execute module top-level code.

## Important syntax and quick revision cheat sheet

| Need | Syntax |
|---|---|
| Whole module | `import module` |
| Specific name | `from module import name` |
| Alias | `import module as alias` |
| Main guard | `if __name__ == "__main__":` |
| Locate module | `importlib.util.find_spec(name)` |
| Run packaged module | `python -m package.module` |

## Common mistakes and interview tips

- Avoid `import *` and wildcard namespace surprises.
- Do not name your file like a standard/installed module.
- Keep imports near the top: standard library, third-party, local.
- A module is cached after its first successful import.

**Revision habit:** explain what each line does, predict the result, run it, and test one edge case.
