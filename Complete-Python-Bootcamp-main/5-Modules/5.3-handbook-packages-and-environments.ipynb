# Handbook Supplement: Modules, pip, and Virtual Environments

Modules keep code in separate files. A package is a collection of modules. This is how a small script grows into an organized project.

## Imports

```python
import math
from random import randint

print(math.sqrt(16))
print(randint(1, 6))
```

Prefer explicit imports and avoid `from module import *`, which hides where names came from. The module `os` helps with operating-system tasks; `random` generates pseudo-random values; `math` provides mathematical functions.

````python
import math
import os

print(math.pi)
print(os.getcwd())
````

## pip and reproducible environments

`pip` installs third-party packages such as Flask. A virtual environment isolates a project's packages from other projects.

```text
python -m venv .venv
# Windows PowerShell
.venv\Scripts\Activate.ps1
python -m pip install flask
python -m pip freeze > requirements.txt
python -m pip install -r requirements.txt
```

Do not commit the `.venv` folder. Commit `requirements.txt` when a project needs a repeatable setup. `pip freeze` records installed versions; it is a snapshot, not an explanation of which packages your code actually uses.

## A module entry point

When a file is imported, its top-level code runs. Protect runnable demo code with:

```python
def main():
    print("Run the program")

if __name__ == "__main__":
    main()
```

Then importing the module gives access to `main` without automatically running the demo.
