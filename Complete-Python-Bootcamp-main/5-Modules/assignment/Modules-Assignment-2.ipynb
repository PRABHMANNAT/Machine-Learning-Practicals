### Q15. Create and import a temporary package.

**Concept tested:** package project

**Question:** Create and import a temporary package.

**Hint:** Write `__init__.py` and core.py.

**Solution approach:** Temporarily add the root, import, call, and clean module cache.

**Step-by-step explanation:** Temporarily add the root, import, call, and clean module cache. Predict each intermediate value, then compare it with the saved result.

**Expected output:** Saved directly beneath the solution cell.

**What was learned:** Packages need coherent structure and import context.

```python
import tempfile,sys,importlib
from pathlib import Path
with tempfile.TemporaryDirectory() as d:
 root=Path(d); pkg=root/'tiny'; pkg.mkdir(); (pkg/'__init__.py').write_text('from .core import double\n'); (pkg/'core.py').write_text('def double(x): return x*2\n'); sys.path.insert(0,str(root))
 try: print(importlib.import_module('tiny').double(9))
 finally: sys.path.remove(str(root)); sys.modules.pop('tiny',None); sys.modules.pop('tiny.core',None)
```
