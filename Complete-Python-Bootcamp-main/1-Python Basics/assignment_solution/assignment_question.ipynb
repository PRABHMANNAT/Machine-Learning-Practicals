**Question 5:** Write a Python program to calculate the factorial of a number.

````python
# Solution
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

num = int(input("Enter a number: "))
print(f"The factorial of {num} is {factorial(num)}")
````

`[::-1]` is Python **slicing**.

General slicing format:

```python
text[start : stop : step]
```

So:

```python
text[::-1]
```

means:

* `start` is empty → start from the default position
* `stop` is empty → go through the whole string
* `step = -1` → move backwards

Example:

```python
text = "hello"
print(text[::-1])
```

Python reads it like:

```text
o → l → l → e → h
```

So output is:

```text
olleh
```

A few more examples:

```python
text = "abcdef"

print(text[::1])    # abcdef
print(text[::2])    # ace
print(text[::-1])   # fedcba
print(text[::-2])   # fdb
```

The important part is the third value:

```python
[start : stop : step]
```

`step = 1` means forward normally, while `step = -1` means backwards.

**Question 20:** Write a Python program to sort a list of numbers in ascending order.

````python
user_input = input("Enter numbers separated by space: ")

items = user_input.split()

numbers = []

for x in items:
    numbers.append(int(x))

numbers.sort()

print(f"Sorted list: {numbers}")
````

````python
# Solution
numbers = [int(x) for x in input("Enter numbers separated by space: ").split()]
numbers.sort()
print(f"Sorted list: {numbers}")
````
