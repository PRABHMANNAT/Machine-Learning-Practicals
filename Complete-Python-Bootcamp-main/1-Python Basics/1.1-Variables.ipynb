# Python Basics 1: Variables

A variable is a **name that stores a value**. Think of it like a label on a box: `age` is the label and `25` is the value.

```python

name = "Maya"
ready = True
```

`=` means "store this value in this name." It does not mean "compare equal".

## Good variable names

Use simple names that explain what they mean. In Python, names often use `snake_case`.

Good names:
- `items_in_cart`
- `student_name`
- `total_price`

Bad names:
- `2cats` (starts with a number)

- `first-name` (uses `-`, which means subtraction)The value has a type, but the variable itself is just a name.

- `if` (this is a reserved Python word)

````python
# A variable can store a number first
thing = 10
print(thing, type(thing).__name__)   # 10, int

# Then the same variable can store text instead
thing = "ten"
print(thing, type(thing).__name__)   # ten, str

# This shows that Python variables can change type.
# The variable name stays the same, but the value changes.
````

## Converting values

Python can change one type into another. For example:
- `int("25")` becomes `25`
- `float("3.5")` becomes `3.5`

- `str(25)` becomes `"25"``input()` always gives text, so if you want a number, convert it first. F-strings let you put values inside text using `{}`.


If the text is not a number, like `int("hello")`, Python raises a `ValueError`.

````python
age_text = "25"
age = int(age_text)
price = float("19.99")
print(f"Age next year: {age + 1}")
print(f"Price: {price}")

# Interactive version:
# name = input("Your name: " )
# age = int(input("Your age: "))
# print(f"Hello {name}; next year: {age + 1}")
````
