## 1. Ways to run Python

```text

```

Put several instructions in a file ending in `.py`, then run it from the terminal:

```text

```

Use the REPL for quick experiments and scripts for programs you want to save and reuse. In Jupyter, each code cell is a small interactive experiment.

````python
# This cell behaves like a tiny REPL experiment.
message = "Hello, Python!"
print(message)
print(2 + 3)
````

## 2. Comments, docstrings, modules, and pip

A one-line comment starts with `#`. A triple-quoted string is a **string literal**, not technically a comment; when placed at the top of a file or function it becomes a docstring that tools can read.

```python
# Explain why, not every obvious action

def greet(name):
    """Return a friendly greeting."""
    return f"Hello, {name}!"
```

A **module** is a `.py` file that can be imported. Python includes built-in modules such as `math`, `random`, and `os`. Third-party packages are installed with `pip`.

```python
import math
print(math.sqrt(25))
```

Typical commands:

```text
python -m pip install package-name
python -m pip freeze > requirements.txt
python -m pip install -r requirements.txt
```

Prefer `python -m pip` because it makes sure pip belongs to the Python interpreter you are using. Install packages only from sources you trust.

````python
# Built-in module: no download is needed.
import math
import random

print(math.sqrt(36))
print(random.randint(1, 6))  # a number like a dice roll
````

## 3. The special value `None`

`None` means "there is no value yet" or "nothing was returned". It is different from `0`, `False`, and `""`.

```python
result = None
if result is None:
    print("No result yet")
```

Use `is None` and `is not None` for this check. A function with no `return` statement returns `None` automatically.

````python
answer = None
print(type(answer).__name__)
if answer is None:
    print("The answer has not been calculated yet.")
````

## 4. Strings: a deeper toolkit

Strings are sequences, so you can index and slice them. Useful methods include:

- `len(text)` - number of characters
- `text.lower()` / `text.upper()` - change case in a new string
- `text.strip()` - remove surrounding whitespace
- `text.startswith(prefix)` / `text.endswith(suffix)` - test boundaries
- `text.count(part)` - count occurrences
- `text.find(part)` - first index, or `-1` if missing
- `text.replace(old, new)` - return text with replacements
- `separator.join(items)` - combine an iterable of strings

```python
raw_name = "  Maya  "
name = raw_name.strip().title()
print(name)
print(" - ".join(["Python", "Basics"]))
```

Escape sequences put special characters inside strings: `\n` creates a new line, `\t` creates a tab, and `\"` puts a quote inside a double-quoted string.

````python
raw_name = "  maya  "
name = raw_name.strip().title()
print(f"Hello, {name}!")
print("Python\nBasics")
print("red,blue,green".split(","))
print(" | ".join(["learn", "build", "repeat"]))
````

## 5. Choosing a collection

The handbook introduces the main containers. Use the simplest one that matches the job:

```python
fruits = ["apple", "mango"]       # list: ordered, changeable
size = (1920, 1080)                 # tuple: ordered, fixed
tags = {"python", "beginner"}      # set: unique values, no indexing
student = {"name": "Maya", "age": 8} # dict: key -> value
```

Lists are good for a sequence you will edit. Tuples are good for a fixed record. Sets remove duplicates and support `union`/`intersection`. Dictionaries make lookup by a meaningful key fast.

Important details:

- An empty dictionary is `{}`; an empty set is `set()`.
- Dictionary keys must be unique; assigning the same key replaces the old value.
- A set cannot contain mutable values such as lists.
- Use `value in collection` to ask whether something is present.

````python
numbers = [1, 2, 2, 3, 3]
unique_numbers = set(numbers)
profile = {"name": "Maya", "score": 95}

print(unique_numbers)
print(profile.get("name"))
print("score" in profile)
````

## 6. Tiny decision and repetition patterns

The later handbook chapters build on the basics like this:

```python
if score >= 40:
    print("Pass")
else:
    print("Try again")

for number in range(1, 4):
    print(number)
```

- `if` chooses a path.
- `for` repeats once for each item.
- `while` repeats while a condition stays true.
- `break` exits now; `continue` skips to the next iteration; `pass` intentionally does nothing.

Always make sure a `while` loop can eventually become false, or it may run forever.

````python
score = 72
if score >= 90:
    grade = "A"
elif score >= 60:
    grade = "B"
else:
    grade = "C"
print(grade)

for number in range(1, 4):
    print("count:", number)
````
