## Multiline comments: the honest explanation

Python has no special multiline-comment symbol. The safest comment style is `#` on every line. Triple quotes create a multiline **string**; when it is not assigned or used, Python ignores the resulting value, so beginners often use it like a comment. At the top of a file or function, it is better understood as a docstring.

```python
# This is a comment.
# It can cover several lines.

"""This is a multiline string.
It can document a module or function."""
```

````python
# A real multiline comment uses # on every line.
# Python ignores each of these lines.

"""This string is a docstring-style note.
It is text, not a special comment syntax."""

print("The program continues.")
````

## First script and REPL

Save this as `hello.py` and run `python hello.py` in a terminal. For tiny experiments, type `python` to open the REPL (Read-Evaluate-Print Loop). In Jupyter, each code cell gives you a similar quick experiment.

```python
print("hello world")
```

Built-in modules come with Python. Third-party packages are installed with pip:

```text
python -m pip install package-name
python -m pip freeze > requirements.txt
```

Use package installation carefully and prefer a virtual environment once projects become larger.

````python
import math
print(math.sqrt(49))
print("Try changing 49 in the REPL or this cell.")
````
