# Python Basics 2: Data Types

A type tells Python what a value is and which operations make sense.

| Type | Meaning | Example |
|---|---|---|
| `int` | whole number | `7` |
| `float` | decimal number | `3.14` |
| `str` | text | `"hello"` |
| `bool` | true/false | `True` |

Ask Python with `type(value)`.

````python
values = (7, 3.14, "hello", True)
for value in values:
    print(value, "->", type(value).__name__)
````

## Numbers

Integers are whole numbers; floats contain decimals. `/` always gives a float. `//` gives floor division, `%` gives the remainder, and `**` means power.

```python
7 / 2   # 3.5
7 // 2  # 3
7 % 2   # 1
2 ** 3  # 8
```

The remainder is useful for testing even numbers: `number % 2 == 0`.

````python
number = 14
print(number / 2)
print(number // 3)
print(number % 2 == 0)
print(2 ** 3)
````

## Strings

Strings are sequences of characters in quotes. Indexes start at zero, and slices exclude the ending index. Strings are immutable: methods return a new string.

```python
text = "hello"
text[0]       # 'h'
text[-1]      # 'o'
text[1:4]     # 'ell'
len(text)     # 5
text.upper()  # 'HELLO'
```

Use f-strings to combine text and values: `f"Welcome, {name}!"`.

````python
text = "hello"
print(text[0], text[-1], text[1:4])
print(len(text), text.upper())
name = "Maya"
print(f"Welcome, {name}!")
````

## Booleans and truthiness

Comparisons produce `True` or `False`. In a condition, Python treats `False`, `None`, `0`, `0.0`, `""`, and empty collections as false; most other values are true.

```python
username = ""
if username:
    print("Welcome")
else:
    print("Please enter a username")
```

Collections you will learn next: `list` (ordered/changeable), `tuple` (ordered/unchangeable), `set` (unique values), and `dict` (key-value pairs).

````python
print(10 > 3)
print("cat" == "dog")
username = ""
if username:
    print("Welcome")
else:
    print("Please enter a username")
````

## Conversion practice

Why does `"Age: " + 8` fail? The left side is text and the right side is a number. Fix it with `"Age: " + str(8)` or, usually better, `f"Age: {8}"`.

Practice: predict the types of `12`, `12.0`, `"12"`, and `12 == 12`; then make a string and print its first character, last character, and length.

**Summary:** types are Python's labels for kinds of values. Convert explicitly and do not confuse `"5"` with `5`.

````python
values = (12, 12.0 , "12" , 12==12)
for value in values:
    print(value, "->", type(value).__name__)

text = "Prabhmannat"
print(text[0], text[-1], len(text))
````
