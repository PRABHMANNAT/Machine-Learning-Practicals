# Python Basics 2 — Data Types: Different Kinds of Toys 🧸

## Goal

A data type tells Python what a value **is** and what operations make sense. You will meet numbers, text, true/false values, `None`, and Python's four main containers.

## 1. Meet the basic types

| Type | Child-friendly meaning | Example |
|---|---|---|
| `int` | Whole number | `7`, `-2`, `0` |
| `float` | Decimal number | `3.14`, `-0.5` |
| `complex` | Real + imaginary number | `2 + 3j` |
| `str` | Text | `"hello"` |
| `bool` | Yes/no truth value | `True`, `False` |
| `NoneType` | No value yet | `None` |

Ask Python with `type(value)`.

```python
values = (7, 3.14, 2 + 3j, "hello", True, None)
for value in values:
    print(repr(value), "->", type(value).__name__)
```

## 2. Numbers: `int`, `float`, and `complex`

Integers have no decimal point. Floats approximate decimal values. Complex numbers use `j` for the imaginary part and are common in engineering, but beginners rarely need them.

`/` gives a float, `//` floors the result, `%` gives the remainder, and `**` means power.

```python
number = 14
print(number / 2)
print(number // 3)
print(number % 2 == 0)
print(2 ** 3)

complex_number = 2 + 3j
print(complex_number.real, complex_number.imag)
```

### Floating-point surprise

Computers store many decimals approximately, so `0.1 + 0.2` is not stored as exactly `0.3`. Use `round()` for display; financial programs often use `decimal.Decimal`.

```python
answer = 0.1 + 0.2
print(answer)
print(round(answer, 2))
```

## 3. Strings are rows of characters

A string uses single, double, or triple quotes. Indexes begin at `0`. Negative indexes count from the end. In a slice `[start:stop:step]`, `stop` is not included.

```python
text = "hello"
print(text[0])       # first character
print(text[-1])      # last character
print(text[1:4])     # indexes 1, 2, 3
print(text[::-1])    # reversed
print(len(text))
print(text.upper())
```

### Strings are immutable

Immutable means “cannot be changed in place.” String methods create a **new** string. Capture the returned value if you want to keep it.

```python
word = "python"
loud_word = word.upper()
print("Original:", word)
print("New:", loud_word)

sentence = "  I like Python  "
print(sentence.strip().replace("like", "love"))
print(sentence.split())
```

### Joining text and using f-strings

`+` concatenates strings, but f-strings are usually easier when values have different types.

```python
first = "Maya"
last = "Patel"
full_name = first + " " + last
score = 9
print(full_name)
print(f"{full_name} scored {score}/10.")
```

## 4. Booleans and comparisons

A Boolean is exactly `True` or `False` (capital first letter). Comparisons create Booleans. `bool` is a special kind of integer in Python, but use it to represent truth—not as an ordinary number.

```python
print(10 > 3)
print("cat" == "dog")
print(isinstance(True, bool))
print(int(True), int(False))
```

## 5. Truthy and falsy values

Inside a condition, Python treats `False`, `None`, numeric zero, empty strings, and empty containers as false. Most non-empty values are true.

```python
test_values = [False, None, 0, 0.0, "", [], "hello", [1], -3]
for value in test_values:
    print(f"{value!r:8} -> {bool(value)}")

username = ""
if username:
    print("Welcome")
else:
    print("Please enter a username")
```

## 6. `None`: absence is its own type

Use `is None`, not `== None`. Zero can be a real answer; `None` means there is no answer yet.

```python
temperature = None
print(type(temperature).__name__)
print(temperature is None)
```

## 7. Container types: boxes that hold other values

| Type | Ordered? | Changeable? | Duplicates? | Example |
|---|---:|---:|---:|---|
| `list` | Yes | Yes | Yes | `["red", "blue"]` |
| `tuple` | Yes | No | Yes | `(10, 20)` |
| `set` | No guaranteed display order | Yes | No | `{"red", "blue"}` |
| `dict` | In insertion order | Yes | Keys unique | `{"name": "Maya"}` |

```python
colors = ["red", "blue", "red"]
point = (10, 20)
unique_colors = {"red", "blue", "red"}
student = {"name": "Maya", "score": 9}

print(colors)
print(point)
print(unique_colors)
print(student["name"])
```

## 8. Mutable versus immutable

- **Mutable** objects can change in place: lists, dictionaries, sets.
- **Immutable** objects cannot: integers, floats, booleans, strings, tuples.

This matters when multiple variables refer to the same object.

```python
shopping = ["milk"]
shopping.append("bread")
print(shopping)

greeting = "hi"
greeting = greeting + "!"  # creates a new string value
print(greeting)
```

## 9. Type conversion and safe inspection

Use `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, and `set()` to convert when the conversion makes sense. Use `isinstance(value, type)` when checking types in real programs.

```python
print(int("12"), float("12.5"), str(12))
print(list("cat"))
print(tuple([1, 2, 3]))
print(set([1, 1, 2, 3]))

age = 12
print(isinstance(age, int))
```

## Checks — Original practice, kept and expanded

Why does `"Age: " + 8` fail? One side is text and the other is a number. Use `str(8)` or an f-string.

```python
values = (12, 12.0, "12", 12 == 12)
for value in values:
    print(value, "->", type(value).__name__)

text = "Prabhmannat"
print(text[0], text[-1], len(text))

print("Age: " + str(8))
print(f"Age: {8}")
```

## Easy revision cheat sheet

| Type | Memory hook | Important detail |
|---|---|---|
| `int` | Whole cookies | No decimal point |
| `float` | Part of a cookie | Approximate decimals |
| `str` | Letters in quotation fences | Indexed, sliced, immutable |
| `bool` | Yes/no switch | `True` or `False` |
| `None` | Empty placeholder | Check with `is None` |
| `list` | Editable shopping list | Ordered and mutable |
| `tuple` | Sealed list | Ordered and immutable |
| `set` | Unique sticker collection | No duplicates |
| `dict` | Labeled drawers | Key → value |

**Fast checks:** `type(x)` tells the exact type; `isinstance(x, int)` asks whether it belongs to a type family; `len(x)` counts items; empty values are usually falsy.
