# Python Basics 0: Syntax and Semantics

Python is a precise language for giving instructions to a computer. Think of a helpful robot: it follows your instructions exactly, so the instructions need the right shape and meaning.

- **Syntax** = the grammar of Python: spelling, punctuation, brackets, colons, and indentation.
- **Semantics** = the meaning: what valid code actually does.


## Case sensitivity and comments

`name`, `Name`, and `NAME` are different names. A comment begins with `#`; Python ignores it, so use comments to explain why code exists.

```python
name = "Maya"
Name = "Leo"
print(name)  # Maya
print(Name)  # Leo
```

````python
name = "Maya"
Name = "Leo"
print(name)
print(Name)
````

## Indentation creates blocks

Indentation (normally four spaces) shows which lines belong together. Imagine putting instructions inside a box. A colon starts the box; the indented lines are inside it.

```python
age = 8
if age >= 5:
    print("You can start school")
    print("This is also inside")
print("This is outside")
```

Never mix tabs and spaces in the same block.

````python
age = 8 #try changing this to 3 and see what happens
if age >= 5:
    print("You can start school")
    print("This is inside the if block")
print("This always runs")
````

## Expressions, statements, and errors

An **expression** produces a value (`2 + 3`, `age > 18`). A **statement** performs an action (`age = 8`, `if`, `print`).

Common error clues:

- `SyntaxError`: Python cannot understand the structure, often a missing quote or colon.
- `IndentationError`: a block is spaced incorrectly.
- `NameError`: a name was misspelled or never created.
- `TypeError`: an operation does not fit the values, such as text plus a number.

Debug from the bottom of the traceback upward: read the final error line, inspect the highlighted line and the line before it, then fix one small thing.

````python
x = 5
y = 10
z = x + y
print("z =", z)
````
