# Python Basics 3 — Operators: Action Buttons 🎮

## Goal

Operators calculate, compare, combine, and test values. You will learn arithmetic, comparison, logical, assignment, membership, identity, bitwise operators, precedence, and short-circuiting.

## 1. Arithmetic operators

| Operator | Meaning | Example → result |
|---|---|---|
| `+` | add | `7 + 2 → 9` |
| `-` | subtract | `7 - 2 → 5` |
| `*` | multiply | `7 * 2 → 14` |
| `/` | true division | `7 / 2 → 3.5` |
| `//` | floor division | `7 // 2 → 3` |
| `%` | remainder | `7 % 2 → 1` |
| `**` | power | `7 ** 2 → 49` |

```python
a, b = 7, 2
print(a + b, a - b, a * b)
print(a / b, a // b, a % b, a ** b)
```

### Remainder solves repeating questions

`number % 2` is `0` for even numbers and `1` for odd numbers. `%` can also wrap values around a cycle, like days of the week or turns in a game.

```python
number = 14
print("even" if number % 2 == 0 else "odd")

day_names = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
day_number = 9
print(day_names[day_number % 7])
```

## 2. Operator precedence: which action happens first?

Python roughly follows PEMDAS: parentheses, exponent, multiplication/division/floor/remainder, then addition/subtraction. Comparisons happen later, then `not`, `and`, `or`.

When in doubt, add parentheses to make your meaning obvious.

```python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(10 + 2 * 3)      # 16
print((10 + 2) * 3)    # 36
```

## 3. Comparison operators ask questions

`==`, `!=`, `>`, `<`, `>=`, and `<=` return Booleans. `=` stores a value; `==` compares two values. String comparison is case-sensitive and follows character ordering.

```python
age = 8
print(age >= 5)
print(age == 18)
print(age != 18)
print("Python" == "python")
print("apple" < "banana")
```

### Chained comparisons

Python lets us write a range check the way we say it aloud. `18 <= temperature <= 25` means both comparisons must be true.

```python
temperature = 22
print(18 <= temperature <= 25)
print(temperature >= 18 and temperature <= 25)  # same question, longer form
```

## 4. Logical operators combine questions

- `and`: both sides must be truthy.
- `or`: at least one side must be truthy.
- `not`: flips truth to false, or false to truth.

```python
has_ticket = True
is_old_enough = False
print(has_ticket and is_old_enough)
print(has_ticket or is_old_enough)
print(not is_old_enough)

is_weekend = True
has_homework = False
can_relax = is_weekend and not has_homework
print("Can relax?", can_relax)
```

### Short-circuiting

Python stops an `and` check as soon as it finds a false value, and stops an `or` check as soon as it finds a true value. This can protect unsafe work.

```python
denominator = 0
safe_to_divide = denominator != 0 and (10 / denominator) > 2
print(safe_to_divide)  # no division-by-zero error

display_name = "" or "Guest"
print(display_name)
```

## 5. Assignment operators update a variable

`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, and `**=` are shortcuts. They calculate with the old value and store the new one.

```python
score = 10
score += 5
score *= 2
print(score)
```

## 6. Membership operators: `in` and `not in`

Membership asks, “Is this item inside that container?” For a dictionary, membership tests keys by default.

```python
fruits = ["apple", "banana", "mango"]
print("banana" in fruits)
print("grape" not in fruits)
print("Py" in "Python")

student = {"name": "Maya", "score": 9}
print("name" in student)
```

## 7. Identity operators: `is` and `is not`

`==` asks whether values are equal. `is` asks whether two names point to the **very same object**. Use `is None` for `None`; do not use `is` as a general equality test.

```python
first = [1, 2]
same = first
copy = [1, 2]

print(first == copy)   # equal contents
print(first is copy)   # different list objects
print(first is same)   # same list object

answer = None
print(answer is None)
```

## 8. Bitwise operators — a small advanced peek

`&`, `|`, `^`, `~`, `<<`, and `>>` work on the binary bits of integers. They are useful for flags and low-level data. Do not confuse `&` with logical `and`, or `|` with logical `or`.

```python
x = 6   # binary 110
y = 3   # binary 011
print(x & y)   # 010 -> 2
print(x | y)   # 111 -> 7
print(x ^ y)   # 101 -> 5
print(x << 1)  # 1100 -> 12
print(x >> 1)  # 0011 -> 3
```

## 9. Operators with strings and containers

Some symbols change meaning depending on the values. `+` joins strings or lists, and `*` repeats them.

```python
print("Py" + "thon")
print("ha" * 3)
print([1, 2] + [3, 4])
print([0] * 5)
```

## Easy revision cheat sheet

| Family | Operators | Question to remember |
|---|---|---|
| Arithmetic | `+ - * / // % **` | “What number do I get?” |
| Comparison | `== != > < >= <=` | “Is this relationship true?” |
| Logical | `and or not` | “How do I combine truth questions?” |
| Assignment | `= += -= *= ...` | “What value should this name store?” |
| Membership | `in`, `not in` | “Is it inside?” |
| Identity | `is`, `is not` | “Is it the exact same object?” |
| Bitwise | `& | ^ ~ << >>` | “How do the integer bits combine?” |

**Top traps:** `=` is not `==`; `/` is not `//`; use `and`/`or` for conditions; use `is` mainly for `None`; add parentheses when precedence is unclear.
