# Python Basics 3: Operators

Operators are symbols or words that calculate, compare, or combine values.

| Operator | Meaning | Example |
|---|---|---|
| `+ - *` | add, subtract, multiply | `7 * 2`
| `/` | true division | `7 / 2`
| `//` | floor division | `7 // 2`
| `%` | remainder | `7 % 2`
| `**` | power | `7 ** 2`

Use parentheses to make the order obvious: `2 + 3 * 4` is `14`, while `(2 + 3) * 4` is `20`.

````python
a, b = 7, 2
print(a + b, a - b, a * b)
print(a / b, a // b, a % b, a ** b)
print(2 + 3 * 4)
print((2 + 3) * 4)
````

## The remainder in real questions

`%` helps test whether a number is even: `number % 2 == 0`. It can also help cycle through repeating patterns, such as every seventh day.

## Comparisons

`==`, `!=`, `>`, `<`, `>=`, and `<=` ask questions and return Booleans. Remember: `=` assigns, `==` compares. Text comparisons are case-sensitive.

````python
number = 14
print("even" if number % 2 == 0 else "odd")
age = 8
print(age >= 5)
print(age == 18)
print("Python" == "python")
````

## Logical operators

- `and`: both conditions must be true.
- `or`: at least one condition must be true.
- `not`: flips a Boolean.

Python short-circuits: `and` stops at the first false part; `or` stops at the first true part.

```python
has_ticket = True
is_old_enough = False
can_enter = has_ticket and is_old_enough
```

````python
has_ticket = True
is_old_enough = False
print(has_ticket and is_old_enough)
print(has_ticket or is_old_enough)
print(not is_old_enough)
````

## Assignment shortcuts and practice

```python
score = 10
score += 5   # score = score + 5
score *= 2   # score = score * 2
```
