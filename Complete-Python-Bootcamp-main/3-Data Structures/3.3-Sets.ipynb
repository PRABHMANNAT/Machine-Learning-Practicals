# Python Data Structures 3 — Sets: A Unique Sticker Box 🟠🟢🔵

## Goal

A set is a mutable collection of **unique, hashable items**. Imagine a sticker box that refuses duplicate stickers. Sets shine at fast membership checks, removing duplicates, and comparing groups.

## 1. Create sets

Use `{1, 2}` for a non-empty set. `{}` creates an empty dictionary, so an empty set must be `set()`. Display order is not a reliable sorting order.

```python
my_set = {1, 2, 3, 4, 5}
empty_set = set()
from_list = set([1, 2, 3, 4, 5, 6])
duplicates_removed = set([1, 2, 3, 6, 5, 4, 5, 6])
print(my_set, type(my_set).__name__)
print(empty_set, type(empty_set).__name__)
print(from_list)
print(sorted(duplicates_removed))
```

## 2. Uniqueness and hashability

Equal values collapse to one member. Set members must be hashable (stable enough to locate again): numbers, strings, and suitable tuples work; lists, dictionaries, and sets do not.

```python
print({1, 1.0, True})  # these compare equal, so only one survives
valid = {"cat", 7, (1, 2)}
print(valid)

try:
    invalid = {[1, 2]}
except TypeError as error:
    print("TypeError:", error)
```

## 3. Add one or many items

`add(x)` adds one item. `update(iterable)` adds every item. Adding an existing item changes nothing.

```python
my_set = {1, 2, 3, 4, 5, 6}
my_set.add(7)
my_set.add(7)
my_set.update([8, 9], {10, 11})
print(sorted(my_set))
```

## 4. Remove safely

- `remove(x)` raises `KeyError` if missing.
- `discard(x)` stays quiet if missing.
- `pop()` removes and returns an **arbitrary** member; do not assume “first” or “last.”
- `clear()` removes everything.

```python
working = {1, 2, 3, 4, 5, 6, 7}
working.remove(7)
working.remove(3)
working.discard(11)
print(sorted(working))

try:
    working.remove(10)  # original error example, now explained safely
except KeyError as error:
    print("KeyError:", error)

popped = working.pop()
print("Popped an arbitrary item:", popped)
working.clear()
print(working)
```

## 5. Membership is the superpower

`x in a_set` is usually average O(1), while scanning a list is O(n). Sets are ideal for repeated membership checks.

```python
allowed_ids = {101, 205, 309, 450}
print(205 in allowed_ids)
print(999 in allowed_ids)
```

## 6. Mathematical operations

| Idea | Method | Operator | Meaning |
|---|---|---|---|
| Union | `a.union(b)` | `a | b` | In either group |
| Intersection | `a.intersection(b)` | `a & b` | In both groups |
| Difference | `a.difference(b)` | `a - b` | In `a`, not `b` |
| Symmetric difference | `a.symmetric_difference(b)` | `a ^ b` | In exactly one group |

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
print("union:", sorted(set1 | set2))
print("intersection:", sorted(set1 & set2))
print("set1 - set2:", sorted(set1 - set2))
print("set2 - set1:", sorted(set2 - set1))
print("symmetric:", sorted(set1 ^ set2))
print("originals unchanged:", sorted(set1), sorted(set2))
```

## 7. Update operations change the left set

`intersection_update`, `difference_update`, `symmetric_difference_update`, and `update` mutate. Operator shortcuts are `&=`, `-=`, `^=`, and `|=`.

```python
working = {1, 2, 3, 4, 5, 6}
working.intersection_update({4, 5, 6, 7, 8, 9})
print(working)

working |= {7, 8}
print(working)
```

## 8. Relationships between sets

`<=` means subset, `<` means proper subset, `>=` means superset, and `isdisjoint` asks whether there is no overlap.

```python
large = {1, 2, 3, 4, 5}
small = {3, 4, 5}
print(large.issubset(small))
print(large.issuperset(small))
print(small <= large)
print(small < large)
print({8, 9}.isdisjoint(large))
```

## 9. Set comprehensions

They look like list comprehensions but use braces and keep only unique results.

```python
squares = {number ** 2 for number in range(-4, 5)}
first_letters = {word[0].lower() for word in ["Apple", "Apricot", "Banana"]}
print(sorted(squares))
print(sorted(first_letters))
```

## 10. `frozenset`: a sealed sticker box

A `frozenset` is immutable and hashable. It can be a dictionary key or member of another set.

```python
permissions = frozenset({"read", "write"})
cache = {permissions: "editor"}
print(cache[frozenset({"write", "read"})])
```

## 11. Remove duplicates—with and without preserving order

`set(items)` removes duplicates but does not promise original order. `list(dict.fromkeys(items))` removes duplicates while keeping first-seen order.

```python
items = [1, 2, 2, 3, 4, 4, 5]
print(sorted(set(items)))
print(list(dict.fromkeys(items)))
```

## 12. Count unique words

Normalize case and punctuation first; otherwise `Python` and `python` are different strings.

```python
import re

text = "In this tutorial we are discussing about sets"
words = text.split()
unique_words = set(words)
print(sorted(unique_words))
print(len(unique_words))

messy = "Python, python! Sets; sets."
normalized = re.findall(r"[a-z]+", messy.lower())
print(set(normalized))
```

## 13. Practical comparisons

Set algebra reads like the real question: mutual friends are an intersection; all skills are a union; missing requirements are a difference.

```python
maya_friends = {"Leo", "Zara", "Noah"}
leo_friends = {"Maya", "Zara", "Ava"}
print("Mutual:", sorted(maya_friends & leo_friends))

required_skills = {"python", "sql", "git"}
my_skills = {"python", "git"}
print("Still learn:", sorted(required_skills - my_skills))
```

## 14. Common mistakes and edge cases

- `{}` is a dictionary, not a set.
- Sets are not indexable or sliceable.
- Do not rely on printed order or which item `pop()` removes.
- `remove` raises on absence; `discard` does not.
- Mutable objects cannot be members.
- Set operators require set-like operands; methods often accept any iterable.
- `True`, `1`, and `1.0` compare equal and collapse in a set.

## Easy revision cheat sheet

| Need | Tool |
|---|---|
| Empty set | `set()` |
| Add one/many | `add(x)` / `update(xs)` |
| Safe remove | `discard(x)` |
| Strict remove | `remove(x)` |
| Either group | `a | b` |
| Both groups | `a & b` |
| Only left | `a - b` |
| Exactly one group | `a ^ b` |
| Containment | `a <= b`, `a >= b` |
| No overlap | `a.isdisjoint(b)` |
| Immutable set | `frozenset(...)` |

**Memory line:** sets store unique hashable items, do not provide positional indexing, and make membership and group comparisons clear.
