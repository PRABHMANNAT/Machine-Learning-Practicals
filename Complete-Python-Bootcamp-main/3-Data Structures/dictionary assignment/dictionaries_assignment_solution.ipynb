### Assignment 14: Counting with Dictionaries

Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.

````python
def count_chars(s):
    count_dict = {}
    for char in s:
        count_dict[char] = count_dict.get(char, 0) + 1
    return count_dict

string = "hello world"
print(count_chars(string))
````

## 1. First: what actually is a dictionary?

Think of a dictionary as a cupboard with **labels**.

```python
student = {
    "name": "Alice",
    "score": 88,
    "age": 20
}
```

Imagine:

```text
LABEL       VALUE
-----------------
name   →    Alice
score  →    88
age    →    20
```

The labels are called **keys**.

The information attached to them is called **values**.

So:

```python
student["name"]
```

means:

> Go to the box labelled `"name"` and give me what's inside.

Result:

```text
Alice
```

And:

```python
student["score"]
```

gives:

```text
88
```

This is the foundation of everything you're doing.

# 3. The three things you absolutely need to know

Suppose:

```python
student = {
    "name": "Alice",
    "score": 88
}
```

If you loop normally:

```python
for x in student:
    print(x)
```

you get the **keys**:

```text
name
score
```

If you want values:

```python
for x in student.values():
    print(x)
```

you get:

```text
Alice
88
```

And this is extremely important:

```python
for key, value in student.items():
    print(key, value)
```

gives:

```text
name Alice
score 88
```

Think of `.items()` as:

> Give me the **key AND its value together**.

So:

```python
for k, v in student.items():
```

simply means:

```text
first time:
k = "name"
v = "Alice"

second time:
k = "score"
v = 88
```

That alone will make Assignments 11 and 12 much easier.

# Assignment 11 — Dictionary Filtering

The question says:

> Create numbers 1–10 as keys and their squares as values.
> Then keep only entries where the key is even.

Don't start with this:

```python
{i: i**2 for i in range(1, 11)}
```

That's a shortcut.

First think manually.

You want:

```text
1 → 1
2 → 4
3 → 9
4 → 16
5 → 25
...
```

Start with:

```python
d = {}
```

Then mentally:

```text
i = 1 → put 1:1
i = 2 → put 2:4
i = 3 → put 3:9
...
```

Normal beginner-friendly version:

```python
d = {}

for i in range(1, 11):
    d[i] = i ** 2
```

Now `d` becomes:

```python
{
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25,
    6: 36,
    7: 49,
    8: 64,
    9: 81,
    10: 100
}
```

Now the question says:

> Keep only even **keys**.

Create another empty dictionary:

```python
even_dict = {}
```

Go through every key-value pair:

```python
for key, value in d.items():
```

Python gives you:

```text
key=1 value=1
key=2 value=4
key=3 value=9
...
```

Then ask:

```text
Is key even?
```

How?

```python
key % 2 == 0
```

If yes, copy it to the new dictionary.

So beginner logic:

```python
even_dict = {}

for key, value in d.items():
    if key % 2 == 0:
        even_dict[key] = value
```

Result:

```text
{
    2: 4,
    4: 16,
    6: 36,
    8: 64,
    10: 100
}
```

Now your screenshot's one-line version:

```python
even_dict = {k: v for k, v in d.items() if k % 2 == 0}
```

means exactly the same thing.

Read it as English:

> For every key and value in `d`, if the key is even, put `key:value` into the new dictionary.

Don't try to memorize this yet.

Master the normal loop first.

# Assignment 13 — `defaultdict`

This one genuinely introduces a **new concept**, so don't worry if it looked strange.

First understand the problem with a normal dictionary.

Suppose:

```python
students = {}
```

You try:

```python
students["A"].append("Alice")
```

Python says:

> Wait. There isn't even an `"A"` key yet.

So you normally have to do:

```python
students["A"] = []
students["A"].append("Alice")
```

Now imagine doing that repeatedly.

`defaultdict` says:

> If a key doesn't exist, I'll automatically create a default value for you.

You write:

```python
from collections import defaultdict
```

Then:

```python
default_dict = defaultdict(list)
```

The important part is:

```python
defaultdict(list)
```

Think:

> Whenever I ask for a missing key, automatically give it an empty list `[]`.

So:

```python
default_dict["a"].append(1)
```

Python sees `"a"` doesn't exist.

Instead of giving an error, it automatically does something like:

```text
"a" → []
```

Then appends `1`:

```text
"a" → [1]
```

Then:

```python
default_dict["a"].append(2)
```

becomes:

```text
"a" → [1, 2]
```

Then:

```python
default_dict["b"].append(3)
```

`"b"` doesn't exist, so Python automatically creates:

```text
"b" → []
```

then appends `3`:

```text
"b" → [3]
```

Final idea:

```text
a → [1, 2]
b → [3]
```

So think of `defaultdict(list)` as:

> **Dictionary where every new key automatically starts with an empty list.**

That's it.

You do not need to deeply learn `collections` yet.

# Assignment 14 — Counting Characters

This one is extremely important because this pattern appears everywhere in programming.

Question:

> `"hello"` contains how many of each character?

You want:

```text
h → 1
e → 1
l → 2
o → 1
```

Imagine you're manually counting.

Start:

```python
count_dict = {}
```

Read `"hello"` one character at a time.

First character:

```text
h
```

Have we seen `h` before?

No.

So:

```text
h → 1
```

Dictionary:

```python
{"h": 1}
```

Next:

```text
e
```

Not seen before:

```python
{"h": 1, "e": 1}
```

Next:

```text
l
```

Not seen:

```python
{"h": 1, "e": 1, "l": 1}
```

Next:

```text
l
```

We've already seen it.

So increase its count:

```text
l: 1 → 2
```

Now:

```python
{
    "h": 1,
    "e": 1,
    "l": 2
}
```

That's the entire algorithm.

The confusing line in your screenshot is:

```python
count_dict[char] = count_dict.get(char, 0) + 1
```

Let's make `.get()` ridiculously simple.

Suppose:

```python
d = {"a": 5}
```

Then:

```python
d.get("a", 0)
```

means:

> Give me `"a"`'s value. If `"a"` doesn't exist, give me `0`.

Since `"a"` exists:

```text
5
```

But:

```python
d.get("z", 0)
```

means:

> Find `z`. If it doesn't exist, give me `0`.

Result:

```text
0
```

So this:

```python
count_dict.get(char, 0) + 1
```

means:

> Get the old count.
> If this character has never appeared, assume old count = 0.
> Then add 1.

Example first `"l"`:

```text
old count = 0
0 + 1 = 1
```

Second `"l"`:

```text
old count = 1
1 + 1 = 2
```

Beautiful once you see it.

Before learning `.get()`, you could write the longer version:

```python
if char in count_dict:
    count_dict[char] = count_dict[char] + 1
else:
    count_dict[char] = 1
```

This is actually what I recommend you learn first.

Only afterward replace it with:

```python
count_dict[char] = count_dict.get(char, 0) + 1
```

Same logic.

# The biggest thing I want you to change

When you see something like:

```python
{k: v for k, v in d.items() if k % 2 == 0}
```

don't immediately try to understand the whole line.

Break it apart.

Ask yourself:

| Question                    | Meaning                     |
| --------------------------- | --------------------------- |
| What data do I start with?  | Dictionary                  |
| What do I need to inspect?  | Each key/value pair         |
| How do I access both?       | `.items()`                  |
| Is there a condition?       | Yes, key must be even       |
| Do I need a new dictionary? | Yes                         |
| What goes into it?          | The accepted key/value pair |

Then write the long version first.

Only after the long version makes complete sense should you shorten it.

## Your dictionary cheat sheet

```python
d = {}
```

Empty dictionary.

```python
d["name"] = "Alice"
```

Add/change value.

```python
d["name"]
```

Get value using key.

```python
for key in d:
```

Loop through keys.

```python
for value in d.values():
```

Loop through values.

```python
for key, value in d.items():
```

Loop through keys and values together.

```python
if key in d:
```

Check whether a key exists.

```python
d.get(key, 0)
```

Get value; use `0` if key doesn't exist.

And the most important pattern for building dictionaries:

```python
result = {}

for something in data:
    key = ...
    value = ...
    result[key] = value
```

If you become comfortable with **that pattern**, most beginner dictionary assignments will become much easier.

For your next few dictionary exercises, I strongly recommend **not using dictionary comprehensions at all**. Solve them using `{}`, `for`, `if`, `.items()`, and `dictionary[key] = value`. Once you can do that without looking up syntax, then learn comprehensions.
