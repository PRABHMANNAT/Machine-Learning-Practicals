# Lists in Real Life — Step-by-Step Examples 🧰

## Goal

This companion notebook turns list tools into small real-world programs. The original to-do list, grades, inventory, and feedback examples are preserved and explained, then expanded from simple to advanced.

## Example 1 — Manage a to-do list

**Story:** One list holds tasks. `append` adds work, `remove` finishes a known task, `in` checks membership, and a loop displays the remaining jobs.

```python
to_do_list = ["Buy Groceries", "Clean the house", "Pay bills"]
to_do_list.append("Schedule meeting")
to_do_list.append("Go For a Run")
to_do_list.remove("Clean the house")

if "Pay bills" in to_do_list:
    print("Don't forget to pay the utility bills")

print("To Do List remaining")
for position, task in enumerate(to_do_list, start=1):
    print(f"{position}. {task}")
```

## Example 2 — Organize student grades

**Steps:** store grades → append a new grade → compute average with `sum/len` → use `max/min`. Always guard against an empty list before division.

```python
grades = [85, 92, 78, 90, 88]
grades.append(95)
average_grade = sum(grades) / len(grades) if grades else 0
print(f"Average Grade: {average_grade:.2f}")
print(f"Highest Grade: {max(grades)}")
print(f"Lowest Grade: {min(grades)}")
```

## Example 3 — Manage inventory

The original example uses a list because only item names matter. A dictionary would be better if each item also needed a quantity.

```python
inventory = ["apples", "bananas", "oranges", "grapes"]
inventory.append("strawberries")
inventory.remove("bananas")

item = "oranges"
print(f"{item} are {'in' if item in inventory else 'out of'} stock.")
print("Inventory List:")
for item in inventory:
    print(f"- {item}")
```

## Example 4 — Collect feedback

Normalize text with `.lower()` so `Great` and `great` match. A generator expression counts comments without building another full list.

```python
feedback = ["Great service!", "Very satisfied", "Could be better", "Excellent experience"]
feedback.append("Not happy with the service")
positive_words = ("great", "excellent")
positive_feedback_count = sum(
    1 for comment in feedback
    if any(word in comment.lower() for word in positive_words)
)
print(f"Positive Feedback Count: {positive_feedback_count}")
for comment in feedback:
    print(f"- {comment}")
```

## Example 5 — Shopping cart totals

Each item is a small dictionary, and the cart is a list of those records. A comprehension calculates line totals.

```python
cart = [
    {"name": "notebook", "price": 50, "quantity": 2},
    {"name": "pen", "price": 10, "quantity": 3},
]
line_totals = [item["price"] * item["quantity"] for item in cart]
for item, line_total in zip(cart, line_totals):
    print(f"{item['name']}: ₹{line_total}")
print("Cart total: ₹", sum(line_totals), sep="")
```

## Example 6 — Clean messy data

Real data may contain `None`, blank text, or extra spaces. Build a new clean list instead of changing the original during iteration.

```python
raw_names = [" Maya ", None, "", "LEO", " zara"]
clean_names = [
    name.strip().title()
    for name in raw_names
    if isinstance(name, str) and name.strip()
]
print(clean_names)
```

## Example 7 — Leaderboard with a custom sort key

Sort records by score descending and then by name ascending. A tuple key can represent multiple rules.

```python
players = [
    {"name": "Zara", "score": 90},
    {"name": "Maya", "score": 95},
    {"name": "Leo", "score": 90},
]
leaderboard = sorted(players, key=lambda player: (-player["score"], player["name"]))
for rank, player in enumerate(leaderboard, start=1):
    print(rank, player["name"], player["score"])
```

## Example 8 — Transpose a matrix

Transposing swaps rows and columns. `zip(*matrix)` unpacks the rows into `zip`. This assumes equally sized rows.

```python
matrix = [[1, 2, 3], [4, 5, 6]]
transposed = [list(column) for column in zip(*matrix)]
print(transposed)
```

## Example 9 — Split data into chunks

Slicing is useful for batches, pages, or mini-batches in machine learning.

```python
items = list(range(1, 11))
chunk_size = 3
chunks = [items[start:start + chunk_size] for start in range(0, len(items), chunk_size)]
print(chunks)
```

## Example 10 — Flatten one level

The two `for` parts mean: visit each row, then each item in that row.

```python
nested = [[1, 2], [3, 4], [5]]
flat = [item for row in nested for item in row]
print(flat)
```

## Example 11 — Moving average window

A sliding slice takes a small window at every valid starting position. This pattern appears in time-series analysis.

```python
readings = [10, 12, 14, 16, 18]
window_size = 3
moving_averages = [
    sum(readings[start:start + window_size]) / window_size
    for start in range(len(readings) - window_size + 1)
]
print(moving_averages)
```

## Example 12 — Undo history as a stack

The newest action sits at the end. `pop()` removes it first.

```python
history = []
history.append("typed hello")
history.append("made text bold")
history.append("deleted title")
last_action = history.pop()
print("Undo:", last_action)
print("Remaining history:", history)
```

## Easy revision cheat sheet

| Real problem | List pattern |
|---|---|
| Ordered tasks | `append`, `remove`, `enumerate` |
| Statistics | `sum`, `len`, `min`, `max` |
| Search inventory | `item in items` |
| Clean records | filtered comprehension |
| Ranking | `sorted(..., key=...)` |
| Grid/matrix | nested lists and `zip(*rows)` |
| Batches | repeated slicing |
| History | stack with `append`/`pop` |

**Choose a list** when order matters, duplicates are allowed, and the collection must change.
