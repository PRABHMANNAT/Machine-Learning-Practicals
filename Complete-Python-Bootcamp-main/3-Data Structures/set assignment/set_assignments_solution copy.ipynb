

# Frozenset

A **frozenset** is an immutable version of a Python set. Once created, it cannot be modified.

## Key Differences from Regular Sets:

| Feature | Set | Frozenset |
|---------|-----|-----------|
| Mutable | ✅ Yes | ❌ No |
| Can add/remove | ✅ Yes | ❌ No |
| Hashable | ❌ No | ✅ Yes |
| Can be dict key | ❌ No | ✅ Yes |
| Can be in a set | ❌ No | ✅ Yes |

## Syntax:

```python
# Create a frozenset
fs = frozenset([1, 2, 3, 4, 5])
print(fs)  # frozenset({1, 2, 3, 4, 5})

# This will raise an error - frozensets are immutable
fs.add(6)  # AttributeError: 'frozenset' object has no attribute 'add'
```

## Common Use Cases:

- **Dictionary keys**: Sets can't be dict keys, but frozensets can
- **Set of sets**: Can't have a set of sets, but can have a set of frozensets
- **Thread-safe data**: Immutable data is safer in multithreaded programs

### Assignment 9: Set and List Conversion

Create a set with the first 5 positive integers. Convert it to a list, append the number 6, and convert it back to a set. Print the resulting set.

````python
k = {1, 2, 3 , 4, 5}
l = list(k)
print(l)  # Output: [1, 2, 3, 4, 5] (order may vary)
l.append(6)
print(l)  # Output: [1, 2, 3, 4, 5, 6]
k = set(l)
print(k)  # Output: {1, 2, 3, 4, 5, 6} (order may vary)
````
