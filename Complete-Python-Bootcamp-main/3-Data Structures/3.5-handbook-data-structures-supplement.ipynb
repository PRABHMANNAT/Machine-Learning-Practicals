# Handbook Supplement: Lists, Tuples, Dictionaries, and Sets

A data structure is a container with rules. Choose the container based on whether you need order, changeability, uniqueness, or key-based lookup.

## Lists and tuples

Lists are ordered and mutable. Indexes start at zero; slices exclude the stop index.

```python
items = [7, 9, "python"]
items.append(10)
items.insert(1, 8)
removed = items.pop(0)
items.remove(9)
```

Useful list methods: `sort()` changes the list in place, `reverse()` reverses it, `append()` adds at the end, `insert(index, value)` adds at a position, `pop()` removes and returns an item, and `remove(value)` removes the first matching value.

Tuples use parentheses and cannot be changed. A one-item tuple needs a comma: `(1,)`, not `(1)`.

````python
marks = [72, 91, 65, 88]
marks.sort()
print(marks)
print(marks[0:2])

coordinates = (10, 20)
print(coordinates.count(10), coordinates.index(20))
# coordinates[0] = 99  # TypeError: tuples are immutable
````

## Dictionaries

A dictionary maps unique keys to values. It is mutable and preserves insertion order in modern Python, but you should use it for lookup rather than relying on position.

```python
student = {"name": "Maya", "marks": [92, 98, 96]}
student["grade"] = "A"
student.update({"city": "Delhi"})
student.get("phone", "not provided")
```

Use `.keys()`, `.values()`, and `.items()` to inspect it. Duplicate keys do not create two entries; the later value replaces the earlier value.

````python
student = {"name": "Maya", "marks": [92, 98, 96]}
student.update({"grade": "A"})
print(student.get("name"))
print(student.get("phone", "not provided"))
for key, value in student.items():
    print(key, "->", value)
````

## Sets

A set stores unique values and is unordered and unindexed. Create an empty set with `set()`, not `{}` (which creates an empty dictionary).

```python
unique = {1, 2, 2, 3}
unique.add(4)
unique.remove(1)
common = unique.intersection({2, 4, 8})
all_values = unique.union({8, 9})
```

Sets are excellent for removing duplicates and membership tests. They cannot contain mutable values such as lists. `20` and `20.0` compare equal, while `"20"` is different.

````python
numbers = [7, 0, 8, 0, 0, 9]
unique = set(numbers)
print(unique)
print(unique.intersection({0, 9}))
print(len({20, 20.0, "20"}))  # 2
````
