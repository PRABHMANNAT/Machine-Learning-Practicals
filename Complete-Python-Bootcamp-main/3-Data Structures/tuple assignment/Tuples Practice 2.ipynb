### Q25. Flatten arbitrarily nested tuples

**Concept tested:** recursion and type checks

**Solution logic:** For each item, recursively expand tuples; otherwise emit the leaf value.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Recursion handles unknown nesting depth.

```python
def flatten_tuple(value):
    result = []
    for item in value:
        result.extend(flatten_tuple(item) if isinstance(item, tuple) else [item])
    return tuple(result)

print(flatten_tuple((1, (2, (3, 4)), 5)))
```

### Q26. Chunk values into fixed tuple groups

### Q28. Cache results using tuple arguments

**Concept tested:** tuple-key memoization

**Solution logic:** Use all function inputs as a compound dictionary key.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Hashable tuples make simple multi-argument cache keys.

```python
cache = {}

def rectangle_area(width, height):
    key = (width, height)
    if key not in cache:
        cache[key] = width * height
    return cache[key]

print(rectangle_area(4, 5), rectangle_area(4, 5), cache)
```

### Q29. Find palindromic tuple slices

**Concept tested:** tuple slicing and reversal

**Solution logic:** Generate all non-trivial slices and keep those equal to their reverse.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Tuple equality and reverse slicing support sequence algorithms.

```python
values = (1, 2, 1, 3)
palindromes = []
for start in range(len(values)):
    for stop in range(start + 2, len(values) + 1):
        piece = values[start:stop]
        if piece == piece[::-1]:
            palindromes.append(piece)
print(palindromes)
```

### Q30. Rotate a square tuple matrix clockwise

**Concept tested:** transpose and reverse

**Solution logic:** Reverse the rows, then transpose them with `zip`.

**Expected output:** The exact executed result is saved directly beneath the solution cell.

**What you learn:** Composition of immutable operations can transform structured data.

```python
matrix = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
rotated = tuple(zip(*matrix[::-1]))
print(rotated)
```
