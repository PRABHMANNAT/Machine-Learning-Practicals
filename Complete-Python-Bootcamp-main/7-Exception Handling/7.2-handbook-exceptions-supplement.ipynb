# Handbook Supplement: Handling Errors

Exceptions are Python's way of reporting that an operation cannot continue normally. Good handling anticipates expected failures and gives the user a useful next step.

## try, except, else, finally

Put only risky code in `try`. Catch specific exceptions you know how to handle. `else` runs when the `try` succeeds. `finally` runs whether it succeeds or fails, so it is useful for cleanup.

```python
try:
    number = int(input("Number: "))
except ValueError:
    print("Please enter whole digits.")
else:
    print(number * 2)
finally:
    print("Finished.")
```

Avoid a bare `except:` unless you immediately re-raise or have a very deliberate reason; it can hide programming bugs.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

result = safe_divide(10, 0)
if result is None:
    print("Division by zero is not allowed.")
else:
    print(result)
````

## Raise useful exceptions

Use `raise` when a value violates the contract of your function. Include a message that explains how to fix it.

```python
def set_age(age):
    if age < 0:
        raise ValueError("age must be non-negative")
    return age
```

Common beginner exceptions: `ValueError` for the right type but wrong value, `TypeError` for the wrong kind of value, `FileNotFoundError` for a missing path, `KeyError` for a missing dictionary key, and `ZeroDivisionError` for division by zero.

````python
def set_age(age):
    if age < 0:
        raise ValueError("age must be non-negative")
    return age

try:
    print(set_age(-5))
except ValueError as error:
    print("Invalid input:", error)
````
