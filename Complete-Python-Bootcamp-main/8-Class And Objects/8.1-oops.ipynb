####  Classes and Objects
Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods.

## OOP explained like you are 5

Imagine that you want to make toy cars. A **class** is the drawing or blueprint used to make them. An **object** is one real toy made from that blueprint. So `Car` is the blueprint, while `audi`, `bmw`, and `tata` are separate toy cars.

### Important words

- **Class:** A blueprint for making objects.
- **Object:** A real thing created from a class.
- **Attribute:** Information belonging to an object, such as its name, age, windows, or balance.
- **Method:** An action an object can perform, such as barking, depositing money, or withdrawing money.
- **`pass`:** Do nothing for now; it lets us create an empty class.
- **`__init__`:** The automatic setup helper that runs when a new object is created.
- **`self`:** This particular object. For example, `self.name` means the name belonging to this particular dog.
- **`return`:** Give an answer back to the code that asked for it.

### What happens in the Car example?

`audi = Car()` and `bmw = Car()` make two different objects from the same `Car` blueprint. `audi.windows = 4` attaches the information `windows` to Audi. Tata only receives `doors`, so asking for `tata.windows` causes an `AttributeError`: Python was never told how many windows Tata has. `dir(tata)` lists the information and built-in abilities available to Tata.

When Python prints something like `<__main__.Car object at 0x...>`, it simply means: 'This is a Car object stored somewhere in the computer's memory.' The strange number is not important here.

### What happens in the Dog example?

`Dog("Buddy", 3)` creates a dog named Buddy who is 3 years old. `Dog("Lucy", 4)` creates a different dog named Lucy who is 4. They use the same blueprint but keep their own information. The `bark()` method uses `self.name`, so Buddy says woof using Buddy's name and Lucy says woof using Lucy's name.

### What happens in the BankAccount example?

`BankAccount("Krish", 5000)` creates Krish's account with a balance of 5000. `deposit(100)` adds 100, making 5100. `withdraw(300)` removes 300, leaving 4800. If we try to withdraw more money than the account has, it prints `Insufficient funds!`. Finally, `get_balance()` gives back the current balance.

### One-sentence summary

**OOP keeps a thing's information and actions together: a Dog object keeps its name and age and also knows how to bark.**

```python
### Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age

## create objects
dog1=Dog("Buddy",3)
print(dog1)
print(dog1.name)
print(dog1.age)
    
    
```

```python
dog2=Dog("Lucy",4)
print(dog2.name)
print(dog2.age)
```

```python
## Define a class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def bark(self):
        print(f"{self.name} says woof")


dog1=Dog("Buddy",3)
dog1.bark()
dog2=Dog("Lucy",4)
dog2.bark()

```

```python
### Modeling a Bank Account

## Define a class for bank account
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient funds!")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. New Balance is {self.balance}")

    def get_balance(self):
        return self.balance
    
## create an account

account=BankAccount("Krish",5000)
print(account.balance)

    
```

```python
## Call isntance methods
account.deposit(100)
```

```python
account.withdraw(300)
```

```python
print(account.get_balance())
```
