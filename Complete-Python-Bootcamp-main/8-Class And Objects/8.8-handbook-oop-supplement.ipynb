Object-oriented programming groups data and the actions that use it. Model the problem by looking for nouns (classes), adjectives (attributes), and verbs (methods). Keep the class small and focused.

## Classes, objects, attributes, and methods

A class is a blueprint. An object is one instance built from it. `__init__` sets instance state; `self` refers to the current object. A class attribute is shared by the class; an instance attribute belongs to one object and takes precedence when names collide.

```python
class Employee:
    company = "Example Ltd"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_salary(self):
        return self.salary * 12
```

````python
class Employee:
    company = "Example Ltd"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def annual_salary(self):
        return self.salary * 12

worker = Employee("Maya", 5000)
print(worker.name, worker.annual_salary(), worker.company)
````

## Static methods, class methods, and properties

- `@staticmethod` is a related utility that does not need `self`.
- `@classmethod` receives `cls` and works with class-level state or alternate constructors.
- `@property` lets an attribute-looking expression run validation or calculation.

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32
```

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return self.celsius * 9 / 5 + 32

print(Temperature(20).fahrenheit)
````

## Inheritance and super

Inheritance lets a child class reuse a parent class. Use `super()` to call parent behavior. Override a method when the child needs a specialized version. Single, multiple, and multilevel inheritance exist, but composition is often easier to understand than a deep hierarchy.

```python
class Animal:
    def speak(self):
        return "sound"

class Dog(Animal):
    def speak(self):
        return "bark"
```

Polymorphism means different objects can respond to the same method name in their own way.
