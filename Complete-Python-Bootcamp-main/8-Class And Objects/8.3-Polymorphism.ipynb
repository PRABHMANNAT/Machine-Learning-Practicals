# Polymorphism in Python — Explained Like You Are 5

## Goal

**Polymorphism** means **one action can behave in many forms**.

Imagine pressing the same **Speak** button:

- A dog says **Woof!**
- A cat says **Meow!**
- A duck says **Quack!**

The button name stays the same, but each animal responds differently.

> **Big idea:** Same instruction, different behavior depending on the object receiving it.

## Important words

| Word | Easy meaning |
|---|---|
| Polymorphism | One action with many forms |
| Method | An action belonging to an object |
| Method overriding | A child replaces a parent's method with its own version |
| Duck typing | Python cares about what an object can do, not only its class name |
| Abstract method | A required method that child classes must create |

## 1. Built-in polymorphism

You have already used polymorphism! The same `len()` function works with strings, lists, and dictionaries.

`len()` asks each object: **How many items do you contain?** Each type knows how to answer.

```python
word = "Python"
fruits = ["apple", "banana", "mango"]
student = {"name": "Riya", "grade": 5}

print("Letters in the word:", len(word))
print("Fruits in the list:", len(fruits))
print("Keys in the dictionary:", len(student))
```

## 2. Operator polymorphism

The `+` symbol also has several forms:

- With numbers, it performs addition.
- With strings, it joins text.
- With lists, it joins the lists.

Python looks at the objects and chooses the correct meaning.

```python
print("Numbers:", 2 + 3)
print("Strings:", "Hello " + "World")
print("Lists:", [1, 2] + [3, 4])
```

## 3. Method overriding — original Animal example

A parent class can provide a general method. Child classes can **override** it with a more suitable version.

All three objects understand `speak()`, but each object gives its own answer.

```python
# Base class
class Animal:
    def speak(self):
        return "Sound of the animal"


# Derived class 1
class Dog(Animal):
    def speak(self):
        return "Woof!"


# Derived class 2
class Cat(Animal):
    def speak(self):
        return "Meow!"


# A function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())


dog = Dog()
cat = Cat()

print(dog.speak())
print(cat.speak())
animal_speak(dog)
```

### One loop, many animals

A list can hold different objects. The loop sends the same `speak()` message to every object, and each object responds in its own way.

```python
class Duck(Animal):
    def speak(self):
        return "Quack!"


animals = [Dog(), Cat(), Duck()]

for animal in animals:
    print(animal.speak())
```

## 4. Polymorphism with functions — original Shape example

Rectangle and Circle both provide an `area()` method. The `print_area()` function does not need separate instructions for every shape. It simply calls `shape.area()`.

This makes the function flexible: any new object with a working `area()` method can be used later.

```python
class Shape:
    def area(self):
        return "The area of the figure"


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


def print_area(shape):
    print(f"The area is {shape.area():.2f}")


rectangle = Rectangle(4, 5)
circle = Circle(3)

print_area(rectangle)
print_area(circle)
```

## 5. Duck typing

Python often follows this playful rule:

> If it walks like a duck and quacks like a duck, treat it like a duck.

Objects do not always need the same parent class. If they provide the method a function needs, the function can use them.

`make_music()` only cares whether an object has a working `play()` method.

```python
class Piano:
    def play(self):
        return "Plink plonk from the piano"


class Guitar:
    def play(self):
        return "Strum strum from the guitar"


def make_music(instrument):
    print(instrument.play())


make_music(Piano())
make_music(Guitar())
```

## 6. Polymorphism with abstract base classes — original Vehicle example

An **abstract base class** can make a rule that every child must follow.

`Vehicle` says: **Every vehicle must have `start_engine()`.** Car and Motorcycle follow that rule in their own ways.

```python
from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"


class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"


def start_vehicle(vehicle):
    print(vehicle.start_engine())


car = Car()
motorcycle = Motorcycle()

start_vehicle(car)
start_vehicle(motorcycle)
```

## 7. Real-world example: payment methods

A shop can call `pay()` without knowing the exact payment type. Card, cash, and UPI objects each perform the action differently.

```python
class CardPayment:
    def pay(self, amount):
        return f"Paid {amount} using a card"


class CashPayment:
    def pay(self, amount):
        return f"Paid {amount} using cash"


class UPIPayment:
    def pay(self, amount):
        return f"Paid {amount} using UPI"


def checkout(payment_method, amount):
    print(payment_method.pay(amount))


payment_methods = [CardPayment(), CashPayment(), UPIPayment()]

for method in payment_methods:
    checkout(method, 500)
```

## 8. Method overloading in Python

Some languages allow several methods with the same name but different parameters. Python does **not** support that traditional style directly: a later method definition replaces an earlier one.

Instead, Python commonly uses default values or `*args` to accept different numbers of inputs.

```python
class Calculator:
    def add(self, first, second=0, third=0):
        return first + second + third


calculator = Calculator()

print(calculator.add(5))
print(calculator.add(5, 3))
print(calculator.add(5, 3, 2))
```

## 9. A common mistake

Polymorphism only works when each object provides the expected method. The next function expects `speak()`. A rock does not provide it, so Python raises `AttributeError`.

We catch the error so the notebook can continue running.

```python
class Rock:
    pass


try:
    animal_speak(Rock())
except AttributeError as error:
    print(f"AttributeError: {error}")
```

## Why is polymorphism useful?

- It lets one function work with many object types.
- It reduces long chains of `if` and `elif` checks.
- New classes can be added without rewriting every function.
- Each class keeps the behavior that belongs to it.
