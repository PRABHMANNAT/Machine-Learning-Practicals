# Abstraction in Python — Explained Like You Are 5

## Goal

**Abstraction** means showing the simple controls and hiding the complicated work behind them.

Think about a television remote. You press the **power button**, and the television turns on. You do not need to understand wires, electricity, or circuits.

> **Big idea:** Show people what they need to use; hide the complicated steps they do not need to see.

## Everyday examples

- You press a car's start button without controlling every engine part.
- You use an ATM's Withdraw button without seeing the bank's server code.
- You call `coffee_machine.make_coffee()` without manually heating water or controlling pumps.

The simple button or method is the **interface**. The hidden steps are the **implementation**.

## Important words

| Word | Easy meaning |
|---|---|
| Abstraction | Hide complexity and show only useful controls |
| Interface | The buttons or methods we are allowed to use |
| Implementation | The hidden steps that make the action work |
| Abstract class | An incomplete blueprint used as a parent |
| Abstract method | A rule saying every child must create this method |
| Concrete class | A finished class that can create objects |

## 1. A simple abstraction without `ABC`

The user calls one easy method: `make_coffee()`. The machine performs several smaller steps inside that method.

The user does not need to call every step separately.

```python
class CoffeeMachine:
    def _heat_water(self):
        print("Heating the water...")

    def _add_coffee(self):
        print("Adding coffee...")

    def make_coffee(self):
        self._heat_water()
        self._add_coffee()
        print("Your coffee is ready!")


machine = CoffeeMachine()
machine.make_coffee()
```

### What does the leading underscore mean?

Names such as `_heat_water()` begin with `_` to tell other programmers:

> This is an internal helper. Please use the main public method when possible.

Python does not make a single underscore truly private. It is a polite naming convention.

## 2. Abstract Base Classes (`ABC`)

Python's `abc` module helps us create official abstract classes.

- Inherit from `ABC` to make an abstract base class.
- Put `@abstractmethod` above a required method.
- A child must implement every abstract method before objects can be created from it.

```python
from abc import ABC, abstractmethod
```

## 3. Original Vehicle and Car example

`Vehicle` is an incomplete blueprint:

- It provides the ready-to-use `drive()` method.
- It requires every child to create `start_engine()`.

`Car` follows the rule, so Python allows us to create a Car object.

```python
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass


class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")


def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()


car = Car()
operate_vehicle(car)
```

## 4. Why can't we create an abstract object?

`Vehicle` is an unfinished promise. Its `start_engine()` method has no real behavior yet, so creating `Vehicle()` would not make sense.

Python stops us with a `TypeError`. We catch it below so the lesson continues.

```python
try:
    unfinished_vehicle = Vehicle()
except TypeError as error:
    print(f"TypeError: {error}")
```

## 5. One rule, several finished children

Both Car and Motorcycle must provide `start_engine()`, but each can start differently.

This combines **abstraction** and **polymorphism**:

- Abstraction creates the common rule.
- Polymorphism lets each child follow the rule differently.

```python
class Motorcycle(Vehicle):
    def start_engine(self):
        print("Motorcycle engine started with a button")


vehicles = [Car(), Motorcycle()]

for vehicle in vehicles:
    operate_vehicle(vehicle)
```

## 6. More than one abstract method: Shapes

An abstract class can require several methods. Every finished shape below must know how to calculate both `area()` and `perimeter()`.

```python
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


rectangle = Rectangle(4, 5)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

## 7. A child must complete every rule

`IncompleteCircle` creates `area()` but forgets `perimeter()`. It is still incomplete, so Python will not create an object from it.

```python
class IncompleteCircle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius**2


try:
    circle = IncompleteCircle(3)
except TypeError as error:
    print(f"TypeError: {error}")
```

### Completing the Circle class

Once Circle implements both required methods, it becomes a concrete class and can create objects.

```python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius**2

    def perimeter(self):
        return 2 * 3.14 * self.radius


circle = Circle(3)
print(f"Area: {circle.area():.2f}")
print(f"Perimeter: {circle.perimeter():.2f}")
```

## 8. Abstract properties

An abstract class can also require information, not just actions. `@property` lets us read a method like an attribute.

Every employee below must provide a `role` property.

```python
class Employee(ABC):
    def __init__(self, name):
        self.name = name

    @property
    @abstractmethod
    def role(self):
        pass

    def introduce(self):
        return f"I am {self.name}, and I work as a {self.role}."


class Developer(Employee):
    @property
    def role(self):
        return "developer"


developer = Developer("Aarav")
print(developer.introduce())
```

## 9. Real-world example: payment system

The checkout code should not care about hidden details such as card networks or UPI servers. It only needs a simple `pay()` method.

`PaymentMethod` defines the rule, and each finished payment class hides its own implementation.

```python
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid {amount} securely using a card"


class UPIPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid {amount} securely using UPI"


def checkout(payment_method, amount):
    print(payment_method.pay(amount))


checkout(CardPayment(), 500)
checkout(UPIPayment(), 500)
```

## 10. Abstraction compared with other OOP ideas

| Concept | Main question | Easy example |
|---|---|---|
| Abstraction | What simple controls should users see? | A car's start button |
| Encapsulation | How do we protect and group data? | Keeping balance inside an account |
| Inheritance | How can a child reuse a parent's code? | Tesla inherits from Car |
| Polymorphism | How can one action have many forms? | Dog and Cat both use `speak()` |

These ideas often work together, but they solve different problems.

## 11. An abstract method can provide shared starter behavior

An abstract method may contain code even though children are still required to override it. A child can call that shared code with `super()` and then add its own steps.

This is useful when every child needs the same beginning or ending behavior.

```python
class Report(ABC):
    @abstractmethod
    def create(self):
        print("Preparing common report information...")


class SalesReport(Report):
    def create(self):
        super().create()
        print("Adding sales numbers to the report.")


SalesReport().create()
```

## 12. Abstraction without `ABC`: `NotImplementedError`

Older or lightweight code sometimes defines a normal parent method that raises `NotImplementedError`. This tells children to replace the method, but Python does not prevent the parent object from being created.

`ABC` and `@abstractmethod` are stronger because Python checks the rule when an object is created.

```python
class SimpleAnimal:
    def speak(self):
        raise NotImplementedError("Child classes must create speak()")


class SimpleDog(SimpleAnimal):
    def speak(self):
        return "Woof!"


print(SimpleDog().speak())

try:
    SimpleAnimal().speak()
except NotImplementedError as error:
    print(f"NotImplementedError: {error}")
```

## 13. Public interface versus hidden helpers

A clear class usually has:

- **Public methods** that users are expected to call, such as `place_order()`.
- **Internal helpers** whose names begin with `_`, such as `_check_stock()`.

The underscore is a convention, not a security lock. It communicates which controls form the simple public interface.

```python
class OnlineStore:
    def _check_stock(self, item):
        print(f"Checking stock for {item}...")
        return True

    def _take_payment(self, amount):
        print(f"Taking payment of {amount}...")

    def place_order(self, item, amount):
        if self._check_stock(item):
            self._take_payment(amount)
            print(f"Order placed for {item}!")


store = OnlineStore()
store.place_order("Python Book", 500)
```

## 14. When should you use abstraction?

Use abstraction when:

- Users need a small, simple interface.
- Several related classes must follow the same rules.
- Complicated implementation details may change later.
- You want to prevent incomplete child classes from being used.

Avoid creating abstract classes for tiny programs with only one simple implementation. Extra structure is helpful only when it makes the code easier to understand and extend.

## Common mistakes

1. Trying to create an object directly from an abstract class.
2. Forgetting to implement one of the required abstract methods.
3. Making every method abstract, leaving no useful shared behavior.
4. Confusing abstraction with privacy—abstraction simplifies an interface; it does not automatically secure data.
5. Building an abstract class before there is a real family of related objects.
