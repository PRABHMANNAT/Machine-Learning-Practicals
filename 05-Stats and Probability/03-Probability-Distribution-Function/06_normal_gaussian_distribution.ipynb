# Normal (Gaussian) Distribution

## The famous bell-shaped pattern

Many measurements gather near a middle value and become less common as we move away from it. The normal distribution is a smooth bell-shaped model for this pattern.

**By the end, you can:** read the mean and standard deviation, use the empirical rule, and calculate a probability as an area under the curve.

## What the two settings mean

Write $X\sim N(\mu,\sigma^2)$.

* $\mu$ (mu) is the **mean**: the centre of the bell.
* $\sigma$ (sigma) is the **standard deviation**: how spread out the bell is.

Changing $\mu$ slides the bell left or right. Changing $\sigma$ makes it narrow/tall or wide/short. The total area under every normal curve is $1$, because all probability together is $100\%$.


## PDF and the 68–95–99.7 rule

The normal probability density function is:

$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2\right)$$

You do not need to memorise every symbol to use the idea. The important part is that **probability is area**, not the height of the curve at one exact point.

For a normal distribution, approximately:

* $68\%$ of values lie within $\mu\pm1\sigma$;
* $95\%$ lie within $\mu\pm2\sigma$;
* $99.7\%$ lie within $\mu\pm3\sigma$.


## Worked example

Suppose test scores are approximately $N(70,10^2)$. The mean is $70$ and the standard deviation is $10$.

* Around $68\%$ of scores are between $60$ and $80$.
* Around $95\%$ are between $50$ and $90$.

To find an exact probability such as $P(60\le X\le80)$, we use a normal CDF in code or a table. The answer is the shaded area between those scores.


```python
# Beginner-friendly guide:
# We draw normal curves with different centres and spreads to see what mu and sigma do.
# A normal CDF gives the area (probability) to the left of a value.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(30, 110, 500)  # Many x-values make a smooth-looking bell curve.
for mean, sd, label in [(70, 10, "mean=70, sd=10"), (70, 5, "mean=70, sd=5"), (80, 10, "mean=80, sd=10")]:
    plt.plot(x, norm.pdf(x, loc=mean, scale=sd), label=label)

probability = norm.cdf(80, loc=70, scale=10) - norm.cdf(60, loc=70, scale=10)
print(f"P(60 <= score <= 80) = {probability:.3f}")
plt.xlabel("Score")
plt.ylabel("Density")
plt.title("How mean and standard deviation change a normal curve")
plt.legend()
plt.show()
```

## Common mix-ups and recap

* A density height is not a probability by itself; an **area over an interval** is a probability.
* Not every real dataset is normal. Always look at the data rather than assuming a bell shape.
* The notation $N(\mu,\sigma^2)$ uses variance ($\sigma^2$), not standard deviation, as the second parameter.

**Remember:** mean chooses the centre; standard deviation chooses the spread.
