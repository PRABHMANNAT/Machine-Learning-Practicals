## When every allowed value gets an equal chance

Uniform means “evenly spread.” There are discrete uniforms, like a fair die, and continuous uniforms, like choosing a random point on a line segment.

## Discrete versus continuous uniform

For a fair six-sided die, the outcomes $1$ through $6$ are equally likely:

$$P(X=x)=\frac16, \qquad x\in\{1,2,3,4,5,6\}$$

For a continuous uniform variable $X\sim U(a,b)$, every **equal-length interval** has the same chance. There are infinitely many possible values between $a$ and $b$, so the probability of one exact value is $0$.


## Continuous uniform formula

For $X\sim U(a,b)$:

$$f(x)=\frac{1}{b-a}\quad\text{for }a\le x\le b$$

$$E[X]=\frac{a+b}{2}, \qquad \operatorname{Var}(X)=\frac{(b-a)^2}{12}$$

The graph is flat because the density stays the same. Probability is still an **area**. Between two values $c$ and $d$ inside the range:

$$P(c\le X\le d)=\frac{d-c}{b-a}$$


## Worked example

Let a bus arrive uniformly at any time from 0 to 30 minutes, so $X\sim U(0,30)$. What is the chance it arrives between minute 10 and minute 18?

$$P(10\le X\le18)=\frac{18-10}{30-0}=\frac8{30}\approx0.267$$

The answer depends on the interval length (8 minutes), not its location. Any 8-minute segment inside the 30-minute window has the same probability.


```python
# Beginner-friendly guide:
# We simulate many random arrival times between 0 and 30 minutes.
# About 8/30 of the simulated times should land between minute 10 and minute 18.
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
arrival_times = rng.uniform(low=0, high=30, size=10_000)
observed = np.mean((arrival_times >= 10) & (arrival_times <= 18))
expected = (18 - 10) / (30 - 0)

print(f"Observed probability: {observed:.3f}")
print(f"Expected probability: {expected:.3f}")
plt.hist(arrival_times, bins=30, density=True, color="mediumseagreen", edgecolor="white")
plt.xlabel("Arrival time in minutes")
plt.ylabel("Density")
plt.title("Random times from a continuous uniform distribution")
plt.show()
```

## Recap

* Discrete uniform: each listed outcome has the same probability.
* Continuous uniform: equal-length intervals have the same probability.
* For continuous variables, do not confuse the flat density height with the probability of one exact point.

**Remember:** uniform means no part of the allowed range is favoured over another equal-size part.
