## Counting how often something happens in a space or time window

The Poisson distribution answers count questions such as: “How many calls arrive in one minute?” or “How many spelling mistakes appear on one page?”

## The child-friendly idea

Imagine raindrops landing on a small square of pavement. We choose a time or area window and count them. The answer can be $0,1,2,3,\ldots$ — there is no fixed maximum beforehand.

The one setting is $\lambda$ (say “lambda”). It is the average number of events in the chosen window.

Write $X\sim\operatorname{Poisson}(\lambda)$.


## Formula and key facts

$$P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}, \qquad k=0,1,2,\ldots$$

Here $k!$ means factorial: $4!=4\times3\times2\times1$.

For a Poisson variable, the average and the variance happen to be the same:

$$E[X]=\lambda, \qquad \operatorname{Var}(X)=\lambda$$

Choose the window carefully. If $\lambda=3$ calls per minute, then a two-minute window has average $6$ calls, assuming the rate stays steady.


## When is Poisson a sensible model?

It works best when events are individual counts, occur independently enough, and have a roughly steady average rate. Examples include website visits per minute or typos per page.

It is different from binomial: binomial has a fixed number of chances; Poisson counts arrivals with no fixed number of chances.

For $λ=3$, the probability of exactly two events is:

$$P(X=2)=\frac{e^{-3}3^2}{2!}\approx0.224$$


```python
# Beginner-friendly guide:
# We compare two average rates. A bigger lambda moves the likely event counts to the right.
# The bars show the chance of seeing 0, 1, 2, and so on events in one chosen window.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

k = np.arange(0, 16)  # Counts that we want to draw on the chart.
for rate, colour in [(2, "teal"), (7, "darkorange")]:
    plt.plot(k, poisson.pmf(k, mu=rate), marker="o", label=f"lambda = {rate}", color=colour)

print(f"P(exactly 2 events when lambda=3) = {poisson.pmf(2, mu=3):.4f}")
plt.xlabel("Number of events")
plt.ylabel("Probability")
plt.title("Poisson probabilities for two average rates")
plt.legend()
plt.show()
```

## Recap

* Poisson counts events in a chosen interval or region.
* $\lambda$ is both the average count and the variance.
* The possible counts start at $0$ and have no fixed upper limit.

**Remember:** Poisson is a “how many arrivals?” model.
