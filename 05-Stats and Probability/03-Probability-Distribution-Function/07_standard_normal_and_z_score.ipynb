## The standard normal distribution

The standard normal is a special normal curve with mean $0$ and standard deviation $1$:

$$Z\sim N(0,1)$$

Its horizontal axis is measured in standard deviations rather than the original unit. A z-score of $0$ is exactly at the average. A z-score of $2$ is two standard deviations above the average.


## The z-score formula

$$z=\frac{x-\mu}{\sigma}$$

* $z>0$: the value is above the mean.
* $z<0$: the value is below the mean.
* $z=0$: the value equals the mean.

Example: a height of $180$ cm in a group with mean $170$ cm and standard deviation $5$ cm has $z=(180-170)/5=2$. It is two standard deviations above that group’s mean.


## Why standardise?

Suppose one student scores $85$ on a test where the mean is $70$ and SD is $10$, while another scores $650$ on a test where the mean is $500$ and SD is $100$.

$$z_1=\frac{85-70}{10}=1.5, \qquad z_2=\frac{650-500}{100}=1.5$$

Their raw scores look different, but both are equally far above their own group averages. A standard normal CDF can turn a z-score into a percentile or probability.


```python
# Beginner-friendly guide:
# First we turn raw scores into z-scores by subtracting the mean and dividing by the standard deviation.
# Then we ask the standard normal curve what fraction of values lie below each z-score.
import numpy as np
from scipy.stats import norm

scores = np.array([55, 70, 85])
mean = 70
sd = 10
z_scores = (scores - mean) / sd  # This is the z-score formula applied to every score.
percentiles = norm.cdf(z_scores) * 100

for score, z, percentile in zip(scores, z_scores, percentiles):
    print(f"Score {score}: z = {z:+.1f}, about the {percentile:.1f}th percentile")
```

```python
# Beginner-friendly guide:
# This chart shows where a z-score of 1.5 sits on the standard normal bell curve.
# The shaded area to the left is the probability of being at or below that z-score.
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 500)
z = 1.5
y = norm.pdf(x)
plt.plot(x, y, color="navy")
plt.fill_between(x, y, where=(x <= z), color="skyblue", alpha=0.7)
plt.axvline(z, color="crimson", linestyle="--", label="z = 1.5")
plt.title("Standard normal curve")
plt.xlabel("z-score")
plt.ylabel("Density")
plt.legend()
plt.show()
```

## Recap and careful language

* A z-score is a relative position, not the original measurement.
* A larger positive z-score means farther above the mean; a more negative one means farther below.
* A z-score is useful only when the mean and standard deviation describe a relevant reference group.

**Remember:** subtract the average, then divide by the usual spread.
