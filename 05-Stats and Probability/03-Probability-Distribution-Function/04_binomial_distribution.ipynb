## The BINS checklist

A binomial random variable counts successes. It needs these four ingredients:

1. **B**inary: every trial is success or failure.
2. **I**ndependent: one trial does not change the next one.
3. Fixed **N**umber of trials: exactly $n$ trials.
4. Same **S**uccess probability: each trial uses the same $p$.

Write $X\sim\operatorname{Binomial}(n,p)$. Its possible values are $0,1,2,\ldots,n$.


## Formula, average, and spread

For exactly $k$ successes:

$$P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}$$

The part $\binom{n}{k}$ counts how many different orders can contain $k$ successes. For example, with three coin tosses, HHT and HTH are different orders but both contain two heads.

$$E[X]=np, \qquad \operatorname{Var}(X)=np(1-p)$$

The expected value is the long-run average count of successes, not a promise for one set of trials.


## Worked example

Suppose a fair coin is tossed $5$ times. What is the chance of exactly $3$ heads?

$$P(X=3)=\binom{5}{3}(0.5)^3(0.5)^2=10\times(0.5)^5=0.3125$$

The chance of **at most** $3$ heads is a cumulative probability:

$$P(X\le3)=P(X=0)+P(X=1)+P(X=2)+P(X=3)$$


```python
# Beginner-friendly guide:
# This code lists the chance of each possible number of heads in five fair tosses.
# scipy.stats.binom already knows the binomial formula, so we can focus on reading the result.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 5
p = 0.5
k = np.arange(n + 1)  # Possible counts: 0, 1, 2, 3, 4, and 5.
probabilities = binom.pmf(k, n=n, p=p)

print(f"P(exactly 3 heads) = {binom.pmf(3, n, p):.4f}")
print(f"P(at most 3 heads) = {binom.cdf(3, n, p):.4f}")

plt.bar(k, probabilities, color="mediumpurple")
plt.xticks(k)
plt.xlabel("Number of heads")
plt.ylabel("Probability")
plt.title("Binomial(5, 0.5): five fair coin tosses")
plt.show()
```

## Common mistakes and recap

* Do not use binomial when the number of trials is not fixed.
* Do not use it if the success chance changes from trial to trial.
* “Exactly $k$” uses the PMF; “at most $k$” adds probabilities and uses the CDF.

**Remember:** binomial = a fixed number of independent Bernoulli trials, counted together.
