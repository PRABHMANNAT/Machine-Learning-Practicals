## Population, sample, parameter, statistic

* **Population:** the full group we want to understand, such as all students at a school.
* **Sample:** the smaller group we actually observe.
* **Parameter:** the true population number, such as population mean $\mu$. It is usually unknown.
* **Statistic:** a number calculated from the sample, such as sample mean $\bar x$.

We use a statistic to estimate a parameter. An **estimator** is the rule or formula; an **estimate** is the number it produces for one particular sample.


## Point estimate versus interval estimate

A **point estimate** gives one best-guess number. For example, a sample mean of $72$ is a point estimate of the population mean.

An **interval estimate** gives a sensible range plus a confidence level. A simple form is:

$$\text{estimate}\ \pm\ (\text{critical value})\times(\text{standard error})$$

For a mean with unknown population standard deviation, a common interval is:

$$\bar x\ \pm\ t^*\frac{s}{\sqrt n}$$

where $s$ is the sample standard deviation and $t^*$ comes from the t distribution.


## What “95% confidence” really means

It does **not** mean there is a 95% probability that the fixed population mean is inside one already-calculated interval. Instead, it describes the method:

If we repeatedly took samples and built intervals the same way, about 95 out of 100 such intervals would capture the true population parameter.

A larger sample usually makes the standard error smaller, which makes the interval narrower. A more demanding confidence level (for example 99%) makes the interval wider.


```python
# Beginner-friendly guide:
# We use a small sample of scores to estimate the unknown average score of the whole population.
# scipy calculates the t-based 95% confidence interval from the sample mean and standard error.
import numpy as np
from scipy import stats

sample = np.array([68, 72, 75, 70, 74, 69, 73, 71, 76, 72])
n = len(sample)
mean = sample.mean()
standard_error = stats.sem(sample)  # How much sample means tend to wiggle from sample to sample.
critical_t = stats.t.ppf(0.975, df=n - 1)  # Middle 95% leaves 2.5% in each tail.
margin_of_error = critical_t * standard_error

print(f"Point estimate (sample mean): {mean:.2f}")
print(f"95% confidence interval: ({mean - margin_of_error:.2f}, {mean + margin_of_error:.2f})")
```

## Recap and good habits

* Samples should be representative; a complicated formula cannot repair a biased sample.
* A point estimate is one value. A confidence interval shows uncertainty around it.
* Bigger samples tend to produce more precise estimates, but only if the collection process is sound.

**Remember:** a sample gives us a careful guess, not perfect certainty.
