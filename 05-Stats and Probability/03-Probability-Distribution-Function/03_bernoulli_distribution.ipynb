## The child-friendly idea

Think of one coin toss. We call the result we care about a **success**. Success does not have to mean “good”; it simply means the outcome we chose to count.

* success → write $X=1$
* not a success (failure) → write $X=0$

So a Bernoulli random variable can only be $0$ or $1$.

Let $p$ be the chance of success. Then the chance of failure is $q=1-p$.


## Formula and useful facts

$$P(X=x)=p^x(1-p)^{1-x}, \qquad x\in\{0,1\}$$

This one compact formula gives both answers:

* if $x=1$, then $P(X=1)=p$
* if $x=0$, then $P(X=0)=1-p$

The average of many repeated $0$s and $1$s is $p$:

$$E[X]=p, \qquad \operatorname{Var}(X)=p(1-p)$$

For example, if a lamp works with probability $0.9$, its expected value is $0.9$. That does **not** mean one lamp gives “0.9 of a working lamp”; it describes the long-run average over many lamps.


## A small worked example

Suppose a quiz question has a $70\%$ chance of being answered correctly. Define $X=1$ for correct and $X=0$ for incorrect.

$$P(X=1)=0.70, \qquad P(X=0)=0.30$$

$$E[X]=0.70, \qquad \operatorname{Var}(X)=0.70\times0.30=0.21$$

Use Bernoulli when there is **one trial** and only two outcomes. If you count successes across many independent trials, you usually need the binomial distribution instead.


```python
# Beginner-friendly guide:
# We make many one-coin experiments. 1 means heads (our success), and 0 means tails.
# The average should get close to p when we repeat the experiment many times.
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)  # A fixed seed makes this example repeatable.
p = 0.70
trials = rng.binomial(n=1, p=p, size=1_000)  # One Bernoulli trial for each result.

print(f"Observed success rate: {trials.mean():.3f}")
print(f"Theoretical success rate: {p:.3f}")

plt.bar([0, 1], [1 - p, p], color=["tomato", "steelblue"])
plt.xticks([0, 1], ["0 = failure", "1 = success"])
plt.ylabel("Probability")
plt.title("A Bernoulli distribution")
plt.show()
```

## Common mix-ups and recap

* A Bernoulli variable is **one** yes/no trial; a binomial variable counts successes in many such trials.
* The values are usually written as $0$ and $1$, so its mean is also the success probability.
* $p$ must be between $0$ and $1$.

**Remember:** one question, two possible labels, one probability $p$.
