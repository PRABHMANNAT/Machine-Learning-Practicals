# Central Limit Theorem (CLT)

## Why averages become bell-shaped

The Central Limit Theorem is one of statistics’ most useful surprises. Even when individual observations are not bell-shaped, the averages of many equally sized samples often become approximately normal.

## The big idea, in tiny steps

1. Start with a population. It might be skewed, bumpy, or far from normal.
2. Take a random sample of size $n$ and calculate its mean.
3. Repeat that sampling many times and collect all those sample means.
4. The distribution of those means is called the **sampling distribution of the mean**.

When $n$ is sufficiently large and samples are independent, that sampling distribution is approximately normal.


## What stays the same and what shrinks

If the population has mean $\mu$ and standard deviation $\sigma$, then the sample mean $\bar X$ has:

$$E[\bar X]=\mu, \qquad \operatorname{SD}(\bar X)=\frac{\sigma}{\sqrt n}$$

The quantity $\sigma/\sqrt n$ is the **standard error**. It gets smaller as sample size grows, so sample means tend to sit closer to the true population mean.

The theorem is about the distribution of **sample means**, not a promise that the raw data itself becomes normal.


## A friendly example

Imagine a population of waiting times that is right-skewed: many short waits and a few long waits. One person’s wait is not normal-looking. But if we repeatedly average 40 people’s waits, those averages make a much more bell-shaped picture.

The CLT helps us make confidence intervals and hypothesis tests because it gives a manageable model for many sample averages.

It is not magic: use random, sufficiently independent observations, and take extra care with tiny samples or extremely unusual distributions.


```python
# Beginner-friendly guide:
# Exponential values are skewed, so the raw observations are not bell-shaped.
# We repeatedly take samples and plot their means to watch the CLT create a bell-like shape.
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
sample_size = 40
number_of_samples = 5_000
sample_means = rng.exponential(scale=1, size=(number_of_samples, sample_size)).mean(axis=1)

print(f"Mean of the sample means: {sample_means.mean():.3f}")
print(f"Standard deviation of sample means: {sample_means.std():.3f}")
plt.hist(sample_means, bins=40, density=True, color="cornflowerblue", edgecolor="white")
plt.xlabel("Sample mean")
plt.ylabel("Density")
plt.title("Many means from samples of 40 skewed observations")
plt.show()
```

## Recap and a common mistake

* The population distribution and the sampling distribution are different things.
* Bigger $n$ makes the standard error smaller by a square-root rule.
* “Large enough” depends on the population shape; $30$ is a common rough starting point, not an unbreakable law.

**Remember:** take many groups, average each group, then look at the averages.
