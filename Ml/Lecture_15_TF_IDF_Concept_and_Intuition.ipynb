# Lecture 15: TF-IDF – Concept, Math, and Intuition


TF-IDF stands for **Term Frequency – Inverse Document Frequency**.  
It is used to convert textual data into numerical vectors by emphasizing important words in a sentence while down-weighting common words.

In this lecture, we’ll cover:
- Term Frequency (TF)
- Inverse Document Frequency (IDF)
- TF-IDF Calculation with example
- Advantages and Disadvantages vs Bag of Words



## 📘 TF-IDF Formulae:

- **Term Frequency (TF)** = (Number of times word appears in a sentence) / (Total words in sentence)  
- **Inverse Document Frequency (IDF)** = log_e (Total number of sentences / Number of sentences containing the word)  
- **TF-IDF** = TF × IDF



### 📝 Example Sentences:

1. `good boy`  
2. `good girl`  
3. `boy girl good`

Vocabulary = `["good", "boy", "girl"]`

We’ll calculate TF and IDF values manually to demonstrate how vectors are created.



### 📊 Term Frequency Table

| Sentence | good      | boy       | girl      |
|----------|-----------|-----------|-----------|
| S1       | 1/2       | 1/2       | 0         |
| S2       | 1/2       | 0         | 1/2       |
| S3       | 1/3       | 1/3       | 1/3       |



### 🧮 Inverse Document Frequency (IDF)

| Word  | IDF Calculation           | Value     |
|-------|---------------------------|-----------|
| good  | log_e(3 / 3) = log(1)     | 0         |
| boy   | log_e(3 / 2)              | ≈ 0.405   |
| girl  | log_e(3 / 2)              | ≈ 0.405   |



### ✅ Final TF-IDF Vectors

| Sentence | TF-IDF (good) | TF-IDF (boy) | TF-IDF (girl) |
|----------|---------------|--------------|---------------|
| S1       | 0             | 0.5 × 0.405  | 0             |
| S2       | 0             | 0            | 0.5 × 0.405   |
| S3       | 0             | 0.333 × 0.405| 0.333 × 0.405 |

This gives meaningful weighting:
- Common words like `good` get ignored.
- Less frequent words like `boy`, `girl` gain importance.



## ✅ Advantages of TF-IDF over Bag of Words:
- Easy to implement and understand.
- Captures **importance** of words.
- Reduces influence of common words (e.g., 'good', 'the').

## ❌ Disadvantages:
- Still sparse (many zero values).
- Suffers from **Out-of-Vocabulary (OOV)** issues (new words get ignored).
- Cannot capture word semantics or relationships.

Despite this, TF-IDF is a strong baseline for many NLP tasks.
