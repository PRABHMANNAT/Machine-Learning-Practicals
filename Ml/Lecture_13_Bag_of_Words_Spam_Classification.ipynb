# Lecture 13: Bag of Words - Spam Classification (NLP Practical)


In this notebook, we will implement the **Bag of Words** model from scratch using a dataset for **spam classification**.  
This lecture includes:
- Data preprocessing
- Text cleaning
- Stemming
- Stopword removal
- Feature extraction with Bag of Words


````python

import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
````

````python
nltk.download('stopwords')
````

````python

# Read dataset with tab separator and custom column names
df = pd.read_csv("spam_classification/SMS_spam_collection.txt", sep='\t', names=['label', 'message'])
df.head()
````

## Text Cleaning & Preprocessing

````python

ps = PorterStemmer()
corpus = []

for i in range(0, len(df)):
    review = re.sub('[^a-zA-Z]', ' ', df['message'][i])
    review = review.lower()
    review = review.split()

    review = [ps.stem(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)

corpus[:5]
````

## Create Bag of Words Model

````python

# Bag of Words with top 100 features and binary format
cv = CountVectorizer(max_features=100, binary=True)
X = cv.fit_transform(corpus).toarray()

# Labels
y = df['label']

X.shape, y.shape
````

````python
pd.DataFrame(X, columns=cv.get_feature_names_out()).head()
````


**Note:**  
- Each row is a message.  
- Each column is a word.  
- `1` indicates the presence of a word, `0` its absence.  
- This is **Binary Bag of Words**.
