# Bag of Words (BoW) in NLP – Super Detailed Implementation

This notebook walks through a full implementation of Bag of Words for spam classification using a real dataset. We’ll:
- Load and inspect the dataset
- Clean and preprocess text (lowercasing, removing special characters, stopword removal, stemming)
- Convert text into vectors using CountVectorizer (Bag of Words)
- Compare different BoW configurations (binary vs normal, max_features)

```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer

# Download stopwords if not already done
nltk.download('stopwords')
```

```python
# Load the dataset (ensure path is correct)
messages = pd.read_csv("spam.csv", sep='\t', names=["label", "message"])
messages.head()
```

### Step 3: Data Cleaning & Preprocessing
We will:
- Remove special characters
- Lowercase text
- Tokenize and remove stopwords
- Apply stemming


```python
ps = PorterStemmer()
corpus = []

for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    review = review.lower().split()
    review = [ps.stem(word) for word in review if word not in stopwords.words('english')]
    corpus.append(' '.join(review))

corpus[:5]
```

### Step 4: Creating Bag of Words using CountVectorizer
- We will use `max_features` to limit vocabulary size
- `binary=True` enables Binary Bag of Words

```python
cv = CountVectorizer(max_features=100, binary=True)
X = cv.fit_transform(corpus).toarray()
y = messages['label']

print(X.shape)
```
