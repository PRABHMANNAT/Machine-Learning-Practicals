### ✅ 2. Compare output between different tokenizers

#### 🔹 `word_tokenize`
- Standard tokenizer from NLTK.
- Separates most words and punctuation.
- Example:
  ```python
  Input: "I'm happy."
  Output: ['I', "'m", 'happy', '.']
  ```

#### 🔹 `WordPunctTokenizer`
- More aggressive: splits on **punctuation too**.
- Example:
  ```python
  Input: "I'm happy."
  Output: ['I', "'", 'm', 'happy', '.']
  ```

#### 🔹 `TreebankWordTokenizer`
- Based on Penn Treebank (used in grammar parsers).
- Smarter with `'s` and punctuation.
- Example:
  ```python
  Input: "She's fine."
  Output: ['She', "'s", 'fine', '.']
  ```

**TL;DR:**

````python
from nltk.tokenize import word_tokenize, WordPunctTokenizer, TreebankWordTokenizer
import nltk
nltk.download('punkt')

text = "She's got 2 cats, and they're awesome!"

# Word Tokenize
print("word_tokenize:", word_tokenize(text))

# WordPunctTokenizer
wp = WordPunctTokenizer()
print("WordPunctTokenizer:", wp.tokenize(text))

# TreebankWordTokenizer
treebank = TreebankWordTokenizer()
print("TreebankWordTokenizer:", treebank.tokenize(text))
````

No, they are not exactly the same, but they often produce the same result.

### `word_tokenize`

```python
word_tokenize(text)
```

- NLTK’s general-purpose tokenizer.
- First detects sentences using Punkt.
- Then tokenizes each sentence using `TreebankWordTokenizer`.
- Best default choice for normal English text.

For your example:

```python
["She", "'s", "got", "2", "cats", ",", "and",
 "they", "'re", "awesome", "!"]
```

### `TreebankWordTokenizer`

```python
TreebankWordTokenizer().tokenize(text)
```

- Directly applies Penn Treebank tokenization rules.
- Handles English contractions such as:

```text
She's  -> She + 's
they're -> they + 're
```

For a single sentence, it gives the same output as `word_tokenize`.

### `WordPunctTokenizer`

```python
WordPunctTokenizer().tokenize(text)
```

- Uses a simpler regular-expression approach.
- Splits punctuation characters individually.
- Therefore:

```text
She's -> She + ' + s
they're -> they + ' + re
```

Output:

```python
["She", "'", "s", "got", "2", "cats", ",", "and",
 "they", "'", "re", "awesome", "!"]
```

### Main difference

| Tokenizer | `"She's"` becomes | Approach |
|---|---|---|
| `word_tokenize` | `["She", "'s"]` | General tokenizer, uses Treebank rules |
| `TreebankWordTokenizer` | `["She", "'s"]` | Linguistic Treebank rules |
| `WordPunctTokenizer` | `["She", "'", "s"]` | Simple regex punctuation splitting |

So, in your example:

```python
word_tokenize(text) == TreebankWordTokenizer().tokenize(text)
```

They match because the input contains only one sentence. However, `word_tokenize` also performs sentence segmentation, while `TreebankWordTokenizer` only tokenizes the text it receives.
