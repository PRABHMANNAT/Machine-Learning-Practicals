# 🧠 NLP Lecture: Lemmatization with NLTK


In this notebook, we explore **Lemmatization**, an important NLP technique used to reduce words to their **root (lemma)** without changing the meaning.

Unlike stemming, lemmatization:
- Returns **real dictionary words**
- Uses the **WordNet corpus**
- Requires **Part-of-Speech (POS)** tagging to work optimally



## 🔍 Difference from Stemming

| Feature         | Stemming             | Lemmatization        |
|------------------|----------------------|-----------------------|
| Output           | Word stem (may not be valid) | Root word (valid dictionary word) |
| Meaning preserved? | ❌ Often no         | ✅ Yes               |
| Uses dictionary? | ❌ No                | ✅ Yes (WordNet)     |
| Example (`goes`) | `goe` ❌             | `go` ✅               |


## 🔧 Using WordNet Lemmatizer

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')  # Optional but recommended for extended support
```

```python

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

### 🔹 Basic Usage

```python
print(lemmatizer.lemmatize("eating"))         # Default is noun → output: eating
print(lemmatizer.lemmatize("eating", "v")) 
# 1st parameter word 2nd parameter pos tag   
print(lemmatizer.lemmatize("eating", "n"))
print(lemmatizer.lemmatize("eating", "a"))
print(lemmatizer.lemmatize("eating", "r"))
```


### 🧠 POS Tags for Lemmatizer
- `'n'` → Noun
- `'v'` → Verb
- `'a'` → Adjective
- `'r'` → Adverb

**Example:** `lemmatizer.lemmatize("going", "v")` → `"go"`


### 🔄 Lemmatize Multiple Words

```python

words = ["eating", "eats", "eaten", "writing", "writes", "programming", "programs", "history", "finally", "finalized"]

# Applying with POS = Verb for best results
for word in words:
    print(f"{word} → {lemmatizer.lemmatize(word, 'v')}")
```

## ⚖️ Fixes Stemming Errors

```python

print(lemmatizer.lemmatize("goes", "v"))         # Expected: go
print(lemmatizer.lemmatize("fairly", "r"))       # Expected: fair
print(lemmatizer.lemmatize("sportingly", "r"))   # May remain same if no root exists
```


## 🧾 Summary

- **Lemmatization** returns **valid root words** using a dictionary (WordNet).
- More accurate than stemming, especially for grammar-sensitive tasks.
- Slower than stemming but **preferred for**:
  - Chatbots
  - Q&A Systems
  - Text Summarization

---

### 💡 Note:
> Lemmatizer takes longer because it compares words against WordNet dictionary using a `morphy()` lookup function.
