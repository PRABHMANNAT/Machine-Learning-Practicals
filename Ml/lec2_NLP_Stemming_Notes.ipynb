# 🧠 NLP Lecture: Stemming with NLTK


This notebook covers the concept of **Stemming** as part of text preprocessing in Natural Language Processing (NLP).  
We'll explore its definition, use-cases, disadvantages, and three types of stemmers:
- Porter Stemmer
- Regexp Stemmer
- Snowball Stemmer



## ✅ What is Stemming?

**Definition** (from Wikipedia):  
Stemming is the process of reducing words to their **root form** or **stem** by chopping off affixes (prefixes or suffixes).  
It helps normalize similar words like `eating`, `eats`, `eaten` → `eat`.

### 🔍 Example:
| Words         | Stem |
|---------------|------|
| eating        | eat  |
| eaten         | eat  |
| goes / going  | go   |
| writes        | write |



## 🎯 Why Use Stemming?

- Reduces vocabulary size
- Improves model generalization
- Removes unnecessary word variations

Helps in tasks like:
- Sentiment classification
- Spam detection
- Search optimization



## ⚠️ Disadvantages of Stemming

- May produce **non-words** (e.g., `history` → `histori`)
- Can **distort meanings**
- Over-aggressive for grammar-sensitive applications like chatbots


## 🔧 1. Porter Stemmer

Porter Stemmer → Removes common endings from English words to get a rough root.
Example: playing → play, studies → studi

Important: Fast and simple, but the result may not be a real word.

```python

from nltk.stem import PorterStemmer

words = ["eating", "eats", "eaten", "writing", "writes", "programming", "programs", "history", "finally", "finalized","ice","icecream"]
ps = PorterStemmer()

for word in words:
    print(f"{word} → {ps.stem(word)}")
```

```python
ps.stem("congratulations")  # Example of stemming a word not in the list
```

## 🔧 2. Regexp Stemmer

Regexp Stemmer → Removes parts of a word using rules that you define with regex.
Example: remove ing → playing → play

Important: Useful when you want full control over which endings/patterns are removed.

```python

from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|ed$|able$', min=4)

for word in words:
    print(f"{word} → {reg_stemmer.stem(word)}")


# The '$' symbol in regex means "end of the word".
# So 'ing$' matches any word that ends with 'ing', like "eating", "playing".
# The pattern 'ing$|s$|ed$|able$' means:
# Remove 'ing', 's', 'ed', or 'able' — *only if they appear at the end* of the word.

# min=4 means: Only apply stemming to words with 4 or more characters after chopping.


```

## 🔧 3. Snowball Stemmer

Snowball Stemmer → A more advanced version of stemming with better rules and support for many languages.
Example: running → run

Important: Usually cleaner and more flexible than Porter Stemmer.

```python

from nltk.stem import SnowballStemmer

snow = SnowballStemmer("english")
#The following languages are supported: Arabic, Danish, Dutch, English, Finnish, French, German, Hungarian, Italian, Norwegian, Portuguese, Romanian, Russian, Spanish and Swedish.
for word in words:
    print(f"{word} → {snow.stem(word)}")
```

## 🔧 Types of Stemming

### 1. **Porter Stemmer**
A rule-based stemmer that removes common morphological and inflectional endings.  
- Simple and fast  
- Often returns non-real words  
- Example: `writing` → `write`, `history` → `histori` ❌

---

### 2. **Regexp Stemmer**
Uses **regular expressions** to remove suffixes or prefixes based on patterns.  
- Highly customizable  
- Depends on manually written rules  
- Example: Pattern `ing$` removes `ing` from `eating` → `eat`

---

### 3. **Snowball Stemmer**
An improved, multilingual version of Porter Stemmer with smarter rules.  
- Produces more accurate stems  
- Supports many languages (e.g., English, French, German)  
- Example: `sportingly` → `sport`, `fairly` → `fair`



## 📊 Comparison

| Stemmer         | Intelligent? | Real Words Output | Multi-Language |
|------------------|--------------|--------------------|----------------|
| Porter Stemmer   | ❌            | Sometimes          | English only   |
| Regexp Stemmer   | ❌ (manual)   | Rarely             | Depends on rule |
| Snowball Stemmer | ✅            | Often              | ✅ Yes         |



## ✅ Summary

- Use **Stemming** to reduce feature space and normalize text.
- For grammar-sensitive tasks, prefer **Lemmatization**.
- Among stemmers:
  - Porter is basic
  - Regex is rule-based
  - Snowball is more accurate and flexible


## ❌ Why Stemming Isn’t Always Enough

While stemming helps reduce word variations and improves model performance in classification tasks, it has **serious limitations**:

---

### 🚨 Problems with Stemming:

1. **Breaks Grammar:**
   - Stemming may produce **non-existent or incorrect words**.
   - Example: `"history"` → `"histori"`, `"congratulations"` → `"congratul"`

2. **Context-Blind:**
   - It blindly chops off suffixes without understanding the meaning.
   - This can lead to **loss of semantic value**.

3. **Snowball Still Fails Sometimes:**
   - Even though **Snowball Stemmer** performs better than Porter, it still gives wrong stems for certain words:
     - `"goes"` → `"goe"` ❌
     - `"does"` → `"doe"` ❌
     - `"history"` → `"histori"` ❌

   So yes, **even the "good guy" messes up** sometimes.

---

## ✅ What To Prefer Instead: **Lemmatization**

**Lemmatization** is smarter.

- It uses a **dictionary** (lexicon) and **part-of-speech tagging** to return the **actual base word (lemma)**.
- It guarantees that the output is a **real, grammatically correct word**.

### 🔍 Example:

| Word           | Lemmatized Output |
|----------------|-------------------|
| goes           | go ✅             |
| better         | good ✅           |
| running        | run ✅            |
| eaten          | eat ✅            |

---

## 🧠 Bottom Line:

> If your task is grammar-sensitive (chatbots, translation, search, summarization) — **use lemmatization**.  
> If you just want to reduce features for classification — **stemming is faster, but riskier**.
