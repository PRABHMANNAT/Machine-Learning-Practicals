# Lecture 19: Word2Vec Architecture – CBOW vs Skip-Gram


In this lecture, we explore the two core architectures behind the **Word2Vec** model:

1. **CBOW (Continuous Bag of Words)**
2. **Skip-Gram**

These models are used to train word embeddings that capture semantic meaning from large text corpora.



## 🧠 1. Continuous Bag of Words (CBOW)

- **Goal**: Predict the target word from a window of surrounding context words.
- **Input**: Multiple context words
- **Output**: One target word

### Example:
Sentence: `The cat sits on the mat`  
If context window = 2 →  
Context: `["The", "cat", "on", "the"]`  
Target: `"sits"`

### Diagram:
```
Context words → [Embedding Layer] → Averaged → [Hidden Layer] → Output (Target Word)
```

### Characteristics:
- Works well for **small datasets**
- Faster to train
- Smoother representations for **frequent words**



## ⚡ 2. Skip-Gram Model

- **Goal**: Predict context words given a single target word.
- **Input**: One word (center word)
- **Output**: Multiple context words

### Example:
Sentence: `The cat sits on the mat`  
Center Word: `"sits"`  
Context Window = 2 →  
Predict: `["The", "cat", "on", "the"]`

### Diagram:
```
Target Word → [Embedding Layer] → [Hidden Layer] → Predict multiple Context Words
```

### Characteristics:
- Better for **rare words**
- More accurate representations
- Slower to train than CBOW



## 📊 CBOW vs Skip-Gram

| Feature                  | CBOW                          | Skip-Gram                     |
|--------------------------|-------------------------------|-------------------------------|
| Input                   | Context Words                 | Target Word                   |
| Output                  | Target Word                   | Context Words                 |
| Training Speed          | Faster                        | Slower                        |
| Works Better On         | Frequent Words                | Rare Words                    |
| Performance             | Smoother                      | More Accurate                 |
| Use Case                | Small datasets                | Large datasets                |



## ✅ Summary

- CBOW: Predicts the word given context
- Skip-Gram: Predicts context given the word
- Both are used to learn efficient word vectors
- Trained via **shallow neural networks**
- Final output: Dense word embeddings
