# Lecture 17: Word Embeddings – Intro to Word2Vec and Techniques


In this lecture, we explore the core concept of **Word Embeddings** – a technique to represent words as vectors in high-dimensional space such that **semantic similarity** is preserved.

We'll also:
- Understand how this differs from earlier approaches (BoW, TF-IDF)
- Classify word embedding techniques
- Preview powerful deep learning-based approaches like **Word2Vec**



## 🧠 What Are Word Embeddings?

> “Word embedding is a type of word representation that allows words with similar meaning to have a similar representation.” — *Wikipedia*

### Key Idea:
- Words are mapped to **dense, real-valued vectors**
- Similar words (like "happy" and "excited") are **close together** in vector space
- Dissimilar or opposite words (like "happy" vs "angry") are far apart



### 📊 Visual Intuition:
Imagine these 2D vector placements (after dimensionality reduction):

- `happy` and `excited` ➝ placed near each other  
- `angry` ➝ far from `happy`

This shows **semantic relationships** are preserved in vector space.



## 🧩 Types of Word Embedding Techniques

### 1. **Frequency-Based (Count-Based)**
- One-Hot Encoding
- Bag of Words
- TF-IDF

Limitations:
- Sparse vectors
- No sense of word meaning
- Cannot capture context

---

### 2. **Prediction-Based (Deep Learning Trained)**
- Word2Vec
  - CBOW (Continuous Bag of Words)
  - Skip-gram
- GloVe
- FastText

These solve the limitations by:
- Learning **dense**, context-aware vectors
- Preserving **semantic and syntactic** meaning



## 🔥 Word2Vec Overview

**Word2Vec** is a deep learning–based word embedding technique that learns word representations efficiently.

It has two architectures:
- **CBOW**: Predicts the target word given the surrounding context
- **Skip-gram**: Predicts context words given a target word

We'll explore both architectures in upcoming lectures.



## ✅ Why Word2Vec?

- Produces **dense vectors** (no sparsity)
- Captures **semantic similarity** (king - man + woman ≈ queen)
- Learns from **context**
- Scalable on large corpora

Unlike one-hot or TF-IDF, it understands that:
- “run” and “jog” are similar
- “apple” and “fruit” are related



## 📌 Summary

- Word embeddings are vector representations of words that preserve meaning.
- BoW and TF-IDF are **basic** embeddings using frequency.
- Word2Vec is a **deep learning** based method that solves many problems.
- CBOW and Skip-gram are two powerful Word2Vec architectures.
- We'll explore these in the next lectures with implementations.
