# 🧠 NLP Notes: Stopwords, Stemming & Lemmatization

In this notebook, we will learn about 3 important **Text Preprocessing** techniques in NLP:
1. **Stopwords Removal** – Remove commonly used words that may not be useful
2. **Stemming** – Reduce words to their base/root form (e.g., "running" → "run")
3. **Lemmatization** – Reduce words to their dictionary form with context (e.g., "better" → "good")

---

## 📌 Why Preprocess Text?

Text data is noisy and full of extra fluff (like “the”, “is”, “have”), so we:
- **Remove Stopwords** to reduce noise.
- **Stem or Lemmatize** to simplify word forms.
- **Convert to lowercase** to avoid duplicates like “India” and “india”.

Let's begin!


## 📦 Step 1: Import Libraries and Download Resources
We’ll use **NLTK** for all our preprocessing tasks.


````python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# Download necessary data
# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('wordnet')
````

## 📄 Step 2: Sample Paragraph

Let’s define a paragraph and break it into sentences.


````python
paragraph = """I have three visions for India. In 3000 years of our history, people from all over the world have come and invaded us, captured our lands, conquered our minds. From Alexander onwards. The Greeks, the Turks, the Moguls, the Portuguese, the British, the French, the Dutch, all of them came and looted us."""
````

## 🪓 Step 3: Sentence Tokenization

We’ll split the paragraph into individual sentences.


````python
sentences = sent_tokenize(paragraph)
print("Tokenized Sentences:")
for sent in sentences:
    print("-", sent)
````

## 🚫 Step 4: Stopwords Removal + Porter Stemmer

We remove common words like “the”, “is”, “of” and apply basic stemming using `PorterStemmer`.


````python
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    filtered = [ps.stem(word.lower()) for word in words if word.lower() not in stop_words]
    sentences[i] = ' '.join(filtered)

print("After Stopword Removal + Porter Stemming:")
for sent in sentences:
    print("-", sent)
````

## ❄️ Step 5: Stopwords Removal + Snowball Stemmer

Let’s reset the paragraph and apply a more modern stemmer – `SnowballStemmer`.


````python
sentences = sent_tokenize(paragraph)
ss = SnowballStemmer("english")

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    filtered = [ss.stem(word.lower()) for word in words if word.lower() not in stop_words]
    sentences[i] = ' '.join(filtered)

print("After Snowball Stemming:")
for sent in sentences:
    print("-", sent)
````

This code removes stopwords and applies **Snowball Stemming** to every sentence.

Creates an English Snowball Stemmer. It reduces words to shorter root forms:


Snowball Stemmer is generally an improved version of Porter Stemmer.

## 📚 Step 6: Lemmatization

Now we lemmatize each word to its dictionary form (e.g., “running” → “run”).


````python
sentences = sent_tokenize(paragraph)
lemmatizer = WordNetLemmatizer()

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    filtered = [lemmatizer.lemmatize(word.lower()) for word in words if word.lower() not in stop_words]
    sentences[i] = ' '.join(filtered)

print("After Lemmatization:")
for sent in sentences:
    print("-", sent)
````

This code cleans the paragraph using **lemmatization**.

```python
sentences = sent_tokenize(paragraph)
```

Breaks the paragraph into separate sentences.

```python
lemmatizer = WordNetLemmatizer()
```

Creates a lemmatizer. It changes words to their proper dictionary form.

Examples:

```text

```

```python
for i in range(len(sentences)):
```

```python

```

Breaks the current sentence into separate words.

```python
filtered = [
    lemmatizer.lemmatize(word.lower())
    for word in words
    if word.lower() not in stop_words
]
```

For each word:

1. Converts it to lowercase.
2. Removes common words such as `the`, `is`, `and`, and `have`.
3. Changes the word to its dictionary form.

Example:

```text

```

```text

```

```python

```

Joins the cleaned words back into a sentence.


Prints a heading.


Prints each cleaned sentence with a dash.

**In one simple sentence:**  
This code separates the paragraph into sentences, removes unnecessary words, changes words to their proper dictionary form, and prints the result.

**Difference from stemming:**  
Stemming simply cuts words, sometimes producing incomplete words. Lemmatization tries to produce a real dictionary word.

## 🔠 Step 7: Lowercase Conversion

To make our model treat “India” and “india” as the same word, we lowercase everything.


````python
sentences = [sent.lower() for sent in sentences]

print("Lowercased Sentences:")
for sent in sentences:
    print("-", sent)
````

## ✅ Summary

| Technique      | Output Form       | Handles Grammar | Speed   |
|----------------|-------------------|------------------|---------|
| Stopwords      | Removes common words | ❌               | Fast    |
| Porter Stemmer | Cuts word endings | ❌               | Fast    |
| Snowball Stemmer | Modern stemmer  | ❌               | Fast    |
| Lemmatizer     | Dictionary root   | ✅ (uses context) | Slower  |

Use **Lemmatization** when meaning matters.  
Use **Stemming** when speed matters.  
Always **remove stopwords** for cleaner data.
