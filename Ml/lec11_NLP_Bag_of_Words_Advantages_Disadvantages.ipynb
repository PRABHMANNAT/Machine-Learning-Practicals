# Bag of Words – Advantages and Disadvantages

## ✅ Advantages of Bag of Words (BoW)
1. **Simple and Intuitive**
   - Very easy to understand and implement.
   - Works well as a baseline for text classification.

2. **Fixed-Length Feature Vector**
   - Regardless of the input sentence length, all vectors have the same size based on vocabulary.
   - ML models expect fixed input sizes — BoW fits perfectly.

## ❌ Disadvantages of Bag of Words (BoW)

1. **Sparse Matrix Problem**
   - Vocabulary can be huge (e.g., 50,000+).
   - Most vectors contain lots of 0s ➝ inefficient and may lead to **overfitting**.

2. **Loss of Word Order (No Context)**
   - Sentences like `"Boy is good"` and `"Good is boy"` generate the same BoW vector.
   - Word position is ignored ➝ **semantic meaning** lost.

3. **Out-of-Vocabulary (OOV) Words**
   - New words in test data (e.g., `"school"`) that weren't in the training vocab get ignored.
   - Can impact prediction quality.

4. **Lack of Semantic Information**
   - Words are binary (0 or 1) or frequency counts.
   - No meaning captured — e.g., `"awesome"` and `"good"` are treated as completely different.

5. **Poor at Handling Negation**
   - Sentences: `"food is good"` vs. `"food is not good"` can have nearly the same BoW vector.
   - Cosine similarity may wrongly show them as **similar**, though their meanings are opposite.

---

## 📌 Code Demo: BoW with Sklearn
Let's practically implement it to visualize the issues.


```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Example sentences
corpus = [
    "He is a good boy",
    "She is a good girl",
    "Boy and girl are good",
    "The food is not good"
]

# Vectorizer instance
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(corpus)

# Vocabulary and matrix
pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
```

📌 Even though `"not"` is a critical word, BoW treats it as just another word.
No context is captured — that's why embeddings like Word2Vec or transformers outperform BoW.
