# Lecture 21: Advantages of Word2Vec

## 1. Dense Matrix Representation
- Unlike Bag of Words or TF-IDF which produce **sparse matrices** (with many 0s), Word2Vec produces **dense vectors**.
- Dense vectors are more efficient and effective for ML models, leading to **better generalization** and **less overfitting**.

## 2. Semantic Similarity
- Word2Vec captures **semantic information**.
- Words like _honest_ and _good_ will have **similar vector representations**.
- You can compute **cosine similarity** to find similar words.

## 3. Fixed Dimensionality
- In Bag of Words / TF-IDF, the vector size depends on **vocabulary size**.
- In Word2Vec, the vector has a **fixed size** (e.g., 300-dim) regardless of corpus size.
- E.g., Google News Word2Vec → trained on 3 billion words → outputs **300-dim vectors**.

## 4. Out-of-Vocabulary Handling
- Word2Vec minimizes **OOV (out-of-vocabulary)** issues because it learns **contextual embeddings**.
- Pretrained models cover vast vocabulary from general domains.

## 5. Summary
- Word2Vec outperforms traditional techniques in **efficiency**, **semantic understanding**, and **generalization**.
- It forms the basis for advanced NLP tasks like **text classification**, **semantic search**, and **chatbots**.
