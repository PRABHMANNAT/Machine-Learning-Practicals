# Lecture 20: Skip-Gram Architecture in Word2Vec


In this lecture, we’ll take a deep dive into the **Skip-Gram** architecture in Word2Vec.  
We’ll explore the working, training process, and when to prefer Skip-Gram over CBOW.



## 🧠 Skip-Gram Architecture: Core Idea

- **Input**: One word (center word)
- **Output**: Predict surrounding context words
- **Works Best**: For large datasets and rare words

---
### 📌 Example Sentence
`Enron company is related to data science`

Suppose center word = `"related"`, window size = 2

Then training samples will be:

| Input     | Outputs                  |
|-----------|--------------------------|
| related   | company, is, to, data    |

---
### 🔧 Neural Network Structure

- **Input Layer**: One-hot vector of size = vocabulary size (e.g., 7)
- **Hidden Layer**: Weight matrix (V × N) → compresses input to embedding size (e.g., 7×5)
- **Output Layer**: Weight matrix (N × V), uses Softmax to predict context words

Each training pair (center word, context word) is trained independently in Skip-Gram.

## ⚙️ Training Details

1. **Input**: One-hot encoded vector of center word
2. **Hidden Layer**: Matrix multiplication → vector representation (size = embedding dim)
3. **Output Layer**: Multiply with second weight matrix, apply softmax
4. **Loss**: Cross-entropy between predicted context and actual context
5. **Backpropagation**: Update both weight matrices until loss minimized

---
### 📐 Matrix Dimensions (Example)

- Vocabulary size = 7
- Embedding size = 5

| Layer            | Matrix Shape       |
|------------------|--------------------|
| Input to Hidden  | 7 × 5              |
| Hidden to Output | 5 × 7              |



## 📌 When to Use Skip-Gram?

| Scenario                   | Recommendation |
|----------------------------|----------------|
| Large dataset              | ✅ Use Skip-Gram |
| Rare words important       | ✅ Use Skip-Gram |
| Faster training needed     | ❌ Prefer CBOW |
| High precision embeddings  | ✅ Use Skip-Gram |



## 🛠️ How to Improve Skip-Gram / CBOW?

- **Increase training data** – more context helps!
- **Increase window size** – captures more semantic relations
- **Use higher dimensional vectors** (e.g., 100 → 300 dims)
- **Use pre-trained vectors** like Google News (3 billion words, 300 dims)

Example:
`cricket` → 300-dimensional dense vector from Google News pretrained Word2Vec



## ✅ Summary

- Skip-Gram flips CBOW: one word in → many context words out
- Works well on **large corpora**, handles **rare words** better
- Training uses simple feedforward network with softmax & cross-entropy
- Better performance with **more data** and **larger window sizes**
- Pretrained models are widely used in real-world applications
