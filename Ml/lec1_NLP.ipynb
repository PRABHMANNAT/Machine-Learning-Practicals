# NLP Basic Terminologies — Super Easy Notes

## 1. Corpus

A **corpus** means the complete text/paragraph we are working with.

Example:

`I like NLP. I am learning Machine Learning.`

The whole paragraph = **Corpus**

---

## 2. Document

A **document** usually means one sentence inside the corpus.

Example:

Corpus:
`I like NLP. I am learning Machine Learning.`

Documents:

* `I like NLP.`
* `I am learning Machine Learning.`

So:

**Corpus = collection of documents/sentences**

---

## 3. Words

All individual words present in the text.

Example:

`I like NLP`

Words:

* I
* like
* NLP

---

## 4. Vocabulary

**Vocabulary = all unique words in the corpus.**

Example:

`I like juice. I drink juice.`

All words:
`I, like, juice, I, drink, juice`

Unique words:
`I, like, juice, drink`

So:

**Vocabulary size = 4**

---

# Tokenization

## What is Tokenization?

**Tokenization = breaking text into smaller parts called tokens.**

Tokens can be:

* Sentences
* Words

---

## Sentence Tokenization

Convert a paragraph into sentences.

Example:

`My name is Krish. I am a YouTuber.`

After tokenization:

* `My name is Krish.`
* `I am a YouTuber.`

Each sentence = **Token**

---

## Word Tokenization

Convert a sentence into words.

Example:

`My name is Krish`

After tokenization:

* My
* name
* is
* Krish

Each word = **Token**

---

# Important Difference

| Term             | Meaning                               |
| ---------------- | ------------------------------------- |
| **Corpus**       | Complete paragraph/text               |
| **Document**     | One sentence                          |
| **Word**         | Individual word                       |
| **Vocabulary**   | Unique words                          |
| **Tokenization** | Breaking text into sentences or words |
| **Token**        | Result after tokenization             |

---

## Easy Example

Text:

`I like apple juice. My friend likes mango juice.`

### Corpus

Entire text.

### Documents

1. `I like apple juice.`
2. `My friend likes mango juice.`

### Words

All individual words.

### Vocabulary

All unique words only.

`juice` appears twice, but is counted **once** in vocabulary.

---

# Why Tokenization?

NLP models need to process text step by step.

So we break:

**Paragraph → Sentences → Words → Vectors**

Later, these words are converted into numerical vectors for Machine Learning.

---

## One-Line Revision

> **Corpus = full text, Document = sentence, Vocabulary = unique words, Tokenization = splitting text into tokens.**

We’ll use the `nltk` library for tokenization.

````python
import sys

# Clear any stale partially initialized NLTK state from earlier failed runs.
for name in list(sys.modules):
    if name == "nltk" or name.startswith("nltk."):
        del sys.modules[name]

import nltk

for package in ["punkt", "punkt_tab"]:
    try:
        nltk.data.find(f"tokenizers/{package}")
    except LookupError:
        nltk.download(package, quiet=True)

print("NLTK tokenizer data is ready.")
````

## 📝 Sample Corpus

````python
corpus = '''Hello, welcome to Krish Naik NLP tutorials.
Please do watch the entire course to become expert in NLP!'''
print(corpus)
````

## ✂️ Sentence Tokenization

````python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(corpus)
print("Sentence Tokenization:")
print(sentences)
````

## ✂️ Word Tokenization

````python
from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)
print("Word Tokenization:")
print(words)
````

## 🪓 WordPunctTokenizer

Splits text into words + punctuation separately. Example: "Hello, world!" → ['Hello', ',', 'world', '!']

here below 's is  spearately tokenized

````python
from nltk.tokenize import WordPunctTokenizer

wp_tokenizer = WordPunctTokenizer()
wp_words = wp_tokenizer.tokenize(corpus)
print("WordPunctTokenizer:")
print(wp_words)
````

## 🌲 TreebankWordTokenizer

Splits text into words using smarter English rules, keeping some punctuation/contractions more naturally grouped.
no sepearate 's

````python
from nltk.tokenize import TreebankWordTokenizer

treebank = TreebankWordTokenizer()
treebank_words = treebank.tokenize(corpus)
print("TreebankWordTokenizer:")
print(treebank_words)
````

## 📌 Summary Table

| Tokenizer              | Sentence Tokenization | Word Tokenization | Special Handling          |
|------------------------|------------------------|--------------------|----------------------------|
| `sent_tokenize()`      | ✅                     | ❌                | Handles `.`, `!`, `?`     |
| `word_tokenize()`      | ❌                     | ✅                | Standard punctuation split |
| `WordPunctTokenizer()` | ❌                     | ✅                | Splits contractions, punctuation |
| `TreebankTokenizer()`  | ❌                     | ✅                | Handles quotes/apostrophes better |
