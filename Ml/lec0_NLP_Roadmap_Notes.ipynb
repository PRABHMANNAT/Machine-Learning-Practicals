# NLP Introduction


## Topic

### 1. NLP Introduction
NLP allows computers to process and understand human language such as English text, emails, commands, and conversations.

### 32 Problem with Text Data
Machine Learning models cannot directly understand raw text. Words and sentences must first be converted into numerical representations before they can be processed by an ML model.

### 3. Spam Classification Example
A classic NLP problem is:

Email subject + body -> Spam / Ham

This is a classification task where the output belongs to a predefined category.

### 4. Text → Vectors
The core idea of NLP is to transform text into meaningful numerical vectors that machine learning and deep learning models can process.

### 5. Text Preprocessing – Part 1
This step includes cleaning text using:
- Tokenization
- Stemming
- Lemmatization
- Stopwords

### 6. Text Preprocessing – Part 2
Basic text vectorization techniques include:
- Bag of Words
- TF-IDF
- Unigrams
- Bigrams / N-grams

### 7. Text Preprocessing – Part 3
More advanced methods include:
- Word2Vec
- Average Word2Vec

### 8. Word Embeddings
Word embeddings are dense vector representations where words with similar meanings tend to have similar numerical values.

### 9. Deep Learning for NLP
The lecture introduces deep learning models such as:
- RNN
- LSTM
- GRU

### 10. Transformers
Transformers are introduced as a more advanced architecture for sequential and language-based tasks.

### 11. BERT
BERT is mentioned as a powerful transformer-based model used in modern NLP.

### 12. NLP Libraries
The lecture highlights two major groups of tools:
- Traditional NLP: NLTK, spaCy
- Deep-learning NLP: TensorFlow, PyTorch

### 13. NLP Applications
Some common real-world NLP applications are:
- Spam detection
- Text summarization
- Voice assistants such as Alexa and Google Assistant
- Command understanding
