# 🧠 NLP Project Lifecycle – Problem to Model (with Preprocessing Recap)

In this notebook, we revise everything we've learned so far in NLP using **sentiment analysis** as an example use case.

We’ll break down how:
- Tokenization
- Lowercasing
- Stopwords
- Stemming & Lemmatization
fit into a **real-world NLP pipeline**.

---



## 🧪 Problem Statement: Sentiment Analysis

We have a dataset with sentences/documents (e.g., reviews, tweets), and our task is to classify them as **positive** or **negative**.

Example:
```text
D1: "The movie was fantastic!" → Positive
D2: "I hated the acting." → Negative
```

---

## 🧱 Structure of NLP Dataset

| Term        | Meaning                                      |
|-------------|----------------------------------------------|
| Document    | One sentence or entry (e.g., a review)       |
| Corpus      | All documents combined (entire dataset)      |
| Vocabulary  | All unique words in the corpus               |

---

## 🔁 NLP Project Pipeline Breakdown

Let’s see where each step fits into solving the problem.



## 🔹 Step 1: Raw Data

We start with a dataset of sentences and labels:
```text
Text → ["I loved the movie", "The food was awful", ...]
Label → ["Positive", "Negative", ...]
```

---

## 🧹 Step 2: Text Preprocessing – Part 1

### ✅ What we do:
1. **Tokenization** – Break paragraph into sentences or words
2. **Lowercasing** – Convert ALL words to lowercase (e.g., `The` and `the` are treated the same)
3. **Regular Expression Cleaning** – Remove:
   - Special characters (`@#$%^&*`)
   - Numbers (optional)
   - Extra whitespace, etc.

➡️ This is our **first cleaning step** to prepare uniform input.

---

## 🔧 Step 3: Text Preprocessing – Part 2

### ✅ What we do:
4. **Stopwords Removal** – Delete words like `the`, `is`, `and`, `to`, `of` etc.
5. **Stemming** – Reduce words to root form (e.g., `running` → `run`)
6. **Lemmatization** – Similar to stemming, but more accurate (uses context)

These help reduce vocabulary size and improve focus on meaningful words.

---

## 🔢 Step 4: Convert Text to Vectors

Now that text is cleaned, we need to **convert it to numbers** so ML models can understand it.

### Techniques:
| Method             | Description |
|--------------------|-------------|
| One-Hot Encoding   | Basic binary representation |
| Bag of Words (BoW) | Word count per sentence     |
| TF-IDF             | Frequency weighted by importance |
| Word2Vec           | Deep learning–based word vectors |
| Avg Word2Vec       | Sentence vector using average word vectors |

➡️ These are vectorization methods — each one turns text into a **vector of numbers**.

---

## 🧠 Step 5: Model Training

Once vectors are ready:
- Feed them into a **machine learning model**
- Example models: Logistic Regression, SVM, Naive Bayes
- Train the model on text vectors and their labels

---

## 📈 Step 6: Prediction and Evaluation

Use test data to:
- Make predictions
- Evaluate accuracy/precision/recall

---

## 📌 Final NLP Pipeline (Overview Diagram)

```
[Raw Text Data] 
      ↓
[Text Preprocessing Part 1]
(Tokenization, Lowercasing, Regex)
      ↓
[Text Preprocessing Part 2]
(Stopwords, Stemming, Lemmatization)
      ↓
[Text to Vectors]
(BoW / TF-IDF / Word2Vec)
      ↓
[Machine Learning Model]
(Logistic Regression, etc.)
      ↓
[Predictions]
```
