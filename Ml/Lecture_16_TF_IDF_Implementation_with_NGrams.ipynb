# Lecture 16: TF-IDF – Python Implementation with Lemmatization and N-Grams


In this notebook, we’ll implement the **TF-IDF Vectorizer** using Python and `scikit-learn` for spam classification.  
We'll also use **Lemmatization** instead of stemming for preprocessing and explore **n-grams** with TF-IDF.


```

import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
nltk.download('stopwords')
nltk.download('wordnet')
```

## Step 1: Load and Inspect the Dataset

```

# Assuming file is tab-separated with no headers
df = pd.read_csv("spam_classification/SMS_spam_collection.txt", sep='\t', names=["label", "message"])
df.head()
```

## Step 2: Clean Text Data using Lemmatization

```

lemmatizer = WordNetLemmatizer()
corpus = []

for i in range(len(df)):
    review = re.sub('[^a-zA-Z]', ' ', df['message'][i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review if word not in stopwords.words('english')]
    corpus.append(' '.join(review))

corpus[:5]
```

## Step 3: Apply TF-IDF Vectorizer (Unigrams Only)

```

tfidf = TfidfVectorizer(max_features=100)
X_uni = tfidf.fit_transform(corpus).toarray()

# Display feature names and first few vectors
pd.DataFrame(X_uni, columns=tfidf.get_feature_names_out()).head()
```

## Step 4: TF-IDF with Bigrams (n_gram_range=(2,2))

```

tfidf_bigram = TfidfVectorizer(max_features=100, ngram_range=(2,2))
X_bi = tfidf_bigram.fit_transform(corpus).toarray()

# Display feature names and vectors
pd.DataFrame(X_bi, columns=tfidf_bigram.get_feature_names_out()).head()
```

## Step 5: TF-IDF with Unigram + Bigram (n_gram_range=(1,2))

```

tfidf_uni_bi = TfidfVectorizer(max_features=100, ngram_range=(1,2))
X_uni_bi = tfidf_uni_bi.fit_transform(corpus).toarray()

# Show sample features and vectors
pd.DataFrame(X_uni_bi, columns=tfidf_uni_bi.get_feature_names_out()).head()
```


### 🧠 Key Takeaways:
- **Lemmatization** improves word normalization compared to stemming.
- `TfidfVectorizer` automatically calculates TF × IDF for each term.
- You can use `ngram_range=(1, 2)` to include both unigrams and bigrams.
- Output vectors are dense with decimal values (word importance), unlike binary Bag of Words.
