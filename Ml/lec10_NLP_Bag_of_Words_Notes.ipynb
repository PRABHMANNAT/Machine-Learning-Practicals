
# Bag of Words (BoW) – NLP Vectorization Technique

In this notebook, we will explore the **Bag of Words (BoW)** method of converting textual data into numerical vectors. This is crucial for machine learning models to understand and process natural language data.

---

### 🎯 Objective

We aim to transform the following **3 positive sentences** into numerical vectors using the BoW method:

```
1. He is a good boy
2. She is a good girl
3. Boy and girl are good
```

All of them have a positive sentiment, so output = 1 for each.



---

## 🔧 Step 1: Text Preprocessing

We convert all words to **lowercase** and **remove stopwords** (e.g., "he", "is", "a", "she", "and", "are").

### Resulting cleaned sentences:
```
1. good boy
2. good girl
3. boy girl good
```



---

## 🧠 Step 2: Vocabulary Building

Now, we extract **unique words** from all cleaned sentences:

| Word  | Frequency |
|-------|-----------|
| good  | 3         |
| boy   | 2         |
| girl  | 2         |

Vocabulary size = **3** → `["good", "boy", "girl"]`



---

## 🔢 Step 3: Vectorizing Sentences (BoW)

### BoW Matrix:

| Sentence          | good | boy | girl |
|-------------------|------|-----|------|
| good boy          |  1   |  1  |  0   |
| good girl         |  1   |  0  |  1   |
| boy girl good     |  1   |  1  |  1   |

### 👉 Output Labels:
```
[1, 1, 1]
```



---

## 🧮 Step 4: Binary vs Count-Based BoW

- **Binary BoW**: Only 1 or 0 (whether word exists)
- **Count BoW**: Uses **actual word frequency**

### Example:
Sentence: `good good girl`

- **Binary**: `[1, 0, 1]`
- **Count**:  `[2, 0, 1]`


**Bag of Words (BoW)** converts text into numbers based on the words it contains.

### Example sentences

```text
Sentence 1: "cat cat dog"
Sentence 2: "cat bird"
```

Vocabulary:

```text
["cat", "dog", "bird"]
```

### 1. Normal Count Bag of Words

Counts how many times each word appears.

| Sentence | cat | dog | bird |
|---|---:|---:|---:|
| cat cat dog | 2 | 1 | 0 |
| cat bird | 1 | 0 | 1 |

Here, the repeated word `cat` gets a higher value.

### 2. Binary Bag of Words

Only checks whether a word is present or absent.

- `1` means the word is present.
- `0` means the word is absent.

| Sentence | cat | dog | bird |
|---|---:|---:|---:|
| cat cat dog | 1 | 1 | 0 |
| cat bird | 1 | 0 | 1 |

Even though `cat` appears twice, binary BoW records it only as `1`.

### Main Difference

| Feature | Count BoW | Binary BoW |
|---|---|---|
| Measures | Word frequency | Word presence |
| Repeated words | Increases the value | Ignored |
| Values | `0, 1, 2, 3...` | Only `0` or `1` |
| Useful when | Repetition is important | Only word occurrence matters |

**Simple meaning:** Count BoW asks, “How many times does this word appear?” Binary BoW asks, “Does this word appear at least once?”
