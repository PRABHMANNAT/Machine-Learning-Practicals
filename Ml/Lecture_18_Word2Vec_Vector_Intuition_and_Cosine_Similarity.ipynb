## 🧠 What is Word2Vec?

- A deep learning–based word embedding technique introduced by **Google in 2013**
- Converts each word into a **dense vector** (typically 100–300 dimensions)
- Similar words → close vectors  
- Opposite words → distant vectors  
- Supports **synonym detection**, **context-aware suggestions**, and more



### 📊 Word Vector Analogy:

If "king", "man", and "queen" have the following vectors:

- king = [0.95, 0.96]
- man  = [0.95, 0.98]
- queen = [0.96, 0.97]

Then:

**king − man + queen ≈ woman**

This vector arithmetic showcases **semantic relationships** that are **preserved geometrically**.



## 🔎 Feature Representation

Each word is mapped to a vector across **learned latent dimensions** (300 in Google's model). These features are not interpretable but may reflect abstract attributes like:

- Gender
- Royalty
- Age
- Food category

For example:

| Word   | Gender | Royalty | Food | Age  | ... |
|--------|--------|---------|------|------|-----|
| Boy    | -1     | 0.01    | 0.00 | 0.03 | ... |
| Girl   | +1     | 0.02    | 0.00 | 0.02 | ... |
| King   | -0.92  | 0.95    | 0.00 | 0.75 | ... |
| Queen  | +0.93  | 0.96    | 0.00 | 0.68 | ... |
| Apple  | 0.01   | 0.00    | 0.91 | 0.82 | ... |
| Mango  | 0.02   | 0.00    | 0.92 | 0.79 | ... |

Word2Vec **learns these embeddings from large corpora** (e.g. Google News, 3 billion words).



## 📐 Cosine Similarity

To compare vectors, we use **cosine similarity**, which measures the **angle** between two vectors.

- **cos(0°) = 1** → vectors are identical
- **cos(90°) = 0** → vectors are orthogonal (unrelated)
- **cos(180°) = -1** → vectors are opposites

### 🧮 Cosine Distance = `1 - cosine_similarity(v1, v2)`

If distance ≈ 0 → words are similar  
If distance ≈ 1 → words are dissimilar

Used in:
- **Recommendation systems**
- **Semantic search**
- **Analogy solving**



## 🧠 Summary

- Word2Vec maps each word to a dense vector using **CBOW** or **Skip-gram**
- These vectors capture **meaning, context, and relationships**
- Enables **vector algebra** (e.g., king − man + woman ≈ queen)
- Cosine similarity quantifies vector similarity
- Outperforms BoW and TF-IDF by avoiding sparsity and encoding semantics
