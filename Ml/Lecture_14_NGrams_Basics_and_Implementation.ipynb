# Lecture 14: N-Grams in NLP – Concepts and Implementation


In this notebook, we dive into **N-Grams**, a powerful technique to capture the contextual meaning of text by considering consecutive word combinations.

We'll cover:
- What N-Grams are and why they're important
- Difference between unigram, bigram, trigram
- Practical usage with `CountVectorizer` from `sklearn`
- Experimenting with `n_gram_range` and `max_features`


```python

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import nltk
nltk.download('stopwords')
```

## Load Preprocessed Corpus

```python

# Load the pre-cleaned corpus from the previous lecture
with open("spam_classification_corpus.txt", "r") as f:
    corpus = f.read().splitlines()

# Confirm sample
corpus[:5]
```

## Unigram Model (n_gram_range=(1,1))

```python

cv_unigram = CountVectorizer(max_features=100, ngram_range=(1,1), binary=True)
X_uni = cv_unigram.fit_transform(corpus).toarray()

# Show top features
cv_unigram.vocabulary_
```

## Unigram + Bigram (n_gram_range=(1,2))

```python

cv_uni_bi = CountVectorizer(max_features=200, ngram_range=(1,2), binary=True)
X_unibi = cv_uni_bi.fit_transform(corpus).toarray()

# Sample vocabulary
list(cv_uni_bi.vocabulary_.items())[:15]
```

## Bigram + Trigram (n_gram_range=(2,3))

```python

cv_bi_tri = CountVectorizer(max_features=200, ngram_range=(2,3), binary=True)
X_bitri = cv_bi_tri.fit_transform(corpus).toarray()

# Sample features
list(cv_bi_tri.vocabulary_.items())[:15]
```

## Unigram + Bigram + Trigram (n_gram_range=(1,3))

```python

cv_uni_bi_tri = CountVectorizer(max_features=500, ngram_range=(1,3), binary=True)
X_all = cv_uni_bi_tri.fit_transform(corpus).toarray()

# Sample features
list(cv_uni_bi_tri.vocabulary_.items())[:15]
```


### 🔍 Why N-Grams?
N-Grams help identify **contextual relationships** between words.

Example:

- `"The food is good"` vs `"The food is not good"`

Without bigrams, both appear similar.  
With bigrams like `"not good"`, we capture that one is positive and the other is negative.

Use different `ngram_range` to explore:
- `(1, 1)` = unigram only
- `(1, 2)` = unigram + bigram
- `(1, 3)` = unigram + bigram + trigram
- `(2, 3)` = bigram + trigram only



### ✅ Tips:
- Start with `(1,1)` and analyze accuracy.
- If poor, increase to `(1,2)` or `(1,3)`.
- Tweak `max_features` as a hyperparameter.
