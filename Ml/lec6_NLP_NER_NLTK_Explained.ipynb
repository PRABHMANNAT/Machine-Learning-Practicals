# 🧠 NLP Lecture: Named Entity Recognition (NER) using NLTK

This notebook will help you understand **Named Entity Recognition (NER)** in Natural Language Processing using NLTK.  
NER helps in identifying **person names, locations, organizations, dates, money , Organization , Percent**, etc., from raw text.

We'll go step-by-step through:
- Tokenization
- POS tagging
- Applying `ne_chunk` for NER
- Visualizing named entities


```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.tree import Tree

# 📦 Required downloads
# nltk.download('punkt')
# nltk.download('maxent_ne_chunker')
# nltk.download('words')
```

```python
# 📝 Sample sentence
sentence = "The Eiffel Tower was built from 1887 to 1889 by French engineer Gustave Eiffel, whose company specialized in building metal frameworks and structures."

# Step 1: Tokenize the sentence
tokens = word_tokenize(sentence)
print(tokens)
```

```python
# Step 2: Part-of-Speech Tagging
tagged_tokens = pos_tag(tokens)
print(tagged_tokens)
```

```python
# Step 3: Named Entity Recognition (NER)
ner_tree = ne_chunk(tagged_tokens)
print(ner_tree)
```

```python
# Step 4: Visualize the named entity tree (only works in Jupyter Notebook)
ner_tree.draw()
```

```python
print(ne_chunk(pos_tag(word_tokenize(sentence)))) same simple from above code
```

## 🔍 NER Output Explanation

The output of `ne_chunk()` is a tree structure where:
- `PERSON` → Names of people (e.g., *Gustave Eiffel*)
- `GPE` → Geopolitical entity (e.g., *France*)
- `ORGANIZATION` → Names of companies, institutions
- `DATE` / `TIME` / `MONEY` → Self-explanatory
- Non-entities just appear as `(word, POS)` tuples.

The `.draw()` method opens a GUI window to visualize the entity tree.
