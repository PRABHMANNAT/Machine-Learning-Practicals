## ✍️ Why is POS Tagging Important?

POS tagging helps NLP systems understand **what role a word plays** in a sentence:  
- Noun, verb, adjective, pronoun, etc.  
- Especially useful in **lemmatization** (to get correct root forms).

Example:
- "running" (verb) → "run"
- "running shoes" (adjective) → stays "running"

If you mess up the tag, you mess up the meaning.

---

## 🔠 Common POS Tags

| Tag | Meaning | Example |
|-----|---------|---------|
| NN | Noun, singular | "monument" |
| NNS | Noun, plural | "monuments" |
| VB | Verb, base form | "run" |
| VBD | Verb, past tense | "ran" |
| JJ | Adjective | "beautiful" |
| RB | Adverb | "quickly" |
| PRP | Personal pronoun | "he", "she" |
| IN | Preposition | "in", "on", "at" |
| DT | Determiner | "the", "a" |

> 🔗 Full list: [NLTK Tag Documentation](https://www.ling.upenn.edu/courses/Fall_2003/ling001/penn_treebank_pos.html)


````python
import nltk
nltk.download('all')
````

````python
# Sample Paragraph (from Dr. A.P.J. Abdul Kalam’s speech)
paragraph = '''
I have three visions for India. In 3000 years of our history, people from all over the world have come
and invaded us, captured our lands, conquered our minds. From Alexander onwards, the Greeks, the Turks,
the Moguls, the Portuguese, the British, the French, the Dutch, all of them came and looted us...
'''

# Convert to sentences
sentences = sent_tokenize(paragraph)
print("Total Sentences:", len(sentences))
print(sentences)
````

````python
# POS tagging after removing stopwords
filtered_sentences = []
stop_words = set(stopwords.words("english"))

for sent in sentences:
    words = word_tokenize(sent)
    filtered_words = [w for w in words if w.lower() not in stop_words and w.isalpha()]
    tagged = nltk.pos_tag(filtered_words)
    filtered_sentences.append(tagged)

# Display POS tagging result
for i, sent in enumerate(filtered_sentences):
    print(f"\nSentence {i+1} POS Tags:")
    print(sent)
````

**In one simple sentence:**  
This code removes stopwords and punctuation, then labels every remaining word as a noun, verb, adjective, or another type of word.

````python
# POS tagging on a custom sentence

sentence = "Taj Mahal is a beautiful monument"
words = word_tokenize(sentence)
tags = nltk.pos_tag(words)

print("POS Tagging Result:")
for word, tag in tags:
    print(f"{word:10s} --> {tag}")
````

## 🐛 Common Errors & Fixes

1. **Missing Tagger Resource:**
   ```
   LookupError: Resource 'averaged_perceptron_tagger' not found
   ```
   🔧 Fix: Run `nltk.download('averaged_perceptron_tagger')`

2. **Passing wrong input to `pos_tag`:**
   - ✅ Correct: `nltk.pos_tag(['Taj', 'Mahal'])`
   - ❌ Wrong: `nltk.pos_tag('Taj Mahal')` (string, not list)

3. **Removing meaningful words:**
   - Be careful with stopwords. Sometimes removing too much ruins context.

> Leave mistakes in while learning. You’ll remember better.
