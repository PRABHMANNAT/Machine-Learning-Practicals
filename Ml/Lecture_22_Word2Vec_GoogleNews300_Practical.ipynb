# Word2Vec Practical (Pre-trained Google News 300) — with Notes

This notebook walks through using **Gensim** to explore the pre-trained **Google News Word2Vec (300-dim)** model and demonstrates:
- Loading the model via `gensim.downloader`
- Inspecting vectors
- Finding similar words
- Computing word similarity
- Performing analogies (e.g., *king - man + woman ≈ queen*)

Along the way, you'll find **concise Markdown notes** summarizing the key ideas for exam/interview prep.

## 🔎 Background

- **Word2Vec** is a shallow neural-network technique (Google, 2013) that learns **dense vectors** for words.
- Pre-trained model used here: **`word2vec-google-news-300`**
  - ~100B tokens (Google News)
  - ~3M words/phrases
  - 300-dimensional vectors
  - ~1.6 GB (downloaded on first use)

## ⚙️ Setup

> If you're running locally, ensure you have sufficient RAM & disk space (~1.6 GB) for the model.

Install and import dependencies:

```python
# !pip install gensim
import gensim
from gensim.models import KeyedVectors
import gensim.downloader as api
```

## 📥 Load the Pre-trained Model

`gensim.downloader` will fetch the model if it's not cached. The object returned exposes a KeyedVectors interface.

```python
# This will download (~1.6 GB) on first run, then cache.
# Uncomment to run in an environment with internet access.
# wv = api.load("word2vec-google-news-300")

# For demonstration in restricted environments, we'll mock a tiny keyed vectors using gensim test data.
try:
    wv = api.load("word2vec-google-news-300")
except Exception as e:
    print("Could not load GoogleNews model here. Using a small fallback model for demo:", e)
    wv = api.load("glove-wiki-gigaword-50")  # smaller 50-d GloVe for examples
```

## 🧪 Inspect Vectors

You can retrieve a word's vector and check its shape (dimension).

```python
word = "king"
vec = wv[word]
vec[:10], vec.shape
```

## 🔍 Find Most Similar Words

`most_similar` uses cosine similarity to find nearest neighbors in the embedding space.

```python
wv.most_similar("cricket")[:10]
```

## 📐 Word-to-Word Similarity

`similarity(w1, w2)` returns cosine similarity in [-1, 1]. Values closer to 1 indicate stronger relatedness.

```python
wv.similarity("hockey", "sports")
```

## 🧮 Analogy: *king - man + woman ≈ ?*

Vector arithmetic often recovers semantic relations.

```python
wv.most_similar(positive=["king", "woman"], negative=["man"], topn=10)
```

## ✅ Summary Notes

- **Pre-trained embeddings** give you strong semantic vectors out of the box.
- Key APIs: `wv[word]`, `wv.most_similar()`, `wv.similarity()`, and analogies with `positive`/`negative`.
- Use pre-trained models when:
  - You need high-quality vectors fast.
  - Your dataset is small.
- Train your own Word2Vec (CBOW/Skip-Gram) with `gensim.models.Word2Vec` when:
  - You have domain-specific text (e.g., medical, legal).
  - You want custom dimensions/window/negative sampling settings.
